```python
# --- SWG Electromagnetism Module · Utilities --------------------------------
# Pure NumPy/Matplotlib helpers shared by downstream experiment cells.

import numpy as np
import matplotlib.pyplot as plt

EPS = 1e-12
rng = np.random.default_rng(7)

# --------- finite differences (periodic) ----------
def grad2(Z):
    gx = 0.5*(np.roll(Z, -1, 1) - np.roll(Z, 1, 1))
    gy = 0.5*(np.roll(Z, -1, 0) - np.roll(Z, 1, 0))
    return gx, gy

def lap2(Z):
    return (np.roll(Z,-1,1) + np.roll(Z,1,1) + np.roll(Z,-1,0) + np.roll(Z,1,0) - 4*Z)

def moving_mean2(Z, k=7):
    k = int(k)
    if k <= 1: return Z
    V = Z
    for _ in range(2):
        V = (np.roll(V,1,0)+V+np.roll(V,-1,0))/3.0
        V = (np.roll(V,1,1)+V+np.roll(V,-1,1))/3.0
    return V

def sat(u, alpha=0.20):  # saturating nonlinearity (keeps dynamics bounded)
    return u / (1.0 + alpha*(u**2))

def wrap_pi(phi):
    return (phi + np.pi) % (2*np.pi) - np.pi

# --------- B-like curl from phase ---------
def current_and_Bz(theta):
    Jx, Jy = grad2(theta)
    # Bz ≡ ∂x Jy − ∂y Jx
    Bz = np.roll(Jy, -1, 1) - Jy - (np.roll(Jx, -1, 0) - Jx)
    return (Jx, Jy), Bz

# --------- SCG invariants on 2D field (A, theta) ----------
def scg_invariants(A, theta, chi_smooth=11):
    # memory proxy χ = smooth envelope of A; curvature face κ = |Δχ|/|χ|
    chi = moving_mean2(A, k=chi_smooth).astype(np.float32)
    kappa = np.abs(lap2(chi)) / (np.abs(chi) + EPS)

    # coherence entropy (amplitude weights)
    w = A / (np.sum(A) + EPS)
    Hc = float(-np.sum(w*np.log(w + EPS)))

    # curvature variance
    Kvar = float(np.var(kappa))

    # energy proxy & flux
    E = A*A
    PhiE = float(np.mean((np.roll(E, -1, 1) - E) / (np.abs(E) + EPS)))

    # phase dislocation density (thresholded gradient jumps along x)
    dth = (np.roll(theta, -1, 1) - theta + np.pi)%(2*np.pi) - np.pi
    med = np.median(dth); mad = 1.4826*np.median(np.abs(dth - med)) + EPS
    rho_phi = float(np.mean(np.abs(dth - med) > (med + 2.0*mad)))

    # order parameter (global phase alignment)
    R = float(np.abs(np.mean(np.exp(1j*theta))))

    # geometry floor: local scale and robust floor (metric p10)
    L_local = 1.0 / np.sqrt(kappa + EPS)
    L_metric = float(np.percentile(L_local, 10))
    L_dyn = 1.0
    L_geom = float(np.sqrt(L_dyn**2 + L_metric**2))
    return dict(Hc=Hc, Kvar=Kvar, PhiE=PhiE, rho_phi=rho_phi, R=R,
                L_metric=L_metric, L_geom=L_geom, L_local=L_local, chi=chi)

# --------- compact panel for A, θ, Bz and metrics ----------
def scg_panel(A, theta, Bz, inv, title=None, skip=6):
    fig, axs = plt.subplots(2, 3, figsize=(12, 7))
    im0 = axs[0,0].imshow(A, cmap='magma'); axs[0,0].set_title("Amplitude A")
    fig.colorbar(im0, ax=axs[0,0], shrink=0.8)
    im1 = axs[0,1].imshow(theta, cmap='twilight'); axs[0,1].set_title("Phase θ")
    fig.colorbar(im1, ax=axs[0,1], shrink=0.8)
    vmin = np.percentile(Bz, 2); vmax = np.percentile(Bz, 98)
    im2 = axs[0,2].imshow(Bz, cmap='coolwarm', vmin=vmin, vmax=vmax)
    axs[0,2].set_title("“B-like” $B_z$"); fig.colorbar(im2, ax=axs[0,2], shrink=0.8)

    Jx, Jy = grad2(theta)
    axs[1,0].quiver(Jx[::skip,::skip], Jy[::skip,::skip], angles='xy', scale_units='xy', scale=1.0)
    axs[1,0].set_title("Current-like J = ∇θ")

    L_local = inv["L_local"]; L_metric = inv["L_metric"]; L_geom = inv["L_geom"]
    axs[1,1].hist(L_local.ravel(), bins=60, color="#555")
    axs[1,1].axvline(L_metric, color='orange', ls='--', label='L_metric (p10)')
    axs[1,1].axvline(L_geom, color='g', ls='--', label='L_geom')
    axs[1,1].set_title("Local scale L"); axs[1,1].legend()

    axs[1,2].axis('off')
    axs[1,2].text(0.0, 1.0,
        (f"H={inv['Hc']:.3f}\nσ²κ={inv['Kvar']:.3e}\nΦ_E≈{inv['PhiE']:.3e}\n"
         f"ρ_φ={inv['rho_phi']:.3f}\nR={inv['R']:.3f}\n"
         f"L_metric≈{L_metric:.3f}\nL_geom≈{L_geom:.3f}"),
        va='top', family='monospace', fontsize=11)
    if title: fig.suptitle(title, fontsize=13)
    plt.tight_layout(); plt.show()

# --------- robust vlims for images ----------
def robust_vlims(Z, lo=2, hi=98):
    vmin = np.percentile(Z, lo); vmax = np.percentile(Z, hi)
    if vmax <= vmin: vmax = vmin + 1e-6
    return vmin, vmax
```

```python
# --- SWG/SCG Electromagnetism · Utilities (append-only) ---
import numpy as np
import matplotlib.pyplot as plt

EPS = 1e-12
rng = np.random.default_rng(7)

# ---------- numerics ----------
def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1) - np.roll(Z, 1,1))
    gy = 0.5*(np.roll(Z,-1,0) - np.roll(Z, 1,0))
    return gx, gy

def lap2(Z):
    return (np.roll(Z,-1,1) + np.roll(Z,1,1) + np.roll(Z,-1,0) + np.roll(Z,1,0) - 4.0*Z)

def moving_mean2(Z, k=7):
    k = int(max(1, k))
    if k == 1: return Z
    v = Z.copy()
    for _ in range(2):
        v = (np.roll(v,1,0)+v+np.roll(v,-1,0))/3.0
        v = (np.roll(v,1,1)+v+np.roll(v,-1,1))/3.0
    return v

def sat(u, alpha=0.20):
    return u/(1.0 + alpha*(u**2))

def robust_vlims(Z, lo=2, hi=98):
    vmin = np.percentile(Z, lo); vmax = np.percentile(Z, hi)
    if vmax <= vmin:
        vmax = vmin + (np.abs(vmin) + 1.0)*1e-6
    return float(vmin), float(vmax)

# ---------- SCG faces & invariants ----------
def curlz_from_J(Jx, Jy):
    # Bz ≡ ∂x Jy - ∂y Jx  (out-of-plane curl)
    return (np.roll(Jy,-1,1) - Jy) - (np.roll(Jx,-1,0) - Jx)

def scg_invariants(A, theta, chi_smooth=11):
    # memory / curvature face κ from χ (curvature is diagnostic face of memory):contentReference[oaicite:2]{index=2}:contentReference[oaicite:3]{index=3}
    chi = moving_mean2(A, k=chi_smooth)
    kappa = np.abs(lap2(chi)) / (np.abs(chi) + EPS)

    # geometry floors (local scale & metric floor):contentReference[oaicite:4]{index=4}:contentReference[oaicite:5]{index=5}
    L_local  = 1.0/np.sqrt(kappa + EPS)
    L_metric = float(np.percentile(L_local, 10))  # robust "metric" floor (p10)
    L_dyn    = 1.0                                 # neutral dynamic floor for static snapshot
    L_geom   = float(np.sqrt(L_dyn**2 + L_metric**2))

    # weights & entropy (coherence entropy ℋ)
    w = A/(np.sum(A) + EPS)
    Hc = float(-np.sum(w*np.log(w + EPS)))

    # curvature variance
    Kvar = float(np.var(kappa))

    # energy proxy & coarse "flux" Φ_E along +x
    E = A*A
    PhiE = float(np.mean((np.roll(E,-1,1) - E)/(np.abs(E)+EPS)))

    # phase dislocation density ρ_φ (MAD threshold on wrapped gradients)
    dthx = (np.roll(theta,-1,1) - theta + np.pi)%(2*np.pi) - np.pi
    med  = np.median(dthx); mad = 1.4826*np.median(np.abs(dthx - med)) + EPS
    rho_phi = float(np.mean(np.abs(dthx - med) > (med + 2.0*mad)))

    # global phase order parameter R
    Rop = float(np.abs(np.mean(np.exp(1j*theta))))

    return dict(H=Hc, Kvar=Kvar, PhiE=PhiE, rho=rho_phi, R=Rop,
                L_metric=L_metric, L_geom=L_geom, L_local=L_local, kappa=kappa)

def circulation_around(theta, cy, cx, r_pix=18, num=512):
    # ∮∇θ·dl ≈ Σ wrap(θ_{i+1}-θ_i) along a discrete circle
    t = np.linspace(0, 2*np.pi, num, endpoint=False)
    yy = (cy + r_pix*np.sin(t)).astype(int)
    xx = (cx + r_pix*np.cos(t)).astype(int)
    thp = theta[yy, xx]
    d = (np.roll(thp,-1) - thp + np.pi)%(2*np.pi) - np.pi
    return float(np.sum(d))
```

```python
# --- SWG/SCG Electromagnetism · Exp 3: Faraday-like induction test (novel) ---
import numpy as np
import matplotlib.pyplot as plt

# Parameters
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cy, cx0 = H//2, W//2
m = +1
osc_amp = 10.0          # px, small back-and-forth motion
T = 64                  # timesteps
dt = 1.0                # unitless time step for derivatives
r_emf = 40              # radius of the loop 𝒞 for ∮E·dl
num_ring = 1024         # samples on loop

# Helper: discrete line integral of a vector field along a circle
def loop_integral(vecx, vecy, cx, cy, r_pix, num=1024):
    t = np.linspace(0, 2*np.pi, num, endpoint=False)
    yy = cy + r_pix*np.sin(t); xx = cx + r_pix*np.cos(t)
    # bilinear sample (fast nearest-neighbor for simplicity)
    iy = np.clip(np.round(yy).astype(int), 0, vecx.shape[0]-1)
    ix = np.clip(np.round(xx).astype(int), 0, vecx.shape[1]-1)
    Ex = vecx[iy, ix]; Ey = vecy[iy, ix]
    tx = -np.sin(t); ty =  np.cos(t)  # tangent unit vector t_hat
    Edl = Ex*tx + Ey*ty
    ds  = r_pix*(2*np.pi/num)
    return float(np.sum(Edl)*ds)

# Pre-allocate series
emf_series = []
dphi_series = []
phi_series  = []
err_series  = []

# For area integral, build a circular mask for Φ_B
Y,X = np.mgrid[0:H,0:W].astype(np.float32)
maskS = ((X-cx0)**2 + (Y-cy)**2) <= r_emf**2
area_px = float(np.sum(maskS))

# Time evolution: oscillate the vortex center
thetas = []
Jxs = []
Jys = []
Bzs = []

for t in range(T):
    cx = cx0 + osc_amp*np.sin(2*np.pi*t/T)
    Xs = (x - cx).astype(np.float32); Ys = (y - cy).astype(np.float32)
    ang = np.arctan2(Ys, Xs)
    theta = (m*ang).astype(np.float32)

    # small envelope to keep invariants finite (fixed in time; doesn't drive induction)
    R = np.hypot(Xs, Ys) + 1e-6
    A = 0.22*np.exp(-(R/24.0)**2).astype(np.float32)

    Jx,Jy = grad2(theta)
    Bz = curlz_from_J(Jx, Jy)

    thetas.append(theta); Jxs.append(Jx); Jys.append(Jy); Bzs.append(Bz)

# Convert to arrays for finite differences
thetas = np.stack(thetas, 0)
Jxs = np.stack(Jxs, 0); Jys = np.stack(Jys, 0)
Bzs = np.stack(Bzs, 0)

# E proxy from SWG: E = -∂t (∇θ)
Ex = - (np.roll(Jxs,-1,axis=0) - Jxs)/dt
Ey = - (np.roll(Jys,-1,axis=0) - Jys)/dt

# Flux Φ_B through S and its time derivative
Phi = np.array([float(np.sum(Bzs[k][maskS])) for k in range(T)], dtype=float)
dPhi = - (np.roll(Phi,-1) - Phi)/dt  # NOTE the minus sign to compare with +∮E·dl

# EMF proxy around loop 𝒞
EMF = np.array([loop_integral(Ex[k], Ey[k], cx0, cy, r_emf, num=num_ring)
                for k in range(T)], dtype=float)

# Alignment check: compare EMF ( ∮E·dl ) vs -dΦ/dt
# Drop the last frame (finite-diff wrap)
valid = slice(0, T-1)
emf_series = EMF[valid]; dphi_series = dPhi[valid]
phi_series  = Phi[valid]

# Metrics
num = emf_series.size
rmse = float(np.sqrt(np.mean((emf_series - dphi_series)**2)))
corr = float(np.corrcoef(emf_series, dphi_series)[0,1])

# Plot
fig,axs = plt.subplots(3,1, figsize=(10.2,7.8), sharex=True)
axs[0].plot(phi_series, lw=1.5); axs[0].set_ylabel("Φ_B (sum Bz)")
axs[0].set_title("Faraday-like test in SWG (moving vortex)")

axs[1].plot(dphi_series, label="-dΦ/dt", lw=1.5)
axs[1].plot(emf_series,  label="∮E·dl (SWG proxy)", lw=1.5, ls="--")
axs[1].set_ylabel("EMF / -dΦ/dt")
axs[1].legend()

axs[2].plot(emf_series - dphi_series, color='crimson', lw=1.2)
axs[2].axhline(0, color='#555', lw=0.8, ls=':')
axs[2].set_ylabel("error")
axs[2].set_xlabel("time (frames)")

plt.tight_layout(); plt.show()

print(f"[Faraday-like]  RMSE={rmse:.3e}   corr={corr:.4f}   frames={num}")
print(f"Loop radius r={r_emf} px, ring samples={num_ring}, osc_amp={osc_amp} px, T={T}")
```

```python
# --- SWG/SCG Electromagnetism · Exp 4: Induction lag law sweep --------------
# Goal: quantify causal retardation between ∮E·dl (emf_series) and -dΦ/dt (dphi_series)
# Model: EMF(t) ≈ ( K_tau * [-dΦ/dt] )(t - Δt), where K_tau is a 1st-order causal kernel
#        y[n] = α y[n-1] + (1-α) x[n],  α = exp(-dt/tau),  tau ≥ 0
# Sweep Δt (lag in frames) and tau; report best corr/RMSE and overlay.

import numpy as np
import matplotlib.pyplot as plt

# ---- guards / capture upstream series ----
assert 'emf_series' in globals() and 'dphi_series' in globals() and 'dt' in globals(), \
    "This cell expects `emf_series`, `dphi_series`, `dt` from the Faraday experiment."

emf = np.asarray(emf_series, float)
dphi = np.asarray(dphi_series, float)

def causal_filter(x, tau):
    """First-order causal kernel with time constant tau (seconds).
       tau=0 -> identity (no filtering)."""
    if tau <= 0:
        return x.copy()
    a = float(np.exp(-dt/tau))
    y = np.empty_like(x)
    y[0] = (1.0 - a)*x[0]
    for n in range(1, len(x)):
        y[n] = a*y[n-1] + (1.0 - a)*x[n]
    return y

def align_and_metrics(y, target, lag):
    """
    Shift y by 'lag' frames to compare with target.
      lag > 0  -> y delayed (compare y[k-lag] to target[k])
      lag < 0  -> y advanced
    Returns (corr, rmse, y_al, t_al)
    """
    L = int(lag)
    if L > 0:
        t_al = target[L:]
        y_al = y[:-L]
    elif L < 0:
        t_al = target[:L]
        y_al = y[-L:]
    else:
        y_al = y; t_al = target
    if len(y_al) < 4:
        return np.nan, np.nan, y_al, t_al
    rmse = float(np.sqrt(np.mean((y_al - t_al)**2)))
    c = float(np.corrcoef(y_al, t_al)[0,1])
    return c, rmse, y_al, t_al

# ---- sweep grid (coarse -> refine) ----
lags = np.arange(-24, 25, 1)                  # frames
taus = np.r_[0.0, np.geomspace(0.25*dt, 16*dt, 16)]  # seconds (include 0=identity)

best = None
records = []
for L in lags:
    for tau in taus:
        y = causal_filter(dphi, tau)          # apply kernel to -dΦ/dt
        c, r, y_al, t_al = align_and_metrics(y, emf, L)
        if not np.isfinite(c): continue
        key = (c, -r)                         # maximize corr, then minimize RMSE
        records.append((L, tau, c, r))
        if best is None or key > (best['corr'], -best['rmse']):
            best = dict(lag=L, tau=tau, corr=c, rmse=r, y_al=y_al, t_al=t_al)

# optional local refine around best
if best is not None:
    L0, tau0 = best['lag'], best['tau']
    lags_ref = np.arange(L0-4, L0+5, 1)
    taus_ref = np.r_[max(0.0, tau0/2), tau0, tau0*2]
    for L in lags_ref:
        for tau in taus_ref:
            y = causal_filter(dphi, tau)
            c, r, y_al, t_al = align_and_metrics(y, emf, L)
            if not np.isfinite(c): continue
            key = (c, -r)
            if key > (best['corr'], -best['rmse']):
                best = dict(lag=L, tau=tau, corr=c, rmse=r, y_al=y_al, t_al=t_al)

# ---- report + overlay ----
assert best is not None, "No valid alignment found."
Lbest, taubest = best['lag'], best['tau']
cbest, rbest = best['corr'], best['rmse']
y_best, t_best = best['y_al'], best['t_al']

fig, axs = plt.subplots(2,1, figsize=(10.2,5.4), sharex=False)

axs[0].plot(t_best, label="EMF (∮E·dl)", lw=1.6)
axs[0].plot(y_best, label=f"Filtered -dΦ/dt  (lag={Lbest} fr, τ={taubest:.3g}s)", lw=1.6, ls='--')
axs[0].set_title("Induction alignment with causal lag/low-pass")
axs[0].legend(loc='best')
axs[0].grid(alpha=0.2)

axs[1].plot(t_best - y_best, color='crimson', lw=1.2)
axs[1].axhline(0, color='#777', lw=0.8, ls=':')
axs[1].set_ylabel("error")
axs[1].set_xlabel("aligned frame index")
axs[1].grid(alpha=0.2)

plt.tight_layout(); plt.show()

print(f"[Induction lag law] best:  lag={Lbest} frames  τ={taubest:.3g} s   corr={cbest:.4f}   RMSE={rbest:.3e}")
print(f"Grid searched lags={lags[0]}..{lags[-1]} frames,  taus∈[{taus.min():.3g},{taus.max():.3g}] s (incl. 0)")
```

```python
# --- SWG/SCG Electromagnetism · Exp 5: induction-wave velocity vs radius ---
# Sweeps loop radius r_emf, repeats Faraday+lag alignment, extracts Δt_best ≈ propagation delay
# → estimates effective coherence-wave speed  c_eff = r / Δt_best

import numpy as np
import matplotlib.pyplot as plt

assert 'Bzs' in globals() and 'Jxs' in globals() and 'Jys' in globals() and 'dt' in globals(), \
    "This cell needs field frames (Bzs, Jxs, Jys, dt) from the Faraday experiment."

# ---------------------------------------------------------------------
def emf_flux_vs_r(r_pix, num_ring=512):
    """Compute Φ_B(t) and ∮E·dl(t) for a given loop radius (reuses existing field stacks)."""
    H, W = Bzs.shape[1:3]
    cy, cx0 = H//2, W//2
    Y,X = np.mgrid[0:H,0:W].astype(np.float32)
    maskS = ((X-cx0)**2 + (Y-cy)**2) <= r_pix**2
    area_px = float(np.sum(maskS))

    # EMF proxy from E = -∂t(∇θ)
    Ex = - (np.roll(Jxs,-1,axis=0) - Jxs)/dt
    Ey = - (np.roll(Jys,-1,axis=0) - Jys)/dt

    # Φ_B
    Phi = np.array([float(np.sum(Bzs[k][maskS])) for k in range(Bzs.shape[0])])
    dPhi = - (np.roll(Phi,-1) - Phi)/dt

    # line integral
    def loop_integral(vecx, vecy, cx, cy, r_pix, num=512):
        t = np.linspace(0, 2*np.pi, num, endpoint=False)
        yy = cy + r_pix*np.sin(t); xx = cx + r_pix*np.cos(t)
        iy = np.clip(np.round(yy).astype(int), 0, vecx.shape[0]-1)
        ix = np.clip(np.round(xx).astype(int), 0, vecx.shape[1]-1)
        Exl = vecx[:, iy, ix]; Eyl = vecy[:, iy, ix]
        tx = -np.sin(t); ty = np.cos(t)
        Edl = Exl*tx + Eyl*ty
        ds  = r_pix*(2*np.pi/num)
        return np.sum(Edl, axis=1)*ds

    EMF = loop_integral(Ex, Ey, cx0, cy, r_pix, num=num_ring)
    return np.asarray(Phi[:-1]), np.asarray(dPhi[:-1]), np.asarray(EMF[:-1])

# ---------------------------------------------------------------------
def best_delay(dphi, emf):
    """Find Δt_best maximizing corr(|shift|≤24) between emf and -dΦ/dt."""
    lags = np.arange(-24,25)
    best_c,best_L=0,0
    for L in lags:
        if L>0:
            c = np.corrcoef(emf[L:], dphi[:-L])[0,1]
        elif L<0:
            c = np.corrcoef(emf[:L], dphi[-L:])[0,1]
        else:
            c = np.corrcoef(emf,dphi)[0,1]
        if np.abs(c)>np.abs(best_c):
            best_c,best_L=c,L
    return best_L*dt,best_c  # Δt_best, corr

# ---------------------------------------------------------------------
radii = np.arange(20, 101, 20)
delays=[]; corrs=[]

for r in radii:
    Phi, dPhi, EMF = emf_flux_vs_r(r)
    Δt, c = best_delay(dPhi, EMF)
    delays.append(Δt); corrs.append(c)

radii=np.array(radii,float); delays=np.array(delays,float); corrs=np.array(corrs,float)
c_eff = radii/np.abs(delays)      # px per time unit
c_eff_mean = np.nanmean(c_eff[np.isfinite(c_eff)])

# ---------------------------------------------------------------------
fig,ax1=plt.subplots(figsize=(7,5))
ax1.plot(radii, np.abs(delays), 'o-', label='|Δt_best|')
ax1.set_xlabel("loop radius r (px)"); ax1.set_ylabel("delay Δt (frames)")
ax1.grid(alpha=0.3)
ax2=ax1.twinx()
ax2.plot(radii, c_eff, 's--', color='tab:orange', label='c_eff = r/|Δt|')
ax2.set_ylabel("c_eff (px/frame)")
ax1.legend(loc='upper left'); ax2.legend(loc='upper right')
plt.title("Effective induction-wave velocity vs radius")
plt.tight_layout(); plt.show()

print(f"[Induction velocity sweep] mean c_eff ≈ {c_eff_mean:.3f} px/frame")
for r,dtb,ce in zip(radii,delays,c_eff):
    print(f"  r={r:4.0f} px  Δt={dtb:6.2f}  c_eff={ce:8.3f}  corr={corrs[np.where(radii==r)][0]:+.3f}")
```

```python
# --- SWG/SCG Electromagnetism · Exp 6: Josephson junction (pairwise sin-link) ---
import numpy as np
import matplotlib.pyplot as plt

EPS = 1e-12

# ---------------- geometry ----------------
H=W=220
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cxL, cxR = int(W*0.30), int(W*0.70)
cy       = H//2
R_sc     = 56
link_w   = 12
link_Lx  = (cxL + R_sc - 4, cxR - R_sc + 4)

def masks():
    X = x; Y = y
    dL = np.hypot(X-cxL, Y-cy); dR = np.hypot(X-cxR, Y-cy)
    SC_L = (dL <= R_sc); SC_R = (dR <= R_sc)
    link = np.zeros_like(X, bool)
    x0,x1 = link_Lx
    link[(Y >= cy-link_w//2) & (Y <= cy+link_w//2) & (X >= x0) & (X <= x1)] = True
    return SC_L, SC_R, link

SC_L, SC_R, LINK = masks()
x0,x1 = link_Lx
xi = (x0 + x1)//2
band = (y[:,0] >= cy-(link_w//2)) & (y[:,0] <= cy+(link_w//2))
band_idx = np.where(band)[0]                     # y-indices in the neck

# ---------------- material ----------------
A_island = 0.34; A_link = 0.22
A0 = np.zeros_like(x, np.float32)
A0[SC_L | SC_R] = A_island
A0[LINK]        = A_link
A0 = moving_mean2(A0, k=9)
A0 += 0.003*(x - W/2).astype(np.float32)
chi0 = moving_mean2(A0, k=11).astype(np.float32)

# ---------------- helper ----------------
def div(Fx, Fy):
    return (np.roll(Fx,-1,1) - Fx) + (np.roll(Fy,-1,0) - Fy)

# ---------------- single DC solve ----------------
def solve_dc(delta, steps=260, dt=0.10,
             kA=0.40, betaA=0.08, lamA=0.18, chi_cpl=0.03, chi_decay=0.01,
             alphaA=0.02, alphaTh=0.02, kTheta_diff=0.06, k_stiff=0.02,
             Jc_link=4.5e-3):
    # fields and velocities
    A     = A0.copy(); A_t = np.zeros_like(A)
    chi   = chi0.copy()
    theta = np.zeros_like(x, np.float32)
    theta_t = np.zeros_like(theta)

    # DC boundary: reservoirs fixed every step
    left_val  = +0.5*delta
    right_val = -0.5*delta

    # small random seed in link so symmetry doesn't kill the torque
    theta[LINK] += 1e-3*np.random.default_rng(0).standard_normal(LINK.sum()).astype(np.float32)

    for _ in range(steps):
        # enforce Dirichlet boundaries
        theta[SC_L] = left_val
        theta[SC_R] = right_val

        # gradients & transport
        Jx, Jy = grad2(theta)
        A2 = A*A

        # conservative transport + tiny diffusion + stiffness
        Fx = A2*Jx; Fy = A2*Jy
        theta_tt = div(Fx, Fy)/(A2 + EPS) + kTheta_diff*lap2(theta) - k_stiff*np.sin(theta)

        # --- pairwise Josephson coupling across the interface (equal & opposite) ---
        # for each y across the neck, couple xi and xi+1
        dloc = theta[band_idx, xi+1] - theta[band_idx, xi]      # Δθ_RL at midline
        torque = Jc_link * np.sin(dloc).astype(np.float32)
        # accumulate torques
        theta_tt[band_idx, xi]   += +torque
        theta_tt[band_idx, xi+1] += -torque

        # phase damping
        theta_tt -= alphaTh*theta_t

        # integrate θ (second order)
        theta_t += dt*theta_tt
        theta    = (theta + dt*theta_t + np.pi)%(2*np.pi) - np.pi

        # amplitude equation (responds to |∇θ|²) + memory + amplitude damping
        gth2 = (Jx*Jx + Jy*Jy)
        A_tt = kA*lap2(A) - betaA*A*gth2 + chi_cpl*(chi - A) - lamA*(A**3) - alphaA*A_t
        A_t  += dt*A_tt
        A     = np.clip(A + dt*A_t, 0.0, None)
        chi  += dt*chi_decay*(A - chi)

        # keep coherence levels
        A[SC_L | SC_R] = np.maximum(A[SC_L | SC_R], A_island*0.95)
        A[LINK]        = np.minimum(A[LINK],        A_link*1.05)

    # current through neck (pairwise jump is numerically robust)
    dth_mid = theta[band_idx, xi+1] - theta[band_idx, xi]
    I_link  = float(np.mean((A2[band_idx, xi] + A2[band_idx, xi+1])*0.5 * dth_mid))
    return I_link, theta, A

# ---------------- DC sweep ----------------
deltas = np.linspace(-np.pi, np.pi, 25)
Ivals  = [solve_dc(d)[0] for d in deltas]
Ivals  = np.asarray(Ivals)

# fit I(Δ) ≈ Ic sin(Δ + φ0)
def fit_sin(x, y):
    S = np.vstack([np.sin(x), np.cos(x)]).T
    a,b = np.linalg.lstsq(S, y, rcond=None)[0]
    Ic   = np.hypot(a,b); phi0 = np.arctan2(b, a)
    yhat = Ic*np.sin(x + phi0)
    R2   = 1 - np.sum((y-yhat)**2)/np.sum((y-y.mean())**2 + 1e-12)
    return Ic, phi0, yhat, R2

Ic, phi0, yhat, R2 = fit_sin(deltas, Ivals)

fig,ax=plt.subplots(1,1, figsize=(6.8,4.2))
ax.plot(deltas, Ivals, 'o', label='measured I(Δ)')
ax.plot(deltas, yhat, '-', label=f'fit: Ic={Ic:.3e}, φ0={phi0:+.2f}, R²={R2:.3f}')
ax.set_xlabel('Δθ (rad)'); ax.set_ylabel('link current density ⟨A²·Δθ_mid⟩')
ax.set_title('DC Josephson relation in SWG (pairwise sin-link)')
ax.legend(); ax.grid(alpha=0.25)
plt.tight_layout(); plt.show()
print(f"[DC Josephson] Ic={Ic:.6e}  phi0={phi0:+.3f}  R^2={R2:.4f}")

# ---------------- AC (Δ(t)=Ω t) ----------------
Ω = 0.18
T = 240
I_t = []
# warm start from Δ=0
_ = solve_dc(0.0, steps=200)
for k in range(T):
    Δk = (Ω*k + np.pi)%(2*np.pi) - np.pi
    I_k, _, _ = solve_dc(Δk, steps=90)
    I_t.append(I_k)
I_t = np.asarray(I_t)

# spectrum
X = np.fft.rfft(I_t - I_t.mean())
freqs = np.fft.rfftfreq(len(I_t), d=1.0)
kpeak = int(np.argmax(np.abs(X[1:]))+1)
f_meas = freqs[kpeak]; Ω_meas = 2*np.pi*f_meas

fig,axs=plt.subplots(2,1, figsize=(7.4,5.8))
axs[0].plot(I_t, lw=1.4)
axs[0].set_title('AC Josephson oscillation (phase ramp)')
axs[0].set_ylabel('⟨A²·Δθ_mid⟩'); axs[0].grid(alpha=0.25)
axs[1].plot(freqs, np.abs(X), lw=1.2); axs[1].set_xlim(0,0.5)
axs[1].set_xlabel('frequency (cycles/frame)'); axs[1].set_ylabel('|FFT|'); axs[1].grid(alpha=0.25)
plt.tight_layout(); plt.show()
print(f"[AC Josephson] drive Ω={Ω:.3f}, measured Ω_peak≈{Ω_meas:.3f} (f≈{f_meas:.3f})")
```

```python
# --- SWG/SCG θ-wave (1D) — Shapiro steps with AC+DC current drive at the link ---
import numpy as np, matplotlib.pyplot as plt

# ---------------- 1D grid and barrier ----------------
Nx   = 400
dx   = 1.0
dt   = 0.1
alpha= 0.02                  # damping
c_eff= 1.0
mid  = Nx//2
w    = 5                     # barrier half-width (link thickness)
steps= 4500
transient = 1500            # frames to discard before averaging

# spatial wave-speed profile (weaker inside barrier)
c2 = np.ones(Nx, float)*c_eff**2
c2[mid-w:mid+w] *= 0.35

# link mask (drive region)
D = np.zeros(Nx, float)
D[mid-w:mid+w] = 1.0

# helpers
def lap1(u):
    return (np.roll(u,-1) - 2*u + np.roll(u,1)) / (dx*dx)

def run_shapiro(I_dc, I_ac, Omega, steps=steps, dt=dt):
    """Integrate θ̈ = c^2 θxx − α θ̇ + (I_dc + I_ac sin Ωt)*D .
       Returns time series for link current J and 'voltage' V across the midline."""
    th   = np.zeros(Nx, float)
    thdot= np.zeros(Nx, float)
    J_hist  = []    # current proxy at link: J ≈ ∂x θ at mid
    V_hist  = []    # voltage proxy: Δ θ̇ across the two sides of the barrier

    for t in range(steps):
        drive = (I_dc + I_ac*np.sin(Omega*t*dt)) * D
        thdd  = c2*lap1(th) - alpha*thdot + drive
        thdot += dt*thdd
        th    += dt*thdot

        # boundaries: left Dirichlet, right Neumann
        th[0]  = 0.0
        th[-1] = th[-2]

        # observables at the link
        J_link = (th[mid+1] - th[mid-1])/(2*dx)
        V_link = thdot[mid+1] - thdot[mid-1]
        J_hist.append(J_link); V_hist.append(V_link)

    return np.asarray(J_hist), np.asarray(V_hist)

# ---------------- sweep I_dc to measure I–V and reveal steps ----------------
Omega = 0.18                  # RF drive (rad/step)
I_ac  = 2.0e-3                # AC amplitude
I_dc_vals = np.linspace(-6e-3, 6e-3, 41)

V_avg = []   # <V> across link (avg after transient)
I_avg = []   # <J> through link (avg after transient)

for Idc in I_dc_vals:
    J, V = run_shapiro(Idc, I_ac, Omega)
    J_tail = J[transient:]
    V_tail = V[transient:]
    I_avg.append(float(np.mean(J_tail)))
    V_avg.append(float(np.mean(V_tail)))

V_avg = np.asarray(V_avg); I_avg = np.asarray(I_avg)

# expected step locations for AC drive: <V> ≈ n*Omega
n_max = 5
step_lines = [n*Omega for n in range(-n_max, n_max+1)]

# ---------------- plots ----------------
fig,axs = plt.subplots(1,2, figsize=(11,4.6))

# (a) I–V curve (dc sweep)
axs[0].plot(I_dc_vals, V_avg, 'o-', lw=1.2)
for s in step_lines:
    axs[0].axhline(s, color='#888', lw=0.8, ls=':')
axs[0].set_title("Shapiro steps: ⟨V⟩ vs I_dc (AC+r.f. at the link)")
axs[0].set_xlabel("I_dc (drive amplitude at link)")
axs[0].set_ylabel("⟨V⟩ ≡ ⟨θ̇_R − θ̇_L⟩")
axs[0].grid(alpha=0.3)

# (b) I_dc vs ⟨I⟩ (to visualize current plateaus if present)
axs[1].plot(I_dc_vals, I_avg, 's-', lw=1.2)
axs[1].set_title("⟨I⟩ vs I_dc (mean current through link)")
axs[1].set_xlabel("I_dc")
axs[1].set_ylabel("⟨I⟩ ≡ ⟨∂x θ⟩ at link")
axs[1].grid(alpha=0.3)

plt.tight_layout(); plt.show()

# quick textual report: step detection by nearest multiple of Ω
# classify each V_avg by n ≈ round(V_avg/Ω)
n_est = np.rint(V_avg / (Omega + 1e-12)).astype(int)
unique_steps, counts = np.unique(n_est, return_counts=True)
print("Detected step indices n (⟨V⟩ ≈ n·Ω):")
for n,c in zip(unique_steps, counts):
    print(f"  n={n:+d}: {c} points")

# sanity check: show one time trace at a bias on a plateau and one off it
idx_plateau = np.argmin(np.abs(V_avg - Omega*1))      # closest to first step (n=+1)
idx_off      = np.argmin(np.abs(I_dc_vals - 0.0))     # near Idc ~ 0
Jp,Vp = run_shapiro(I_dc_vals[idx_plateau], I_ac, Omega)
Jo,Vo = run_shapiro(I_dc_vals[idx_off], I_ac, Omega)

fig,axs = plt.subplots(2,1, figsize=(10.2,5.4), sharex=True)
axs[0].plot(Vp[:1500], lw=1.0, label=f"plateau n≈+1, I_dc={I_dc_vals[idx_plateau]:.2e}")
axs[0].plot(Vo[:1500], lw=1.0, ls='--', label=f"off-plateau, I_dc={I_dc_vals[idx_off]:.2e}")
axs[0].set_ylabel("V(t)"); axs[0].legend(); axs[0].grid(alpha=0.3)
# spectra
Xp = np.fft.rfft(Vp - Vp.mean()); Xo = np.fft.rfft(Vo - Vo.mean())
f  = np.fft.rfftfreq(Vp.size, d=dt)
axs[1].plot(f, np.abs(Xp), lw=1.0, label="plateau"); axs[1].plot(f, np.abs(Xo), lw=1.0, ls='--', label="off")
axs[1].set_xlim(0, 0.5/dt); axs[1].set_xlabel("frequency"); axs[1].set_ylabel("|FFT|"); axs[1].legend()
axs[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()

print(f"Ω = {Omega:.3f} rad/step  → expected step spacing in ⟨V⟩ = Ω.")
```

```python
# --- SWG/SCG θ-wave (1D) — coupled junctions and synchronization ---
import numpy as np, matplotlib.pyplot as plt

# ---------------- grid and parameters ----------------
Nx   = 600
dx   = 1.0
dt   = 0.1
alpha= 0.006
c_eff= 1.0
steps= 9000
mid1, mid2 = 200, 400     # positions of the two weak links
w = 5                     # half-width of each link region
link_mask1 = np.zeros(Nx, float); link_mask1[mid1-w:mid1+w] = 1.0
link_mask2 = np.zeros(Nx, float); link_mask2[mid2-w:mid2+w] = 1.0

# base wave-speed profile (lower inside both links)
c2 = np.ones(Nx, float)*c_eff**2
c2[mid1-w:mid1+w] *= 0.45
c2[mid2-w:mid2+w] *= 0.45

# drive parameters
I_ac1, I_ac2 = 1.0e-2, 1.0e-2    # RF drive amplitudes
Omega1, Omega2 = 0.180, 0.184    # slightly different frequencies
I_dc1, I_dc2 = 0.0, 0.0
coupling_gain = 1.0e-3           # linear coupling between junctions (χ-mediated)
transient = 3000

# initial fields
theta = np.zeros(Nx)
theta_dot = np.zeros_like(theta)

# barrier drive masks
drive1 = link_mask1.copy()
drive2 = link_mask2.copy()

# helper
def lap1(u):
    return (np.roll(u,-1) - 2*u + np.roll(u,1)) / (dx*dx)

# time evolution
J1_hist, J2_hist = [], []

for t in range(steps):
    # local drives on each junction
    I_drive = (I_dc1 + I_ac1*np.sin(Omega1*t*dt))*drive1 \
            + (I_dc2 + I_ac2*np.sin(Omega2*t*dt))*drive2

    # coupling through coherence field: θ̈ += κ(θ_other_link - θ_link)
    Δ12 = np.mean(theta[mid2-w:mid2+w]) - np.mean(theta[mid1-w:mid1+w])
    coupling_term = np.zeros_like(theta)
    coupling_term[mid1-w:mid1+w] =  coupling_gain * Δ12
    coupling_term[mid2-w:mid2+w] = -coupling_gain * Δ12

    # full θ̈
    theta_ddot = c2*lap1(theta) - alpha*theta_dot + I_drive + coupling_term

    # integrate
    theta_dot += dt*theta_ddot
    theta     += dt*theta_dot

    # boundary: left Dirichlet, right Neumann
    theta[0]  = 0.0
    theta[-1] = theta[-2]

    # record link currents (∂xθ) around midpoints
    J1 = (theta[mid1+1]-theta[mid1-1])/(2*dx)
    J2 = (theta[mid2+1]-theta[mid2-1])/(2*dx)
    J1_hist.append(J1); J2_hist.append(J2)

J1_hist, J2_hist = np.asarray(J1_hist), np.asarray(J2_hist)

# ---------------- analysis ----------------
# correlation of the two junction currents
corr = np.corrcoef(J1_hist[transient:], J2_hist[transient:])[0,1]

# FFTs
def spectrum(sig):
    X = np.fft.rfft(sig - np.mean(sig))
    freqs = np.fft.rfftfreq(sig.size, d=dt)
    return freqs, np.abs(X)

f1,A1 = spectrum(J1_hist[transient:])
f2,A2 = spectrum(J2_hist[transient:])

# ---------------- plots ----------------
fig,axs=plt.subplots(3,1, figsize=(9,7))
axs[0].plot(J1_hist[:1500], lw=1.1, label="J1 (Ω1)")
axs[0].plot(J2_hist[:1500], lw=1.1, ls='--', label="J2 (Ω2)")
axs[0].set_title(f"Coupled junction currents (first 1500 steps)")
axs[0].set_ylabel("J(link)"); axs[0].legend(); axs[0].grid(alpha=0.3)

axs[1].plot(f1, A1/A1.max(), label="|FFT| J1")
axs[1].plot(f2, A2/A2.max(), label="|FFT| J2", ls='--')
axs[1].set_xlim(0, 0.5/dt); axs[1].set_xlabel("frequency"); axs[1].set_ylabel("norm |FFT|")
axs[1].set_title("Spectra of both junctions"); axs[1].legend(); axs[1].grid(alpha=0.3)

axs[2].scatter(J1_hist[transient:], J2_hist[transient:], s=1, alpha=0.3)
axs[2].set_xlabel("J1"); axs[2].set_ylabel("J2")
axs[2].set_title(f"Phase-space correlation after transient (corr={corr:.3f})")
axs[2].grid(alpha=0.3)

plt.tight_layout(); plt.show()
print(f"Coupling_gain={coupling_gain:.2e}, corr(J1,J2)={corr:+.3f}")
```

```python
# --- SWG/SCG θ-wave (2D) — synchronization front from a driven seed ----------
import numpy as np, matplotlib.pyplot as plt

# ---------------- grid & parameters ----------------
H=W=160
dt = 0.08
T  = 2400
alpha = 0.010                  # damping (light)
c_eff = 1.00                   # uniform wave speed
c2 = np.ones((H,W), np.float32)*(c_eff**2)

# Drive: small seed in the top-left corner
seed_h, seed_w = 22, 22
DRIVE = np.zeros((H,W), np.float32)
DRIVE[0:seed_h, 0:seed_w] = 1.0

I_dc = 0.0
I_ac = 1.2e-2                  # RF amplitude — tune 8e-3 .. 2e-2
Omega = 0.18                   # rad/step

# Init fields
rng = np.random.default_rng(7)
theta    = (1e-3 * rng.standard_normal((H,W))).astype(np.float32)
theta_dt = np.zeros_like(theta)

# helpers
wrap = lambda a: ((a + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

def global_order(th):
    z = np.exp(1j*th)
    R = np.abs(np.mean(z))
    return float(R)

def local_order(th, k=7):
    c = np.cos(th).astype(np.float32)
    s = np.sin(th).astype(np.float32)
    c_s = moving_mean2(c, k=k)
    s_s = moving_mean2(s, k=k)
    Rloc = np.sqrt(c_s*c_s + s_s*s_s)
    return Rloc

def count_vortices(th):
    # count phase defects by plaquette circulation
    dtx = wrap(np.roll(th,-1,1) - th)
    dty = wrap(np.roll(th,-1,0) - th)
    circ = wrap(dtx + np.roll(dty,-1,1) - np.roll(dtx,-1,0) - dty)
    return int(np.sum(np.abs(circ) > np.pi)), circ

# storage (sampled every k steps)
k_sample = 10
R_hist, F_hist, D_hist, E_hist = [], [], [], []

# energy diagnostic (θ-sector only): ½ θ̇² + ½ c²|∇θ|²
def energy_theta(th, thdt):
    gx,gy = grad2(th)
    return float(0.5*np.mean(thdt*thdt) + 0.5*np.mean(c2*(gx*gx+gy*gy)))

# ---------------- integrate ----------------
for t in range(T):
    # transport + drive
    lap = lap2(theta)
    drive = (I_dc + I_ac*np.sin(Omega*t*dt)) * DRIVE
    theta_dd = c2*lap - alpha*theta_dt + drive

    # update
    theta_dt = theta_dt + dt*theta_dd
    theta    = wrap(theta + dt*theta_dt)

    if (t % k_sample) == 0:
        Rg = global_order(theta)
        Rloc = local_order(theta, k=9)
        frac = float(np.mean(Rloc >= 0.85))     # coherent-area fraction
        nd, _ = count_vortices(theta)
        R_hist.append(Rg); F_hist.append(frac); D_hist.append(nd)
        E_hist.append(energy_theta(theta, theta_dt))

# ---------------- front-speed estimate (50%→90%) ----------------
times = np.arange(0, T, k_sample)*dt
idx50 = int(np.argmax(np.array(F_hist) >= 0.50))
idx90 = int(np.argmax(np.array(F_hist) >= 0.90))
def radius_from_frac(f):
    # effective radius from coherent area fraction (disc approximation)
    A = f * H * W
    r = np.sqrt(max(A,1e-6)/np.pi)
    return r
if idx50>0 and idx90>idx50:
    r50, r90 = radius_from_frac(F_hist[idx50]), radius_from_frac(F_hist[idx90])
    v_front  = (r90 - r50) / ((times[idx90] - times[idx50]) + 1e-9)   # px per step
else:
    v_front = np.nan

# ---------------- snapshots ----------------
def im_phase(ax, th, title):
    im = ax.imshow(th, cmap='twilight', vmin=-np.pi, vmax=np.pi)
    ax.set_title(title); plt.colorbar(im, ax=ax, shrink=0.78)

# choose early & late frames nearest to ~10% and last sample
snap_a = max(0, int(0.1*len(R_hist)) - 1)
snap_b = len(R_hist) - 1

# ---------------- plots ----------------
fig = plt.figure(figsize=(12,6.8))
gs = fig.add_gridspec(2,3, height_ratios=[1,1.2])

# coherence metrics
ax0 = fig.add_subplot(gs[0,0])
ax0.plot(times, R_hist, label='global R'); ax0.plot(times, F_hist, label='coherent frac ≥0.85')
ax0.axhline(0.5, color='#aaa', ls=':'); ax0.axhline(0.9, color='#aaa', ls=':')
ax0.set_title("Coherence metrics"); ax0.set_xlabel("time"); ax0.legend(); ax0.grid(alpha=0.3)

# defects + energy
ax1 = fig.add_subplot(gs[0,1])
ax1.plot(times, D_hist, label='vortex count')
ax1.set_title("Topological defects"); ax1.set_xlabel("time"); ax1.legend(); ax1.grid(alpha=0.3)

ax2 = fig.add_subplot(gs[0,2])
ax2.plot(times, E_hist, label='E_θ'); ax2.set_title("θ-sector energy"); ax2.set_xlabel("time"); ax2.grid(alpha=0.3)

# snapshots
ax3 = fig.add_subplot(gs[1,0]); im_phase(ax3, np.array(theta), f"Phase θ (final)")
ax4 = fig.add_subplot(gs[1,1]); im_phase(ax4, np.array(theta),  f"")  # placeholder re-used below

# show early snapshot by replaying stored indices: resim quickly to snap times
# (fast: re-run small loop up to snap time)
theta_tmp = (1e-3 * rng.standard_normal((H,W))).astype(np.float32)
theta_dt_tmp = np.zeros_like(theta_tmp)
for ti in range((snap_a+1)*k_sample):
    lap = lap2(theta_tmp)
    drive = (I_dc + I_ac*np.sin(Omega*ti*dt)) * DRIVE
    theta_dd = c2*lap - alpha*theta_dt_tmp + drive
    theta_dt_tmp = theta_dt_tmp + dt*theta_dd
    theta_tmp    = wrap(theta_tmp + dt*theta_dt_tmp)
im_phase(ax4, theta_tmp, f"Phase θ (early ~t={times[snap_a]:.1f})")

# drive mask overlay (for reference)
ax5 = fig.add_subplot(gs[1,2])
im5 = ax5.imshow(DRIVE, cmap='gray_r', vmin=0, vmax=1)
ax5.set_title("Drive mask (seed zone)"); plt.colorbar(im5, ax=ax5, shrink=0.78)

plt.tight_layout(); plt.show()

print(f"Front speed estimate (50%→90% coherence): v ≈ {v_front:.3f} px/step")
print(f"Final: global R={R_hist[-1]:.3f}, coherent fraction={F_hist[-1]:.3f}, vortices={D_hist[-1]}")
```

```python
# --- SWG/SCG θ-wave (2D) — coherence phase diagram: I_ac × disorder -----------
import numpy as np, matplotlib.pyplot as plt

# base parameters
H=W=120
dt = 0.08
T  = 900
alpha = 0.015
c_eff = 1.0
Omega = 0.18
seed_h=seed_w=8
I_dc = 0.0

drive_levels = np.array([2e-3,4e-3,6e-3,8e-3,1.0e-2,1.4e-2])
disorder_levels = np.array([0.00,0.04,0.08,0.12,0.16,0.20])

rng = np.random.default_rng(42)

# --- helper functions --------------------------------------------------------
wrap=lambda a:((a+np.pi)%(2*np.pi)-np.pi).astype(np.float32)

def global_order(th):
    z=np.exp(1j*th); return float(np.abs(np.mean(z)))

def count_vortices(th):
    dtx=wrap(np.roll(th,-1,1)-th); dty=wrap(np.roll(th,-1,0)-th)
    circ=wrap(dtx+np.roll(dty,-1,1)-np.roll(dtx,-1,0)-dty)
    return int(np.sum(np.abs(circ)>np.pi))

def energy_theta(th,thdt,c2):
    gx,gy=grad2(th)
    return float(0.5*np.mean(thdt*thdt)+0.5*np.mean(c2*(gx*gx+gy*gy)))

def run_once(I_ac,σ_dis):
    c2=(c_eff**2)*(1.0+σ_dis*rng.standard_normal((H,W))).astype(np.float32)
    DRIVE=np.zeros((H,W),np.float32)
    DRIVE[0:seed_h,0:seed_w]=1.0
    DRIVE[-seed_h:,-seed_w:]=1.0
    theta=(0.5*rng.standard_normal((H,W))).astype(np.float32)
    theta_dt=np.zeros_like(theta)
    for t in range(T):
        lap=lap2(theta)
        drive=(I_dc+I_ac*np.sin(Omega*t*dt))*DRIVE
        theta_dd=c2*lap - alpha*theta_dt + drive
        theta_dt += dt*theta_dd
        theta     = wrap(theta + dt*theta_dt)
    R=global_order(theta)
    V=count_vortices(theta)
    E=energy_theta(theta,theta_dt,c2)
    return R,V,E

# --- parameter sweep ---------------------------------------------------------
Rmap=np.zeros((len(drive_levels),len(disorder_levels)))
Vmap=np.zeros_like(Rmap)
Emap=np.zeros_like(Rmap)

for i,Iac in enumerate(drive_levels):
    for j,σ in enumerate(disorder_levels):
        R,V,E=run_once(Iac,σ)
        Rmap[i,j],Vmap[i,j],Emap[i,j]=R,V,E
        print(f"I_ac={Iac:.3e}, σ_dis={σ:.2f} → R={R:.3f}, V={V:3d}, E={E:.3e}")

# --- visualization -----------------------------------------------------------
fig,axs=plt.subplots(1,3,figsize=(13,4.8))
im0=axs[0].imshow(Rmap,origin='lower',extent=[disorder_levels[0],disorder_levels[-1],
               drive_levels[0],drive_levels[-1]],aspect='auto',vmin=0,vmax=1,cmap='viridis')
axs[0].set_title("Global coherence R")
axs[0].set_xlabel("disorder σ_dis"); axs[0].set_ylabel("drive I_ac"); plt.colorbar(im0,ax=axs[0],shrink=0.8)

im1=axs[1].imshow(Vmap,origin='lower',extent=[disorder_levels[0],disorder_levels[-1],
               drive_levels[0],drive_levels[-1]],aspect='auto',cmap='magma')
axs[1].set_title("Vortex count"); axs[1].set_xlabel("σ_dis"); plt.colorbar(im1,ax=axs[1],shrink=0.8)

im2=axs[2].imshow(Emap,origin='lower',extent=[disorder_levels[0],disorder_levels[-1],
               drive_levels[0],drive_levels[-1]],aspect='auto',cmap='cividis')
axs[2].set_title("θ-sector energy"); axs[2].set_xlabel("σ_dis"); plt.colorbar(im2,ax=axs[2],shrink=0.8)

plt.tight_layout(); plt.show()

# identify regime boundaries heuristically
#  (R>0.98 => global order; 0.9<R≤0.98 => propagating; R≤0.9 => turbulent)
phase_regime=np.zeros_like(Rmap,int)
phase_regime[Rmap>0.98]=2
phase_regime[(Rmap>0.9)&(Rmap<=0.98)]=1
phase_regime[Rmap<=0.9]=0

fig,ax=plt.subplots(figsize=(6,5))
c=ax.imshow(phase_regime,origin='lower',extent=[disorder_levels[0],disorder_levels[-1],
            drive_levels[0],drive_levels[-1]],aspect='auto',cmap='coolwarm')
ax.set_xlabel("disorder σ_dis"); ax.set_ylabel("drive I_ac")
ax.set_title("Phase diagram regions: 2=ordered,1=front,0=defect/turbulent")
plt.colorbar(c,ax=ax,shrink=0.8); plt.tight_layout(); plt.show()
```

```python
# --- SWG/SCG Aharonov–Bohm analogue: ring interferometer ---------------------
import numpy as np, matplotlib.pyplot as plt

# ---------------- grid & ring geometry ---------------------------------------
H=W=220
dt = 0.06
steps_prop = 3600
alpha = 0.012
c_eff = 1.0
rng = np.random.default_rng(9)

y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X = x-cx; Y = y-cy
R = np.sqrt(X*X + Y*Y)
R_in, R_out = 58, 72            # annulus radii
R_core = 40                     # inner core (confined region)

# ring mask (annulus)
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# readout slit on the right side of the ring
slit_w = 6
SLIT = (RING) & (np.abs(X - R_out*np.cos(0)) <= slit_w) & (np.abs(Y) < 3*slit_w)

# wave speed: fast only on ring, very slow elsewhere (waveguide)
c2 = np.zeros((H,W), np.float32)
c2[RING] = (c_eff**2)
c2[~RING] = 0.02*(c_eff**2)

# ---------------- confined curl: "flux thread" in CORE -----------------------
# We inject a localized torque that maintains an azimuthal phase gradient ONLY in core.
def core_torque(theta, PhiB):
    # produce curl ∝ PhiB inside CORE, zero outside; equal & opposite on boundary to confine
    ang = np.arctan2(Y, X)
    target = PhiB * ang   # phase pattern that would carry circulation PhiB
    d = ((target - theta + np.pi)%(2*np.pi) - np.pi)
    tau = np.zeros_like(theta, np.float32)
    tau[CORE] = 0.002 * d[CORE]          # weak restoring to target inside core
    return tau

# ---------------- packet launch: split-beam around ring ----------------------
def launch_split(theta):
    # two small packets on opposite sides of the readout slit, with opposite k around the ring
    ang = np.arctan2(Y, X)
    # packet centers ~ ±90° from slit (slit at 0°)
    mask1 = RING & (np.abs(ang - (+np.pi/2)) < 0.15)
    mask2 = RING & (np.abs(ang - (-np.pi/2)) < 0.15)
    theta = theta.copy()
    theta[mask1] += 0.8
    theta[mask2] -= 0.8
    return theta

# ---------------- evolution ----------------------
def evolve_ring(PhiB, steps=steps_prop):
    theta = (1e-3*rng.standard_normal((H,W))).astype(np.float32)
    theta = launch_split(theta)
    theta_dt = np.zeros_like(theta)
    wrap = lambda a: ((a + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

    # measure complex field on the slit to read interference
    z_hist = []

    for t in range(steps):
        lap = (np.roll(theta,-1,1)+np.roll(theta,1,1)+np.roll(theta,-1,0)+np.roll(theta,1,0)-4*theta)
        # transport only along ring (c2 is near-zero elsewhere)
        theta_dd = c2*lap - alpha*theta_dt + core_torque(theta, PhiB)

        theta_dt += dt*theta_dd
        theta    = wrap(theta + dt*theta_dt)

        # readout: complex sum on slit
        ph = theta[SLIT]
        z = np.mean(np.exp(1j*ph)) if ph.size>0 else 0.0
        z_hist.append(z)

    z_hist = np.asarray(z_hist, np.complex64)
    # final interference phase (mean over last quarter)
    z_tail = z_hist[int(0.75*len(z_hist)):]
    z_mean = np.mean(z_tail) if z_tail.size>0 else 0.0
    phase = float(np.angle(z_mean))
    vis   = float(np.abs(z_mean))
    return phase, vis, z_hist

# ---------------- sweep enclosed flux & controls -----------------------------
PhiB_vals = np.linspace(-2*np.pi, 2*np.pi, 13)    # sweep ±2π
phases = []; visibilities = []

for PhiB in PhiB_vals:
    ph, vis, _ = evolve_ring(PhiB)
    phases.append(ph); visibilities.append(vis)

phases  = np.unwrap(np.array(phases))
visibilities = np.array(visibilities)

# baseline control (no core)
def evolve_no_core(steps=steps_prop):
    theta = (1e-3*rng.standard_normal((H,W))).astype(np.float32)
    theta = launch_split(theta)
    theta_dt = np.zeros_like(theta)
    wrap = lambda a: ((a + np.pi)%(2*np.pi) - np.pi).astype(np.float32)
    z_hist=[]
    for t in range(steps):
        lap = (np.roll(theta,-1,1)+np.roll(theta,1,1)+np.roll(theta,-1,0)+np.roll(theta,1,0)-4*theta)
        theta_dd = c2*lap - alpha*theta_dt
        theta_dt += dt*theta_dd
        theta    = wrap(theta + dt*theta_dt)
        ph = theta[SLIT]; z = np.mean(np.exp(1j*ph)) if ph.size>0 else 0.0
        z_hist.append(z)
    z_tail = np.asarray(z_hist, np.complex64)[int(0.75*steps):]
    return float(np.angle(np.mean(z_tail)))

phase_baseline = evolve_no_core()

# “zero enclosed flux but same local flow” control: two opposite-signed threads
def evolve_dual_threads(PhiB, steps=steps_prop):
    theta = (1e-3*rng.standard_normal((H,W))).astype(np.float32)
    theta = launch_split(theta)
    theta_dt = np.zeros_like(theta)
    wrap = lambda a: ((a + np.pi)%(2*np.pi) - np.pi).astype(np.float32)
    # split core into two small disks with ±PhiB so net flux zero
    ang = np.arctan2(Y, X)
    coreA = (R<=R_core) & (X<0)
    coreB = (R<=R_core) & (X>=0)
    z_hist=[]
    for t in range(steps):
        lap = (np.roll(theta,-1,1)+np.roll(theta,1,1)+np.roll(theta,-1,0)+np.roll(theta,1,0)-4*theta)
        tau = np.zeros_like(theta,np.float32)
        # apply opposite torques that cancel net flux
        targetA = (+PhiB)*ang; dA = ((targetA - theta + np.pi)%(2*np.pi) - np.pi)
        targetB = (-PhiB)*ang; dB = ((targetB - theta + np.pi)%(2*np.pi) - np.pi)
        tau[coreA] = 0.002*dA[coreA]; tau[coreB] = 0.002*dB[coreB]
        theta_dd = c2*lap - alpha*theta_dt + tau
        theta_dt += dt*theta_dd
        theta    = wrap(theta + dt*theta_dt)
        ph = theta[SLIT]; z = np.mean(np.exp(1j*ph)) if ph.size>0 else 0.0
        z_hist.append(z)
    z_tail = np.asarray(z_hist, np.complex64)[int(0.75*steps):]
    return float(np.angle(np.mean(z_tail)))

phase_dual = np.array([evolve_dual_threads(P) for P in PhiB_vals])

# ---------------- plots -----------------------------
fig,axs = plt.subplots(1,2, figsize=(10.5,4.0))
axs[0].plot(PhiB_vals, phases - phase_baseline, 'o-', label='confined core')
axs[0].plot(PhiB_vals, phase_dual - phase_dual[PhiB_vals.size//2], 's--', label='dual threads (net flux 0)')
axs[0].set_xlabel("enclosed flux Φ_B (rad-equivalent)")
axs[0].set_ylabel("interference phase shift Δφ")
axs[0].set_title("SWG AB analogue: Δφ(Φ_B)")
axs[0].legend(); axs[0].grid(alpha=0.3)

axs[1].plot(PhiB_vals, visibilities, 'o-')
axs[1].set_xlabel("Φ_B"); axs[1].set_ylabel("visibility |⟨e^{iφ}⟩| at slit")
axs[1].set_title("Interference visibility vs enclosed flux")
axs[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
# --- SWG/SCG AB-II — Flux quantization & persistent current on a ring --------
# Phase-only reduction consistent with SWG canonical transport (θ̈ − α θ̇ + ...).
# Append-only; pure NumPy + Matplotlib.

import numpy as np
import matplotlib.pyplot as plt

# ---------------- geometry: ring (annulus) ----------------
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2,H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)

R_in, R_out = 58, 72          # ring thickness ~14 px (adjustable)
R_core       = 40             # inner core (excluded)
R_pad        = 10             # guard band to keep dynamics off the bulk

RING  = (R>=R_in) & (R<=R_out)
CORE  = (R<=R_core)
MASK  = RING.copy()           # dynamics primarily on the ring
FREE  = RING | CORE           # allowed support for soft torque where needed

# Tangent unit vectors on ring for J‖ estimate
ang = np.arctan2(Y, X)
tx, ty = -np.sin(ang), np.cos(ang)        # θ increases CCW; tangent points CCW

# Sample a 1-px wide median-radius contour for circulation/J readout
r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

# ---------------- integrator params ----------------
dt    = 0.10           # time step
T     = 3500           # total steps; enough to relax
alpha = 0.030          # damping (weak → persistent current remains)
kappa = 0.80           # phase diffusion on ring
lambda_s = 0.015       # soft local “Josephson stiffness” (stabilizes windings)
pin_core  = 0.004      # very weak core pin (keeps the thread centered)
guard     = 0.003      # suppress residual dynamics off the ring

def lap(Z):
    return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)

def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def wrap_pi(th):
    return ((th + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

def circulation(theta, path_mask, n_samp=2048):
    # Sample theta around a median-radius circle and sum wrapped differences.
    # Build parametric circle samples and bilinear interpolate theta.
    ys, xs = np.where(path_mask)
    if xs.size == 0:
        return 0.0
    r0 = np.mean(R[path_mask])
    t = np.linspace(0, 2*np.pi, n_samp, endpoint=False)
    xs_f = cx + r0*np.cos(t); ys_f = cy + r0*np.sin(t)
    # bilinear
    xi, yi = np.floor(xs_f).astype(int), np.floor(ys_f).astype(int)
    dx, dy = xs_f - xi, ys_f - yi
    xi1 = (xi+1).clip(0,W-1); yi1 = (yi+1).clip(0,H-1)
    th00 = theta[yi, xi]; th10 = theta[yi, xi1]; th01 = theta[yi1, xi]; th11 = theta[yi1, xi1]
    ths = (1-dx)*(1-dy)*th00 + dx*(1-dy)*th10 + (1-dx)*dy*th01 + dx*dy*th11
    d = (np.roll(ths,-1) - ths + np.pi)%(2*np.pi) - np.pi
    return float(np.sum(d))

def ring_current(theta, path_mask):
    gx, gy = grad(theta)
    # project gradient onto tangent along the ring sample
    Jpar = (gx*tx + gy*ty)[path_mask]
    return float(np.mean(Jpar)) if Jpar.size else 0.0

def energy(theta):
    # Minimal phase energy: 0.5*kappa*|∇θ|^2 + 0.5*lambda_s*(1-cosθ) on ring
    gx, gy = grad(theta)
    E_grad = 0.5*kappa*np.mean((gx**2 + gy**2)[RING])
    E_loc  = 0.5*lambda_s*np.mean((1.0 - np.cos(theta))[RING])
    return float(E_grad + E_loc)

# ---------------- experiment sweep over windings n ----------------
Ns = [0,1,2,3,4]   # try additional n if desired
records = []

for n in Ns:
    # --- initialize phase and its velocity
    theta = wrap_pi(n * ang)       # target winding
    thdot = np.zeros_like(theta)

    # small random dither for symmetry breaking (optional)
    rng = np.random.default_rng(n+123)
    theta += 1e-4*rng.standard_normal(theta.shape).astype(np.float32)

    E_trace, C_trace, J_trace = [], [], []

    for tstep in range(T):
        # Core torque: keep a centered “flux thread” (weak pin), zero outside core
        tau_core = np.zeros_like(theta)
        tau_core[CORE] = -pin_core * wrap_pi(theta[CORE])  # relax to 0 mod 2π

        # Phase transport (ring-only dominant):
        thdd = kappa*lap(theta)
        # Soft local stiffness (weak sine term) only on ring → stabilizes windings
        thdd[RING] += -lambda_s*np.sin(theta[RING])
        # Gentle guard off-ring to silence stray waves
        thdd[~(RING|CORE)] += -guard*theta[~(RING|CORE)]
        # Add core torque
        thdd += tau_core

        # Damped wave step: θ̈ − α θ̇ + … = 0
        thdot = (1.0 - alpha*dt)*thdot + dt*thdd
        theta = wrap_pi(theta + dt*thdot)

        # logs (sparse after initial transient)
        if tstep % 10 == 0:
            E_trace.append(energy(theta))
            C_trace.append(circulation(theta, PATH))
            J_trace.append(ring_current(theta, PATH))

    E_ss  = np.mean(E_trace[-25:]) if E_trace else energy(theta)
    circ  = np.mean(C_trace[-25:]) if C_trace else circulation(theta, PATH)
    J_ss  = np.mean(J_trace[-25:]) if J_trace else ring_current(theta, PATH)

    records.append(dict(n=n, E=E_ss, circ=circ, J=J_ss))

# ---------------- plots & report ----------------
Evals = [r["E"] for r in records]
circs = [r["circ"] for r in records]
Jvals = [r["J"] for r in records]

fig,axs = plt.subplots(1,3, figsize=(12,3.6))

# (a) Energy vs n — parabolic wells expected
axs[0].plot(Ns, Evals, "o-"); axs[0].set_xlabel("winding n"); axs[0].set_ylabel("steady energy")
axs[0].set_title("Energy vs winding — wells at integer n")

# (b) Circulation quantization
axs[1].plot(Ns, np.array(circs)/(2*np.pi), "o-")
axs[1].axhline(0, color="#888", ls=":")
axs[1].set_xlabel("winding n"); axs[1].set_ylabel("∮∇θ·dl / (2π)")
axs[1].set_title("Circulation ≈ n (quantized)")

# (c) Persistent current vs n
axs[2].plot(Ns, Jvals, "o-")
axs[2].axhline(0, color="#888", ls=":")
axs[2].set_xlabel("winding n"); axs[2].set_ylabel("<J‖> on ring")
axs[2].set_title("Persistent ring current")

plt.tight_layout(); plt.show()

# Pretty print quick summary
for r in records:
    print(f"n={r['n']}:  E≈{r['E']:.5f}   circ≈{r['circ']/(2*np.pi):.3f} quanta   <J‖>≈{r['J']:.5e}")
```

```python
# --- SWG/SCG AB-III — Phase slips & critical flux on a ring ------------------
# Hysteresis of persistent current vs enclosed flux with thermal noise.
# Append-only; no external deps besides NumPy/Matplotlib.

import numpy as np
import matplotlib.pyplot as plt

# ---------------- ring geometry (same shape as AB-II) ----------------
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)

R_in, R_out = 58, 72
R_core      = 40
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# median-radius path for readout (1 px band)
r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

# Tangent unit vectors (CCW)
tx, ty = -np.sin(ang), np.cos(ang)

# ---------------- numerics ----------------
dt     = 0.10
alpha  = 0.030        # damping
kappa  = 1.40         # curvature transport on ring
lambda_s = 0.050      # soft local stiffness (locks winding)
pin_core = 0.020      # enforces core “flux” target
guard  = 0.004        # off-ring quench

sigma_noise = 6e-3    # thermal noise amplitude (on θ̇)
steps_per_phi = 60    # integration steps per Φ increment
N_points = 160        # points per sweep branch

# helpers
def lap(Z):
    return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)

def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def wrap_pi(th): return ((th + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

def circulation(theta, path_mask, n_samp=2048):
    # bilinear sample around median radius
    ys, xs = np.where(path_mask)
    if xs.size == 0: return 0.0
    r0 = np.mean(R[path_mask])
    t = np.linspace(0, 2*np.pi, n_samp, endpoint=False)
    xs_f = cx + r0*np.cos(t); ys_f = cy + r0*np.sin(t)
    xi, yi = np.floor(xs_f).astype(int), np.floor(ys_f).astype(int)
    dx, dy = xs_f - xi, ys_f - yi
    xi1 = (xi+1).clip(0,W-1); yi1 = (yi+1).clip(0,H-1)
    th00 = theta[yi, xi]; th10 = theta[yi, xi1]; th01 = theta[yi1, xi]; th11 = theta[yi1, xi1]
    ths = (1-dx)*(1-dy)*th00 + dx*(1-dy)*th10 + (1-dx)*dy*th01 + dx*dy*th11
    d = (np.roll(ths,-1) - ths + np.pi)%(2*np.pi) - np.pi
    return float(np.sum(d))

def ring_current(theta, path_mask):
    gx, gy = grad(theta); Jpar = (gx*tx + gy*ty)[path_mask]
    return float(np.mean(Jpar)) if Jpar.size else 0.0

# ---------------- sweep engine ----------------
def run_sweep(phi_vals, theta_init=None, seed=0):
    rng = np.random.default_rng(seed)
    # start near n=0 unless a state is provided
    theta = wrap_pi(theta_init) if theta_init is not None else wrap_pi(0*ang)
    thdot = np.zeros_like(theta)

    J_list, n_list = [], []
    for phi in phi_vals:
        # target core phase (= Φ_B * angle) — this biases the “enclosed curl”
        theta_target_core = wrap_pi(phi * ang)

        for _ in range(steps_per_phi):
            # θ̈ = transport + locking + pin_core + guard − α θ̇ + noise
            thdd = np.zeros_like(theta)

            # ring transport + local stiffness
            thdd[RING] += kappa*lap(theta)[RING] - lambda_s*np.sin(theta[RING])

            # pin the core toward target (enclosed flux bias)
            err = wrap_pi(theta[CORE] - theta_target_core[CORE])
            thdd[CORE] += -pin_core * err

            # gentle off-ring guard
            thdd[~(RING|CORE)] += -guard*theta[~(RING|CORE)]

            # damped Verlet-like step + thermal noise on θ̇
            thdot = (1.0 - alpha*dt)*thdot + dt*thdd + np.sqrt(dt)*sigma_noise*rng.standard_normal(thdot.shape)
            theta = wrap_pi(theta + dt*thdot)

        # readouts after settling at this Φ
        circ = circulation(theta, PATH)
        n_est = circ/(2*np.pi)
        Jpar = ring_current(theta, PATH)
        J_list.append(Jpar); n_list.append(n_est)
    return np.array(J_list), np.array(n_list), theta

# forward and backward sweeps
phi_forward  = np.linspace(-4*np.pi, +4*np.pi, N_points)
phi_backward = np.linspace(+4*np.pi, -4*np.pi, N_points)

J_f, n_f, theta_f_end = run_sweep(phi_forward, theta_init=None, seed=11)
J_b, n_b, theta_b_end = run_sweep(phi_backward, theta_init=theta_f_end, seed=12)

# Detect phase slips as integer changes in round(n)
def detect_slips(phi_vals, n_vals):
    n_round = np.round(n_vals).astype(int)
    slips = []
    for k in range(1, len(n_round)):
        if n_round[k] != n_round[k-1]:
            slips.append((phi_vals[k], n_round[k-1], n_round[k]))
    return n_round, slips

nF_round, slips_F = detect_slips(phi_forward, n_f)
nB_round, slips_B = detect_slips(phi_backward, n_b)

# ---------------- plots ----------------
fig,axs = plt.subplots(1,3, figsize=(13,3.6))

# (1) Persistent current hysteresis J vs Φ_B
axs[0].plot(phi_forward,  J_f, label="forward", lw=1.6)
axs[0].plot(phi_backward, J_b, label="backward", lw=1.6)
axs[0].axhline(0, color="#888", ls=":"); axs[0].axvline(0, color="#888", ls=":")
axs[0].set_xlabel("enclosed flux Φ_B (rad)"); axs[0].set_ylabel("<J‖> on ring")
axs[0].set_title("Hysteresis of persistent current"); axs[0].legend()

# (2) Quantized winding n vs Φ_B (forward/back)
axs[1].step(phi_forward,  nF_round, where="post", label="forward")
axs[1].step(phi_backward, nB_round, where="post", label="backward")
axs[1].set_xlabel("Φ_B (rad)"); axs[1].set_ylabel("winding n (rounded)")
axs[1].set_title("Phase slips (quantized steps)"); axs[1].legend()

# (3) Unrounded n estimate (to see approach to slip)
axs[2].plot(phi_forward,  n_f, lw=1.2, label="forward raw")
axs[2].plot(phi_backward, n_b, lw=1.2, label="backward raw")
axs[2].set_xlabel("Φ_B (rad)"); axs[2].set_ylabel("n_est = ∮∇θ·dl/(2π)")
axs[2].set_title("Approach to instability (raw n)"); axs[2].legend()

plt.tight_layout(); plt.show()

# ---------------- text report of critical flux values ----------------
def pretty_slips(tag, slips):
    if not slips:
        print(f"{tag}: no slip detected")
    else:
        for phi, n0, n1 in slips:
            print(f"{tag}: Φ_c≈{phi:+.3f} rad,  n: {n0} → {n1}")

pretty_slips("forward", slips_F)
pretty_slips("backward", slips_B)
```

```python
# --- SWG/SCG AB-IV — Weak-link ring with AC flux drive (Josephson analogue) ---
# Append-only; NumPy + Matplotlib only.

import numpy as np, matplotlib.pyplot as plt

# ---------------- ring geometry ----------------
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)                    # [-pi, pi)

R_in, R_out = 58, 72
R_core      = 40
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# median-radius 1-px path for currents
r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

# Tangent on ring (CCW)
tx, ty = -np.sin(ang), np.cos(ang)

# ---------------- weak link (small angular sector near +x) ----------------
link_center = 0.0          # angle (rad), 0 = +x
link_halfw  = 8*np.pi/180  # half-width in radians (~8°)
def ang_wrap(a):           # shortest |a|
    return (a + np.pi)%(2*np.pi) - np.pi

WEAK = RING & (np.abs(ang_wrap(ang - link_center)) <= link_halfw)

# Two narrow arcs on either side of the link to read Δθ_link
side_w = 5*np.pi/180
SIDE_PLUS  = RING & (ang_wrap(ang - (link_center + side_w))  >= 0) & (np.abs(ang_wrap(ang - (link_center + side_w)))  <= 3*side_w)
SIDE_MINUS = RING & (ang_wrap(ang - (link_center - side_w))  <= 0) & (np.abs(ang_wrap(ang - (link_center - side_w)))  <= 3*side_w)

# ---------------- numerics & drive ----------------
dt      = 0.10
T_total = 12000                    # total steps (~ long enough for steady oscillations)
alpha   = 0.030                    # damping
kappa   = 1.40                     # curvature transport
lambda_strong = 0.050              # bulk locking
lambda_weak   = 0.010              # weak-link locking  (much smaller)
pin_core = 0.020                   # core “flux thread” pin
guard    = 0.004                   # off-ring quench
sigma_noise = 5e-3                 # small thermal drive

Phi_dc  = 0.0
Phi_ac  = 2.0                      # AC amplitude (rad)
Omega   = 0.010                    # drive frequency (rad/step) →  f = Omega/(2π dt)

# helpers
def lap(Z):
    return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)
def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def wrap_pi(th): return ((th + np.pi)%(2*np.pi) - np.pi).astype(np.float32)
def ring_current(theta, path_mask):
    gx, gy = grad(theta); Jpar = (gx*tx + gy*ty)[path_mask]
    return float(np.mean(Jpar)) if Jpar.size else 0.0
def delta_theta_link(theta):
    # phase drop across link = mean(θ on + side) − mean(θ on − side), wrapped
    thp = np.mean(theta[SIDE_PLUS])  if np.any(SIDE_PLUS)  else 0.0
    thm = np.mean(theta[SIDE_MINUS]) if np.any(SIDE_MINUS) else 0.0
    return float(wrap_pi(thp - thm))

# initialize
theta = wrap_pi(0.0*ang)                   # start near n=0
thdot = np.zeros_like(theta)

# spatially varying stiffness (weak link sector)
lambda_map = np.where(WEAK, lambda_weak, lambda_strong).astype(np.float32)

# recorders
J_series, dth_series, V_series = [], [], []    # V ≡ d/dt Δθ_link
Phi_series = []

# integration
rng = np.random.default_rng(2025)
for t in range(T_total):
    # AC+DC flux bias via core target
    Phi_t = Phi_dc + Phi_ac*np.sin(Omega*t)
    theta_target_core = wrap_pi(Phi_t * ang)

    # phase transport
    thdd = np.zeros_like(theta)
    # ring transport + local locking (with weak-link map)
    thdd[RING] += kappa*lap(theta)[RING] - lambda_map[RING]*np.sin(theta[RING])
    # pin the core to the drive
    err = wrap_pi(theta[CORE] - theta_target_core[CORE])
    thdd[CORE] += -pin_core * err
    # guard off-ring
    thdd[~(RING|CORE)] += -guard*theta[~(RING|CORE)]

    # update (damped + small thermal drive on θ̇)
    thdot = (1.0 - alpha*dt)*thdot + dt*thdd + np.sqrt(dt)*sigma_noise*rng.standard_normal(thdot.shape)
    theta = wrap_pi(theta + dt*thdot)

    # readouts every step
    J_series.append(ring_current(theta, PATH))
    dth = delta_theta_link(theta); dth_series.append(dth)
    Phi_series.append(Phi_t)
# voltage-like signal (finite diff)
V_series = np.gradient(np.array(dth_series), dt)

# ---------------- analyses ----------------
# steady part (drop initial transient)
burn = int(0.2*T_total)
t = np.arange(T_total)*dt
J = np.array(J_series)[burn:]
V = np.array(V_series)[burn:]
Phi = np.array(Phi_series)[burn:]
t_st = t[burn:]

# spectrum of V
N = len(V)
win = 0.5*(1 - np.cos(2*np.pi*np.arange(N)/max(N,1)))
Vw = V*win
Fw = np.fft.rfft(Vw)
freqs = np.fft.rfftfreq(N, d=dt)
P = (Fw*np.conj(Fw)).real

# expected drive frequency in Hz-equivalent steps: f_drive = Omega/(2π) (per step).
f_drive = Omega/(2*np.pi)

# ---------------- plots ----------------
fig,axs = plt.subplots(2,2, figsize=(12,7))

axs[0,0].plot(t_st, Phi, lw=1.0, label="Φ_B(t)")
axs[0,0].plot(t_st, dth_series[burn:], lw=1.0, label="Δθ_link(t)")
axs[0,0].set_title("Flux drive and link phase drop")
axs[0,0].set_xlabel("time"); axs[0,0].legend(); axs[0,0].grid(alpha=0.3)

axs[0,1].plot(t_st, J, lw=1.0)
axs[0,1].set_title("Persistent ring current J(t)")
axs[0,1].set_xlabel("time"); axs[0,1].grid(alpha=0.3)

axs[1,0].plot(freqs, P, lw=1.0)
axs[1,0].axvline(f_drive, color="crimson", ls="--", label="drive f")
axs[1,0].set_xlim(0, 3*f_drive)             # show first few harmonics vicinity
axs[1,0].set_xlabel("frequency (cycles/step)")
axs[1,0].set_ylabel("|FFT(V)|^2"); axs[1,0].set_title("Voltage-like spectrum |FFT(Δθ̇)|^2")
axs[1,0].legend(); axs[1,0].grid(alpha=0.3)

# phase portrait: V vs Δθ_link (Josephson-like loop)
axs[1,1].plot(dth_series[burn:], V, lw=0.8)
axs[1,1].set_xlabel("Δθ_link"); axs[1,1].set_ylabel("V ≡ d(Δθ)/dt")
axs[1,1].set_title("Josephson-like phase portrait"); axs[1,1].grid(alpha=0.3)

plt.tight_layout(); plt.show()

print(f"Drive f ≈ {f_drive:.5f} cycles/step.  Strong spectral peak near f? → Look at left-bottom panel.")
```

```python
# --- SWG/SCG AB-VI — Weak-link ring: frequency sweep & amplification ----------
# Single-tone flux drive, lock-in readout of V = d/dt(Δθ_link).
# Produces |H(f)|, phase(f), resonance peak and a quick Q estimate.

import numpy as np, matplotlib.pyplot as plt

# ---------------- common ring geometry (same as AB-IV/V) ----------------
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)

R_in, R_out = 58, 72
R_core      = 40
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# 1-px mid-radius readout path for ring current (not essential here)
r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

# tangent along ring (CCW)
tx, ty = -np.sin(ang), np.cos(ang)

# weak link sector near +x
def wrap_ang(a): return (a+np.pi)%(2*np.pi) - np.pi
link_center = 0.0
link_halfw  = 8*np.pi/180
WEAK = RING & (np.abs(wrap_ang(ang - link_center)) <= link_halfw)

# side arcs for Δθ_link readout
side_w = 5*np.pi/180
SIDE_PLUS  = RING & (wrap_ang(ang - (link_center + side_w))  >= 0) & (np.abs(wrap_ang(ang - (link_center + side_w)))  <= 3*side_w)
SIDE_MINUS = RING & (wrap_ang(ang - (link_center - side_w))  <= 0) & (np.abs(wrap_ang(ang - (link_center - side_w)))  <= 3*side_w)

# ---------------- dynamics parameters ----------------
dt      = 0.10
alpha   = 0.035      # damping
kappa   = 1.45       # curvature transport
lambda_strong = 0.050
lambda_weak   = 0.010
pin_core = 0.020
guard    = 0.004
sigma_noise = 5e-3   # small “thermal” background

# spatial stiffness map (weak link in sector)
lambda_map = np.where(WEAK, lambda_weak, lambda_strong).astype(np.float32)

def lap(Z):  return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)
def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def wrap_pi(th): return ((th + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

def delta_theta_link(theta):
    thp = np.mean(theta[SIDE_PLUS])  if np.any(SIDE_PLUS)  else 0.0
    thm = np.mean(theta[SIDE_MINUS]) if np.any(SIDE_MINUS) else 0.0
    return float(wrap_pi(thp - thm))

# ---------------- lock-in loop for one frequency ----------------
def run_one_frequency(Omega, Phi_dc=0.0, Phi_ac=1.5, T_total=14000, burn_frac=0.25, seed=777):
    rng = np.random.default_rng(seed)
    theta = wrap_pi(0.0*ang); thdot = np.zeros_like(theta)
    # integrate
    dth_trace = []
    for t in range(T_total):
        Phi_t = Phi_dc + Phi_ac*np.sin(Omega*t)
        theta_target_core = wrap_pi(Phi_t * ang)

        thdd = np.zeros_like(theta)
        # ring transport + local stiffness (with weak link)
        thdd[RING] += kappa*lap(theta)[RING] - lambda_map[RING]*np.sin(theta[RING])
        # pin the core to flux target
        err = wrap_pi(theta[CORE] - theta_target_core[CORE])
        thdd[CORE] += -pin_core * err
        # off-ring guard
        thdd[~(RING|CORE)] += -guard*theta[~(RING|CORE)]

        # damped+noisy update
        thdot = (1.0 - alpha*dt)*thdot + dt*thdd + np.sqrt(dt)*sigma_noise*rng.standard_normal(thdot.shape)
        theta = wrap_pi(theta + dt*thdot)

        dth_trace.append(delta_theta_link(theta))

    dth = np.array(dth_trace)
    V = np.gradient(dth, dt)  # voltage-like signal

    # lock-in at Omega after burn-in
    burn = int(burn_frac*T_total)
    t_arr = np.arange(burn, T_total)
    s = np.sin(Omega*t_arr); c = np.cos(Omega*t_arr)
    v_seg = V[burn:]
    # complex lock-in: H = <V * e^{-iΩt}> / (Φ_ac)  (units: response per rad)
    X = np.mean(v_seg * c); Y = np.mean(v_seg * s)
    H = (X - 1j*Y) / (0.5*Phi_ac)   # factor 0.5 because average sin/cos power
    mag = np.abs(H); phase = np.angle(H)

    # keep example time segments for plotting later (optional)
    return mag, phase, (t_arr*dt, V[burn:])

# ---------------- frequency sweep ----------------
# center sweep around the ring's typical θ-mode; tune bounds if needed
f_min, f_max = 0.0003, 0.0035   # cycles/step (≈ Ω/2π)
Nf = 28
freqs = np.linspace(f_min, f_max, Nf)
Omegas = 2*np.pi*freqs

gains, phases = [], []
examples = {}   # store one near-resonance and one off-resonance trace

mag_peak, idx_peak = -1.0, 0
for k, Om in enumerate(Omegas):
    mag, ph, ex = run_one_frequency(Om, Phi_dc=0.0, Phi_ac=1.8, T_total=14000, burn_frac=0.25, seed=800+k)
    gains.append(mag); phases.append(ph)
    if mag > mag_peak:
        mag_peak, idx_peak = mag, k
        examples["on"] = (freqs[k], ex)
    # also keep a clearly off-resonance point
examples["off"] = (freqs[0], run_one_frequency(Omegas[0], Phi_dc=0.0, Phi_ac=1.8, T_total=14000, burn_frac=0.25, seed=999)[2])

gains = np.array(gains); phases = np.unwrap(np.array(phases))

# quick Q estimate from -3 dB points around the peak
def estimate_Q(freqs, gains):
    if gains.size < 3: return np.nan, (np.nan, np.nan, np.nan)
    k0 = int(np.argmax(gains)); f0 = freqs[k0]; g0 = gains[k0]
    thr = g0/np.sqrt(2.0)
    # left
    kL = k0
    while kL > 0 and gains[kL] > thr: kL -= 1
    # right
    kR = k0
    while kR < len(gains)-1 and gains[kR] > thr: kR += 1
    if kL == k0 or kR == k0: return np.nan, (f0, np.nan, np.nan)
    B = freqs[kR] - freqs[kL]
    Q = f0/B if B>0 else np.nan
    return Q, (f0, freqs[kL], freqs[kR])

Q_est, (f0, fL, fR) = estimate_Q(freqs, gains)

# ---------------- plots ----------------
fig,axs = plt.subplots(2,2, figsize=(12,7))

# |H(f)| (gain)
axs[0,0].plot(freqs, gains, "o-", label="|H(f)|")
if np.isfinite(Q_est):
    axs[0,0].axvline(f0, color="crimson", ls="--", label=f"peak f0≈{f0:.5f}")
    axs[0,0].axvspan(fL, fR, color="crimson", alpha=0.12, label=f"-3 dB band (Q≈{Q_est:.1f})")
axs[0,0].set_xlabel("frequency (cycles/step)"); axs[0,0].set_ylabel("|H| (arb.)")
axs[0,0].set_title("Lock-in gain |H(f)|"); axs[0,0].legend()

# phase(f)
axs[0,1].plot(freqs, phases, "o-")
axs[0,1].set_xlabel("frequency (cycles/step)"); axs[0,1].set_ylabel("phase (rad)")
axs[0,1].set_title("Lock-in phase response")

# example time traces
for tag, (f_ex, (t_ex, V_ex)) in examples.items():
    if tag == "on":
        axs[1,0].plot(t_ex, V_ex, lw=0.9, label=f"on-res f≈{f_ex:.5f}")
    else:
        axs[1,0].plot(t_ex, V_ex, lw=0.7, label=f"off-res f≈{f_ex:.5f}", alpha=0.8)
axs[1,0].set_xlabel("time"); axs[1,0].set_ylabel("V(t)=d(Δθ)/dt")
axs[1,0].set_title("Voltage-like time-domain examples")
axs[1,0].legend(); axs[1,0].grid(alpha=0.3)

# phase portrait near resonance
if "on" in examples:
    f_on, (t_on, V_on) = examples["on"]
    # approximate Δθ by integrating V (visualization only)
    dth_on = np.cumsum(V_on)* (t_on[1]-t_on[0])
    axs[1,1].plot(dth_on - np.mean(dth_on), V_on, lw=0.8)
    axs[1,1].set_xlabel("Δθ_link (approx)"); axs[1,1].set_ylabel("V")
    axs[1,1].set_title("Phase portrait near resonance"); axs[1,1].grid(alpha=0.3)

plt.tight_layout(); plt.show()

if np.isfinite(Q_est):
    print(f"Resonance peak f0 ≈ {f0:.6f} cycles/step,  Q ≈ {Q_est:.2f}  (band ≈ [{fL:.6f}, {fR:.6f}])")
else:
    print("Could not estimate Q (peak too flat or sweep too short).")
```

```python
# --- SWG/SCG AB-VII — Injection locking: phase locking & coherence narrowing ---
# Strong main drive near f0, weak detuned drive. Watch two spectral lines fuse (locking).

import numpy as np, matplotlib.pyplot as plt

# ---------- ring geometry (same as AB-IV/V/VI) ----------
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)

R_in, R_out = 58, 72
R_core      = 40
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

tx, ty = -np.sin(ang), np.cos(ang)

def wrap_ang(a): return (a+np.pi)%(2*np.pi) - np.pi
link_center = 0.0
link_halfw  = 8*np.pi/180
WEAK = RING & (np.abs(wrap_ang(ang - link_center)) <= link_halfw)

side_w = 5*np.pi/180
SIDE_PLUS  = RING & (wrap_ang(ang - (link_center + side_w))  >= 0) & (np.abs(wrap_ang(ang - (link_center + side_w)))  <= 3*side_w)
SIDE_MINUS = RING & (wrap_ang(ang - (link_center - side_w))  <= 0) & (np.abs(wrap_ang(ang - (link_center - side_w)))  <= 3*side_w)

# ---------- dynamics ----------
dt      = 0.10
alpha   = 0.035
kappa   = 1.45
lambda_strong = 0.050
lambda_weak   = 0.010
pin_core = 0.020
guard    = 0.004
sigma_noise = 4e-3

lambda_map = np.where(WEAK, lambda_weak, lambda_strong).astype(np.float32)

def lap(Z):  return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)
def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def wrap_pi(th): return ((th + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

def delta_theta_link(theta):
    thp = np.mean(theta[SIDE_PLUS])  if np.any(SIDE_PLUS)  else 0.0
    thm = np.mean(theta[SIDE_MINUS]) if np.any(SIDE_MINUS) else 0.0
    return float(wrap_pi(thp - thm))

# ---------- drive: main + weak detuned ----------
# With AB-VI you saw f0 ~ 0.00113 cycles/step; center main tone near that.
f0_guess = 0.00113    # cycles/step (update from your AB-VI readout if needed)
Omega0 = 2*np.pi*f0_guess
detunes = [ +2.5e-4, +1.5e-4, +8.0e-5, +4.0e-5, +1.0e-5, +0.0 ]   # shrink Δf → locking
A_main, A_inj = 2.2, 0.6                                             # main stronger

# acquisition
T_block   = 9000        # steps per detune
burn_frac = 0.25
Nfft_max  = 4096        # cap for FFT windowing display

def run_block(Delta_f, seed=2027):
    rng = np.random.default_rng(seed)
    theta = wrap_pi(0.0*ang); thdot = np.zeros_like(theta)
    V = []
    for t in range(T_block):
        Phi_t = A_main*np.sin(Omega0*t) + A_inj*np.sin(2*np.pi*(f0_guess+Delta_f)*t)
        theta_target_core = wrap_pi(Phi_t * ang)

        thdd = np.zeros_like(theta)
        thdd[RING] += kappa*lap(theta)[RING] - lambda_map[RING]*np.sin(theta[RING])
        err = wrap_pi(theta[CORE] - theta_target_core[CORE])
        thdd[CORE] += -pin_core * err
        thdd[~(RING|CORE)] += -guard*theta[~(RING|CORE)]

        thdot = (1.0 - alpha*dt)*thdot + dt*thdd + np.sqrt(dt)*sigma_noise*rng.standard_normal(thdot.shape)
        theta = wrap_pi(theta + dt*thdot)

        # voltage-like signal
        if t>0:
            V.append( (delta_theta_link(theta) - prev_dth)/dt )
        prev_dth = delta_theta_link(theta)
    V = np.array(V, dtype=float)
    burn = int(burn_frac*V.size)
    V = V[burn:]
    # spectrum & peak finder
    N = min(len(V), Nfft_max)
    Vw = V[:N] * (0.5*(1 - np.cos(2*np.pi*np.arange(N)/N)))
    F  = np.fft.rfft(Vw); freqs = np.fft.rfftfreq(N, d=dt)
    P  = (F*np.conj(F)).real
    return freqs, P, V

# ---------- sweep over detunes ----------
results = []
for Δf in detunes:
    freqs, P, V = run_block(Δf)
    # detect the two largest peaks (main & injection or a single locked line)
    idxs = np.argsort(P)[-6:][::-1]
    peaks = [(freqs[i], P[i]) for i in idxs]
    results.append(dict(Delta_f=Δf, freqs=freqs, P=P, V=V, peaks=peaks))

# ---------- analyze locking & linewidth ----------
def main_inj_lines(peaks, f_main, f_inj, tol=2e-4):
    # pick peaks near f_main or f_inj; if one line dominates ~ locked
    near = lambda f0: [(f,p) for f,p in peaks if abs(f-f0)<tol]
    pm = near(f_main)
    pi = near(f_inj)
    return pm[:1], pi[:1]

f_main = f0_guess
out_rows = []
for r in results:
    Δf = r["Delta_f"]
    f_inj = f0_guess + Δf
    pm, pi = main_inj_lines(r["peaks"], f_main, f_inj)
    # estimate linewidth around the strongest peak by half-maximum width
    Pf = r["P"]; ff = r["freqs"]
    k0 = int(np.argmax(Pf)); P0 = Pf[k0]
    Lk = k0
    while Lk>0 and Pf[Lk] > 0.5*P0: Lk -= 1
    Rk = k0
    while Rk<len(Pf)-1 and Pf[Rk] > 0.5*P0: Rk += 1
    fL, fR = ff[max(Lk,0)], ff[min(Rk, len(ff)-1)]
    linewidth = fR - fL
    out_rows.append((Δf, pm, pi, linewidth))

# ---------- plots ----------
fig,axs = plt.subplots(2,2, figsize=(12,7))

# 1) show spectra for first and last detune
for j,idx in enumerate([0, len(results)-1]):
    r = results[idx]
    axs[0,0].plot(r["freqs"], r["P"], lw=0.9, label=f"Δf={r['Delta_f']:+.1e}")
axs[0,0].set_xlim(max(0, f0_guess-0.002), f0_guess+0.002); axs[0,0].legend()
axs[0,0].set_title("Voltage spectrum around f0"); axs[0,0].set_xlabel("f (cycles/step)"); axs[0,0].set_ylabel("|FFT|^2")

# 2) peak separation vs detune (locking → separation→0)
sep = []
for r in results:
    # find two largest distinct peaks in window
    ff, PP = r["freqs"], r["P"]
    mask = (ff>f0_guess-0.002) & (ff<f0_guess+0.002)
    ii = np.argsort(PP[mask])[-4:][::-1]
    fpk = ff[mask][ii]
    if len(fpk)>=2:
        sep.append( (r["Delta_f"], np.abs(fpk[0]-fpk[1])) )
    else:
        sep.append( (r["Delta_f"], 0.0) )
sep = np.array(sep, float)
axs[0,1].plot(sep[:,0], sep[:,1], "o-")
axs[0,1].set_xlabel("injection detune Δf (cycles/step)")
axs[0,1].set_ylabel("peak separation (cycles/step)")
axs[0,1].set_title("Peak separation → 0 (locking)")

# 3) linewidth vs detune (narrowing when locked)
lw = np.array([row[3] for row in out_rows], float)
axs[1,0].plot([row[0] for row in out_rows], lw, "o-")
axs[1,0].set_xlabel("Δf"); axs[1,0].set_ylabel("half-max linewidth")
axs[1,0].set_title("Coherence narrowing under locking")

# 4) time-domain snippets (final block)
tt = np.arange(results[-1]["V"].size)*dt
axs[1,1].plot(tt, results[-1]["V"], lw=0.8)
axs[1,1].set_xlabel("time"); axs[1,1].set_ylabel("V = d(Δθ)/dt")
axs[1,1].set_title("Locked regime: voltage time series"); axs[1,1].grid(alpha=0.3)

plt.tight_layout(); plt.show()

# text summary
print("Detune  | peak near f0? | peak near f_inj? | linewidth (cycles/step)")
for (Δf, pm, pi, lwid) in out_rows:
    tagm = f"{pm[0][0]:.5f}" if pm else "-"
    tagi = f"{pi[0][0]:.5f}" if pi else "-"
    print(f"{Δf:+.1e} | {tagm:<12} | {tagi:<13} | {lwid:.5e}")
```

```python
# --- SWG/SCG AB-VIII — Two weak-link rings coupled by a meta-χ bus ----------
# Ring A is driven; Ring B is undriven. χ-bus enables mutual synchronization.

import numpy as np, matplotlib.pyplot as plt

# ---------------- shared geometry helpers ----------------
H=W=240
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)

R_in, R_out = 58, 72
R_core      = 40
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# readout path (mid radius)
r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

def wrap_ang(a): return (a+np.pi)%(2*np.pi) - np.pi
link_center = 0.0
link_halfw  = 8*np.pi/180
WEAK = RING & (np.abs(wrap_ang(ang - link_center)) <= link_halfw)

# side arcs for link phase drop
side_w = 5*np.pi/180
SIDE_PLUS  = RING & (wrap_ang(ang - (link_center + side_w))  >= 0) & (np.abs(wrap_ang(ang - (link_center + side_w)))  <= 3*side_w)
SIDE_MINUS = RING & (wrap_ang(ang - (link_center - side_w))  <= 0) & (np.abs(wrap_ang(ang - (link_center - side_w)))  <= 3*side_w)

# tangential unit vectors (for current if needed)
tx, ty = -np.sin(ang), np.cos(ang)

# ---------------- dynamics (per ring) ----------------
dt      = 0.10
alpha   = 0.035
kappa   = 1.45
lambda_strong = 0.050
lambda_weak   = 0.010
pin_core = 0.020
guard    = 0.004
sigma_noise = 4e-3

lambda_map = np.where(WEAK, lambda_weak, lambda_strong).astype(np.float32)

def lap(Z):  return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)
def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def wrap_pi(th): return ((th + np.pi)%(2*np.pi) - np.pi).astype(np.float32)

def delta_theta_link(theta):
    thp = np.mean(theta[SIDE_PLUS])  if np.any(SIDE_PLUS)  else 0.0
    thm = np.mean(theta[SIDE_MINUS]) if np.any(SIDE_MINUS) else 0.0
    return float(wrap_pi(thp - thm))

# ---------------- χ-bus (meta memory) ----------------
# meta_chi slowly averages ring χ fields and feeds them back.
chi_cpl_in   = 0.06   # bus -> ring
chi_cpl_out  = 0.04   # ring -> bus
chi_leak     = 0.001  # slow decay of meta bus

# ---------------- drive: only Ring A ----------------
Phi_dc = 0.0
Phi_ac = 2.0
f0_guess = 0.00113                 # cycles/step (from AB-VI)
Omega = 2*np.pi*f0_guess

# ---------------- integration ----------------
T_total = 16000
burn    = int(0.20*T_total)

rngA = np.random.default_rng(101)
rngB = np.random.default_rng(202)

# Ring A
thA   = wrap_pi(0.0*ang); dthA = np.zeros_like(thA)
chiA  = np.zeros_like(thA)
# Ring B
thB   = wrap_pi(0.0*ang); dthB = np.zeros_like(thB)
chiB  = np.zeros_like(thB)
# meta bus
chiM  = np.zeros_like(thA)

# logs
dlinkA, dlinkB, V_A, V_B, PLI = [], [], [], [], []
Phi_series = []

prevA = delta_theta_link(thA)
prevB = delta_theta_link(thB)

for t in range(T_total):
    # drives: only A is driven
    PhiA = Phi_dc + Phi_ac*np.sin(Omega*t)
    PhiB = 0.0
    Phi_series.append(PhiA)

    # target cores
    tgtA = wrap_pi(PhiA * ang)
    tgtB = wrap_pi(PhiB * ang)

    # -------- ring updates (A & B) --------
    # A transport + weak-link stiffness
    thddA = np.zeros_like(thA)
    thddA[RING] += kappa*lap(thA)[RING] - lambda_map[RING]*np.sin(thA[RING])
    # pin cores
    errA = wrap_pi(thA[CORE] - tgtA[CORE]); thddA[CORE] += -pin_core*errA
    # off-ring guard
    thddA[~(RING|CORE)] += -guard*thA[~(RING|CORE)]
    # χ coupling (bus → ring), slow EMA for χ_A itself
    chiA = (1.0-chi_leak)*chiA + chi_cpl_in*(chiM - chiA)
    # damped + thermal
    dthA = (1.0 - alpha*dt)*dthA + dt*thddA + np.sqrt(dt)*sigma_noise*rngA.standard_normal(dthA.shape)
    thA  = wrap_pi(thA + dt*dthA)

    # B (no direct drive)
    thddB = np.zeros_like(thB)
    thddB[RING] += kappa*lap(thB)[RING] - lambda_map[RING]*np.sin(thB[RING])
    errB = wrap_pi(thB[CORE] - tgtB[CORE]); thddB[CORE] += -pin_core*errB
    thddB[~(RING|CORE)] += -guard*thB[~(RING|CORE)]
    chiB = (1.0-chi_leak)*chiB + chi_cpl_in*(chiM - chiB)
    dthB = (1.0 - alpha*dt)*dthB + dt*thddB + np.sqrt(dt)*sigma_noise*rngB.standard_normal(dthB.shape)
    thB  = wrap_pi(thB + dt*dthB)

    # bus write (ring → bus) every few steps
    if (t % 10) == 0:
        chiM = (1.0 - chi_leak)*chiM + chi_cpl_out*(0.5*(thA+thB) - chiM)

    # readouts
    dA = delta_theta_link(thA); dB = delta_theta_link(thB)
    vA = (dA - prevA)/dt; vB = (dB - prevB)/dt
    prevA, prevB = dA, dB

    dlinkA.append(dA); dlinkB.append(dB)
    V_A.append(vA);   V_B.append(vB)

    # phase-locking index (instant): PLI = |<e^{i(θA-θB)}>| over short window
    if t > 200:
        win = 200
        phi = np.array(dlinkA[-win:]) - np.array(dlinkB[-win:])
        PLI.append(np.abs(np.mean(np.exp(1j*phi))))
    else:
        PLI.append(0.0)

# numpy arrays & burn-in removal
t_arr = np.arange(T_total)*dt
Phi_series = np.array(Phi_series)[burn:]
dA = np.array(dlinkA)[burn:]; dB = np.array(dlinkB)[burn:]
VA = np.array(V_A)[burn:];    VB = np.array(V_B)[burn:]
PLI = np.array(PLI)[burn:]
t_st = t_arr[burn:]

# spectra near f0
def spectrum(sig, dt, nmax=4096):
    N = min(len(sig), nmax)
    w = 0.5*(1 - np.cos(2*np.pi*np.arange(N)/N))
    F = np.fft.rfft(sig[:N]*w)
    f = np.fft.rfftfreq(N, d=dt)
    P = (F*np.conj(F)).real
    return f, P

fA, PA = spectrum(VA, dt); fB, PB = spectrum(VB, dt)

# peak finder near drive
def peak_near(f, P, f0, bw=0.002):
    m = (f > f0-bw) & (f < f0+bw)
    k = np.argmax(P[m]); idx = np.where(m)[0][0] + k
    return f[idx], P[idx]

f0A, pA = peak_near(fA, PA, f0_guess); f0B, pB = peak_near(fB, PB, f0_guess)

# cross-correlation (normalized)
def norm_xcorr(a, b, maxlag=600):
    a = (a - np.mean(a)) / (np.std(a)+1e-12)
    b = (b - np.mean(b)) / (np.std(b)+1e-12)
    lags = np.arange(-maxlag, maxlag+1)
    xc = np.array([np.mean(a[max(0,lag):len(a)+min(0,lag)] * b[max(0,-lag):len(b)-max(0,lag)]) for lag in lags])
    return lags*dt, xc

lags, xc = norm_xcorr(VA, VB, maxlag=800)
xc_peak = xc[np.argmax(np.abs(xc))]

# ---------------- plots ----------------
fig,axs = plt.subplots(2,2, figsize=(12,7))

# A/B link phases & locking index
axs[0,0].plot(t_st, dA, lw=0.9, label="Δθ_link (A)")
axs[0,0].plot(t_st, dB, lw=0.9, label="Δθ_link (B)")
axs[0,0].set_title("Weak-link phase drops"); axs[0,0].legend(); axs[0,0].grid(alpha=0.3)

axs[0,1].plot(t_st, PLI, lw=1.0, color="crimson")
axs[0,1].set_title("Phase Locking Index (|⟨e^{i(Δθ_A−Δθ_B)}⟩|, window≈200)"); axs[0,1].set_ylim(0,1); axs[0,1].grid(alpha=0.3)

axs[1,0].plot(fA, PA/np.max(PA)+1e-12, lw=0.9, label="Ring A")
axs[1,0].plot(fB, PB/np.max(PB)+1e-12, lw=0.9, label="Ring B")
axs[1,0].axvline(f0_guess, color="#888", ls="--", label="drive f0")
axs[1,0].set_xlim(max(0,f0_guess-0.002), f0_guess+0.002)
axs[1,0].set_title("Voltage spectra (normalized)"); axs[1,0].legend(); axs[1,0].grid(alpha=0.3)

axs[1,1].plot(lags, xc, lw=0.9)
axs[1,1].set_title("Normalized cross-correlation VA ⟷ VB"); axs[1,1].set_xlabel("lag (time)"); axs[1,1].grid(alpha=0.3)

plt.tight_layout(); plt.show()

print(f"Peaks near f0:  A→ f≈{f0A:.6f},  B→ f≈{f0B:.6f} cycles/step")
print(f"Mean PLI (last 20%): {np.mean(PLI[int(0.8*len(PLI)):]):.3f}   |xcorr|_max≈{np.max(np.abs(xc)):.3f}")
```

```python
# --- SWG/SCG AB-IX — Multi-ring coherence network (reduced weak-link model) ---
# Each ring -> single DOF φ_k(t) ≡ Δθ_link^(k).
# 2nd-order phase transport with damping + local locking + network coupling.
# Sweep coupling K and measure global order R and pairwise PLI.

import numpy as np, matplotlib.pyplot as plt

# ---------------- reduced weak-link ring model ----------------
# φ̈_k + α φ̇_k + λ sin φ_k  =  ω_k  + K * Σ_j A_kj * sin(φ_j - φ_k) + drive_k(t)
# (Kuramoto-with-inertia + local curvature-locking λ sinφ_k)

def simulate_network(N=8, topology="ring", K=0.05, T=18000, dt=0.10,
                     alpha=0.035, lam=0.25,  # damping / local locking
                     f0=0.00113, spread=0.02,  # natural freq spread (±%)
                     drive_amp=2.0, drive_omega=2*np.pi*0.00113,  # drive 1st node
                     noise=2e-3, seed=7, burn_frac=0.3):
    rng = np.random.default_rng(seed)
    # natural frequencies (cycles/step -> rad/step)
    freqs = f0 * (1.0 + spread*rng.uniform(-1, 1, size=N))
    omega = 2*np.pi*freqs

    # adjacency
    A = np.zeros((N,N), float)
    if topology == "line":
        for k in range(N-1):
            A[k, k+1] = A[k+1, k] = 1.0
    elif topology == "ring":
        for k in range(N):
            A[k, (k+1)%N] = A[(k+1)%N, k] = 1.0
    elif topology == "all":
        A[:] = 1.0; np.fill_diagonal(A, 0.0)
    deg = np.maximum(A.sum(1), 1.0)
    # normalized Laplacian-style factor for stability
    def coupling_term(phi):
        # Σ_j A_kj sin(φ_j - φ_k) / deg_k
        S = np.zeros_like(phi)
        for k in range(N):
            S[k] = np.sum(A[k]*np.sin(phi - phi[k]))/deg[k]
        return S

    # state
    phi  = rng.uniform(-np.pi, np.pi, size=N)
    phid = np.zeros(N)
    # record after burn
    burn = int(burn_frac*T)
    Phi = []  # phases post-burn

    for t in range(T):
        # external drive only on node 0 (network entrainment test)
        drv = np.zeros(N)
        drv[0] = drive_amp*np.sin(drive_omega*t)

        # 2nd-order step
        phidd = - alpha*phid - lam*np.sin(phi) + omega + K*coupling_term(phi) + drv
        phid  = phid + dt*phidd + np.sqrt(dt)*noise*rng.standard_normal(N)
        phi   = (phi + dt*phid + np.pi)%(2*np.pi) - np.pi

        if t >= burn:
            Phi.append(phi.copy())

    Phi = np.array(Phi)               # (T_burn, N)
    # global order parameter R(t) and time-avg R
    R_t = np.abs(np.exp(1j*Phi).mean(axis=1))
    R   = float(R_t.mean())
    # pairwise phase-locking index (PLI) over the post-burn window
    def pli(a, b):
        return float(np.abs(np.exp(1j*(a-b)).mean()))
    PLI = np.zeros((N,N))
    for i in range(N):
        for j in range(N):
            PLI[i,j] = pli(Phi[:,i], Phi[:,j])
    return dict(R=R, R_t=R_t, PLI=PLI, Phi=Phi, freqs=freqs, A=A)

# ---------------- sweep coupling and summarize ----------------
def sweep_coupling(K_vals, N=8, topology="ring", seed=7):
    R_list = []; results_store = {}
    for K in K_vals:
        out = simulate_network(N=N, topology=topology, K=K, seed=seed)
        R_list.append(out["R"])
        results_store[K] = out
    return np.array(R_list), results_store

# settings
N = 8
topology = "ring"      # "line" | "ring" | "all"
K_vals = np.linspace(0.00, 0.14, 12)   # coupling sweep

R_vals, store = sweep_coupling(K_vals, N=N, topology=topology, seed=11)

# find a representative weak/strong point
K_lo = K_vals[1]
K_hi = K_vals[-2]
out_lo = store[K_lo]; out_hi = store[K_hi]

# ---------------- plots ----------------
fig,axs = plt.subplots(2,2, figsize=(12,7))

# (1) Global order vs coupling
axs[0,0].plot(K_vals, R_vals, "o-")
axs[0,0].axhline(0.8, color="#888", ls=":")
axs[0,0].set_xlabel("coupling K"); axs[0,0].set_ylabel("global order R = |⟨e^{iφ}⟩|")
axs[0,0].set_title(f"Synchronization curve (N={N}, topology={topology})")

# (2) PLI heatmap at strong coupling
im = axs[0,1].imshow(out_hi["PLI"], vmin=0, vmax=1, cmap="viridis")
axs[0,1].set_title(f"PLI matrix at K={K_hi:.3f}")
fig.colorbar(im, ax=axs[0,1], fraction=0.046)

# (3) Phase trajectories (subset) at weak coupling
Tvis = min(1500, out_lo["Phi"].shape[0])
for k in range(min(N,6)):
    axs[1,0].plot(np.unwrap(out_lo["Phi"][:Tvis,k]), lw=0.8, label=f"node {k}")
axs[1,0].set_title(f"Unwrapped phases at K={K_lo:.3f} (weak coupling)")
axs[1,0].set_xlabel("time step"); axs[1,0].legend(fontsize=8, ncol=2)

# (4) Phase trajectories at strong coupling (collapse to common line)
Tvis = min(1500, out_hi["Phi"].shape[0])
for k in range(min(N,6)):
    axs[1,1].plot(np.unwrap(out_hi["Phi"][:Tvis,k]), lw=0.8, label=f"node {k}")
axs[1,1].set_title(f"Unwrapped phases at K={K_hi:.3f} (locked)")
axs[1,1].set_xlabel("time step"); axs[1,1].legend(fontsize=8, ncol=2)

plt.tight_layout(); plt.show()

# quick textual summary
# threshold ≈ smallest K with R >= 0.8
thr_idx = np.argmax(R_vals >= 0.8) if np.any(R_vals >= 0.8) else None
if thr_idx is None:
    print("No full-network lock within K sweep. Increase K range or lower α.")
else:
    print(f"Estimated sync threshold K* ≈ {K_vals[thr_idx]:.3f}  (R≈{R_vals[thr_idx]:.2f}) for topology={topology}")
```

```python
# --- SWG/SCG AB-X — Geometry-floor perturbation & adaptive coherence ----------
# One node’s local "metric" (natural frequency) is periodically modulated.
# Observe global R(t) and node phase adaptation.

import numpy as np, matplotlib.pyplot as plt

# ----- parameters (based on AB-IX model) -----
N = 8
topology = "ring"
dt = 0.10
T = 20000
alpha = 0.05           # more damping
lam = 0.20             # local locking
K = 0.08               # coupling strength (above threshold)
f0 = 0.00113           # base frequency (cycles/step)
spread = 0.04          # intrinsic dispersion
noise = 2e-3
burn = int(0.2*T)
rng = np.random.default_rng(27)

# Geometry-floor perturbation on node 0
pert_amp = 0.05        # ±5% modulation of metric term
pert_freq = 2*np.pi*0.0003  # slow modulation frequency

# natural frequencies (in cycles/step -> rad/step)
freqs = f0*(1.0 + spread*rng.uniform(-1,1,N))
omega = 2*np.pi*freqs

# adjacency
A = np.zeros((N,N))
for k in range(N): A[k,(k+1)%N] = A[(k+1)%N,k] = 1.0
deg = np.maximum(A.sum(1),1.0)

def coupling_term(phi):
    S = np.zeros_like(phi)
    for k in range(N):
        S[k] = np.sum(A[k]*np.sin(phi - phi[k]))/deg[k]
    return S

# state
phi  = rng.uniform(-np.pi,np.pi,N)
phid = np.zeros(N)

R_hist, phi_hist = [], []

for t in range(T):
    # geometric perturbation on node 0 (modulates its natural freq)
    metric_mod = 1.0 + pert_amp*np.sin(pert_freq*t)
    omega_eff = omega.copy()
    omega_eff[0] *= metric_mod

    phidd = -alpha*phid - lam*np.sin(phi) + omega_eff + K*coupling_term(phi)
    phid  = phid + dt*phidd + np.sqrt(dt)*noise*rng.standard_normal(N)
    phi   = (phi + dt*phid + np.pi)%(2*np.pi) - np.pi

    if t >= burn:
        R_hist.append(np.abs(np.exp(1j*phi).mean()))
        phi_hist.append(phi.copy())

R_hist = np.array(R_hist)
phi_hist = np.array(phi_hist)
t_arr = np.arange(burn,T)*dt

# ----- analysis -----
R_mean = R_hist.mean()
R_std  = R_hist.std()
print(f"Global coherence <R>={R_mean:.3f} ± {R_std:.3f}")

# node 0 vs average phase
phi0 = np.unwrap(phi_hist[:,0])
phi_mean = np.unwrap(np.angle(np.exp(1j*phi_hist).mean(1)))
dphi0 = phi0 - phi_mean

# FFT of global R(t) to show response to perturbation
Nfft = 4096
w = 0.5*(1 - np.cos(2*np.pi*np.arange(min(len(R_hist),Nfft))/min(len(R_hist),Nfft)))
F = np.fft.rfft(R_hist[:len(w)]*w)
f = np.fft.rfftfreq(len(w), d=dt)
P = (F*np.conj(F)).real

# expected modulation frequency (cycles/step)
f_mod = pert_freq/(2*np.pi)

# ----- plots -----
fig,axs = plt.subplots(2,2, figsize=(12,7))

# 1. global coherence over time
axs[0,0].plot(t_arr, R_hist, lw=0.8)
axs[0,0].set_title("Global coherence R(t)")
axs[0,0].set_xlabel("time"); axs[0,0].set_ylabel("R")

# 2. spectral response of R
axs[0,1].plot(f, P/np.max(P)+1e-12)
axs[0,1].axvline(f_mod, color="crimson", ls="--", label="modulation f")
axs[0,1].set_xlim(0, 3*f_mod)
axs[0,1].set_title("R(t) spectral response")
axs[0,1].set_xlabel("frequency (cycles/step)"); axs[0,1].legend()

# 3. phase deviation of perturbed node
axs[1,0].plot(t_arr, dphi0, lw=0.8, color="tab:orange")
axs[1,0].set_title("Phase deviation node 0 − global ⟨φ⟩")
axs[1,0].set_xlabel("time"); axs[1,0].set_ylabel("Δφ₀ (rad)")

# 4. instantaneous node phases (subset)
for k in range(min(N,6)):
    axs[1,1].plot(t_arr, np.unwrap(phi_hist[:,k]), lw=0.8, label=f"node {k}")
axs[1,1].set_title("Unwrapped node phases")
axs[1,1].set_xlabel("time"); axs[1,1].legend(fontsize=8, ncol=2)

plt.tight_layout(); plt.show()
```

```python
# --- SWG/SCG AB-XI — Curvature-energy redistribution under geometry modulation ---
# Tracks internal energies, pump/dissipation power, and χ-bus exchange in a weak-link
# network (reduced phase model used in AB-IX / AB-X).

import numpy as np, matplotlib.pyplot as plt

# ---------------- network & dynamics (as in AB-X) ----------------
N = 8
topology = "ring"
dt = 0.10
T  = 22000
burn = int(0.20*T)

alpha = 0.05      # damping
lam   = 0.20      # local curvature locking
K     = 0.08      # coupling strength

f0     = 0.00113          # base nat. freq (cycles/step)
spread = 0.04             # ± dispersion
noise  = 2e-3
rng    = np.random.default_rng(42)

# geometry-floor perturbation on node 0 (metric modulation)
pert_amp  = 0.05                   # ±5%
pert_omega= 2*np.pi*0.0003         # rad/step
f_mod     = pert_omega/(2*np.pi)   # cycles/step (for annotation)

# natural freqs (cycles/step -> rad/step)
freqs = f0*(1.0 + spread*rng.uniform(-1,1,N))
omega0 = 2*np.pi*freqs

# adjacency (ring)
A = np.zeros((N,N))
for k in range(N): A[k,(k+1)%N] = A[(k+1)%N,k] = 1.0
deg = np.maximum(A.sum(1), 1.0)

def coup(phi):
    S = np.zeros_like(phi)
    for k in range(N):
        S[k] = np.sum(A[k]*np.sin(phi - phi[k]))/deg[k]
    return S

# state
phi  = rng.uniform(-np.pi, np.pi, N)
phid = np.zeros(N)

# logs
Ekin, Eloc, Eint = [], [], []
P_pump, P_diss, P_bus = [], [], []
Cum_pump, Cum_diss, Cum_bus = [], [], []
residual = []  # Eint + Cum_bus + Cum_diss - Cum_pump (≈ const)

cum_p, cum_d, cum_b = 0.0, 0.0, 0.0

for t in range(T):
    # geometry-floor modulation on node 0
    metric_mod = 1.0 + pert_amp*np.sin(pert_omega*t)
    omega_eff  = omega0.copy()
    omega_eff[0] *= metric_mod
    domega     = omega_eff - omega0  # only node 0 nonzero

    # dynamics (inertial Kuramoto + local locking + coupling)
    phidd = -alpha*phid - lam*np.sin(phi) + omega_eff + K*coup(phi)

    # instantaneous powers (before update)
    Pp = float(np.dot(phid, domega))                 # pump (work rate from metric mod)
    Pd = float(alpha*np.sum(phid*phid))             # dissipation (always ≥0)
    Pc = float(np.dot(phid, K*coup(phi)))           # power delivered to nodes by coupling
    # by definition χ-bus accumulates with opposite sign:
    Pb = -Pc

    # step
    phid = phid + dt*phidd + np.sqrt(dt)*noise*rng.standard_normal(N)
    phi  = (phi + dt*phid + np.pi)%(2*np.pi) - np.pi

    # energies after step
    Ek = 0.5*np.sum(phid*phid)
    Ul = lam*np.sum(1.0 - np.cos(phi))
    Ei = Ek + Ul

    # accumulate
    cum_p += Pp*dt
    cum_d += Pd*dt
    cum_b += Pb*dt

    # record (after burn)
    if t >= burn:
        Ekin.append(Ek); Eloc.append(Ul); Eint.append(Ei)
        P_pump.append(Pp); P_diss.append(Pd); P_bus.append(Pb)
        Cum_pump.append(cum_p); Cum_diss.append(cum_d); Cum_bus.append(cum_b)
        residual.append(Ei + cum_b + cum_d - cum_p)

# arrays & timebase
t_arr = np.arange(len(Eint))*dt

Ekin, Eloc, Eint = map(np.array, (Ekin, Eloc, Eint))
P_pump, P_diss, P_bus = map(np.array, (P_pump, P_diss, P_bus))
Cum_pump, Cum_diss, Cum_bus = map(np.array, (Cum_pump, Cum_diss, Cum_bus))
residual = np.array(residual)

# ---------------- plots ----------------
fig,axs = plt.subplots(2,2, figsize=(12,7))

# (1) internal energies (kinetic, local, total)
axs[0,0].plot(t_arr, Ekin, lw=0.8, label="E_kin")
axs[0,0].plot(t_arr, Eloc, lw=0.8, label="E_loc")
axs[0,0].plot(t_arr, Eint, lw=1.2, label="E_int = E_kin+E_loc")
axs[0,0].set_title("Internal curvature energies"); axs[0,0].legend(); axs[0,0].grid(alpha=0.3)
axs[0,0].set_xlabel("time")

# (2) instantaneous powers (rates)
axs[0,1].plot(t_arr, P_pump, lw=0.8, label="P_pump")
axs[0,1].plot(t_arr, P_diss, lw=0.8, label="P_diss")
axs[0,1].plot(t_arr, P_bus,  lw=0.8, label="P_bus (χ-bus)")
axs[0,1].axhline(0, color="#888", ls=":")
axs[0,1].set_title("Power channels"); axs[0,1].legend(); axs[0,1].grid(alpha=0.3)
axs[0,1].set_xlabel("time")

# (3) cumulative energy exchange
axs[1,0].plot(t_arr, Cum_pump, lw=1.0, label="∫P_pump dt (work in)")
axs[1,0].plot(t_arr, Cum_diss, lw=1.0, label="∫P_diss dt (loss)")
axs[1,0].plot(t_arr, Cum_bus,  lw=1.0, label="∫P_bus dt (χ-bus store)")
axs[1,0].set_title("Cumulative energy flow"); axs[1,0].legend(); axs[1,0].grid(alpha=0.3)
axs[1,0].set_xlabel("time")

# (4) conservation residual (should be ~flat)
axs[1,1].plot(t_arr, residual - residual[0], lw=1.0)
axs[1,1].axhline(0, color="#888", ls=":")
axs[1,1].set_title("Conservation residual  E_int + E_bus + E_diss − W_pump")
axs[1,1].set_xlabel("time"); axs[1,1].grid(alpha=0.3)

plt.tight_layout(); plt.show()

print(f"<E_int> = {Eint.mean():.4e}  |  <P_pump>={P_pump.mean():+.3e}  <P_bus>={P_bus.mean():+.3e}  <P_diss>={P_diss.mean():+.3e}")
print(f"Residual drift (std): {np.std(residual - residual[0]):.3e}  (smaller is better)")
```

```python
# --- SWG/SCG AB-XII — χ-energy flux & continuity in a weak-link ring ---------
import numpy as np, matplotlib.pyplot as plt

# ---------------- geometry (same as AB-IV family) ----------------
H=W=180
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)

R_in, R_out = 44, 58
R_core      = 34
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# mid-ring path (for quick scalars if needed)
r_mid = 0.5*(R_in+R_out)
PATH = np.logical_and(np.abs(R - r_mid) <= 0.75, RING)

# weak link near +x
def wrap_pi(z): return ((z + np.pi)%(2*np.pi) - np.pi).astype(np.float32)
def lap(Z):
    return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)
def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def div(Fx, Fy):
    return 0.5*(np.roll(Fx,-1,1)-np.roll(Fx,1,1)) + 0.5*(np.roll(Fy,-1,0)-np.roll(Fy,1,0))

link_center = 0.0
link_halfw  = 8*np.pi/180
ANG = ang
WEAK = RING & (np.abs(((ANG-link_center)+np.pi)%(2*np.pi)-np.pi) <= link_halfw)

# ---------------- dynamics & drive ----------------
dt      = 0.10
steps   = 12000
alpha   = 0.035
kappa   = 1.45
lambda_strong = 0.050
lambda_weak   = 0.010
pin_core = 0.020     # core pin (implements flux bias)
guard    = 0.004     # quiet off-ring

Phi_dc = 0.0
Phi_ac = 2.0
Omega  = 2*np.pi*0.00113    # ~ your ring’s f0 from AB-VI

lambda_map = np.where(WEAK, lambda_weak, lambda_strong).astype(np.float32)

# state
theta = wrap_pi(0.0*ANG)
thdot = np.zeros_like(theta)

# logs
E_total, W_pump, W_diss, W_guard, Res_global = [], [], [], [], []
cum_p, cum_d, cum_g = 0.0, 0.0, 0.0

# choose frames to snapshot
snap_idx = np.linspace(int(0.5*steps), steps-1, 4, dtype=int)
snaps_e, snaps_Jx, snaps_Jy = [], [], []
snaps_divJ, snaps_rhs = [], []

rng = np.random.default_rng(2025)

for t in range(steps):
    # flux drive on the core
    Phi_t  = Phi_dc + Phi_ac*np.sin(Omega*t)
    tgt    = wrap_pi(Phi_t * ANG)

    # PDE pieces
    Lap = lap(theta)
    f_core  = np.zeros_like(theta); f_core[CORE] = -pin_core * wrap_pi(theta[CORE] - tgt[CORE])
    f_guard = np.zeros_like(theta); f_guard[~(RING|CORE)] = -guard*theta[~(RING|CORE)]
    f_all   = f_core + f_guard

    # θ̈ = κΔθ − λ sinθ + f_all − α θ̇
    thdd = kappa*Lap - lambda_map*np.sin(theta) + f_all - alpha*thdot

    # update + tiny thermal dither (optional)
    thdot = thdot + dt*thdd + np.sqrt(dt)*4e-3*rng.standard_normal(thdot.shape)
    theta = wrap_pi(theta + dt*thdot)

    # --- energy densities & flux (AFTER update) ---
    gx, gy = grad(theta)
    e_kin  = 0.5*(thdot**2)
    e_grad = 0.5*kappa*(gx*gx + gy*gy)
    e_loc  = 0.5*lambda_map*(1.0 - np.cos(theta))
    e      = e_kin + e_grad + e_loc

    # Jχ = -κ θ̇ ∇θ
    Jx = -kappa * thdot * gx
    Jy = -kappa * thdot * gy
    divJ = div(Jx, Jy)

    # time derivative of e via finite diff (only after first step)
    if t == 0:
        e_prev = e.copy()
        de_dt  = np.zeros_like(e)
    else:
        de_dt  = (e - e_prev)/dt
        e_prev = e.copy()

    # local continuity residual: r = de/dt + ∇·J − (θ̇ f_all − α θ̇²)
    rhs = thdot*f_all - alpha*(thdot**2)
    rmap = de_dt + divJ - rhs

    # accumulate global integrals (over whole grid)
    E_total.append(float(np.sum(e)))
    cum_p += float(np.sum(thdot*f_core))*dt
    cum_d += float(np.sum(alpha*thdot*thdot))*dt
    cum_g += float(np.sum(thdot*f_guard))*dt
    W_pump.append(cum_p); W_diss.append(cum_d); W_guard.append(cum_g)
    Res_global.append(float(np.sum(rmap)))   # domain integral (should hover near 0)

    # snapshots
    if t in snap_idx:
        snaps_e.append(e.copy()); snaps_Jx.append(Jx.copy()); snaps_Jy.append(Jy.copy())
        snaps_divJ.append(divJ.copy()); snaps_rhs.append(rhs.copy())

# -------------- plots: energy/flux maps + balance ----------------
fig,axs = plt.subplots(3,4, figsize=(14,8))
for k in range(4):
    im0 = axs[0,k].imshow(snaps_e[k], cmap="inferno"); axs[0,k].set_title(f"e(x,y) @ {snap_idx[k]}")
    axs[0,k].axis("off"); fig.colorbar(im0, ax=axs[0,k], fraction=0.046)

    # overlay coarse quiver of Jχ
    skip=6
    axs[1,k].imshow(np.zeros_like(snaps_e[k]), cmap="gray_r", vmin=0, vmax=1); axs[1,k].axis("off")
    axs[1,k].quiver(x[::skip,::skip], y[::skip,::skip],
                    snaps_Jx[k][::skip,::skip], snaps_Jy[k][::skip,::skip],
                    color="turquoise", angles="xy", scale_units="xy", scale=1.0, width=0.002)
    axs[1,k].set_title("Jχ field (arrows)")

    # continuity: compare -divJ vs (rhs - de/dt)
    im2 = axs[2,k].imshow(-(snaps_divJ[k]), cmap="coolwarm"); axs[2,k].axis("off")
    fig.colorbar(im2, ax=axs[2,k], fraction=0.046)
    axs[2,k].set_title("−∇·Jχ  (sign conv.)")

plt.tight_layout()
plt.show()

# global balance timeline
E_total = np.array(E_total)
W_pump  = np.array(W_pump)
W_diss  = np.array(W_diss)
W_guard = np.array(W_guard)
Res_global = np.array(Res_global)

tarr = np.arange(len(E_total))*dt

fig,axs = plt.subplots(1,2, figsize=(12,4))
axs[0].plot(tarr, E_total, label="E_int")
axs[0].plot(tarr, W_pump, label="∫pump")
axs[0].plot(tarr, W_diss, label="∫diss")
axs[0].plot(tarr, W_guard, label="∫guard")
axs[0].set_title("Global energy balance (cumulatives)"); axs[0].legend(); axs[0].grid(alpha=0.3)
axs[1].plot(tarr, np.cumsum(Res_global), label="∫∫(de/dt+divJ−rhs) dxdt")
axs[1].axhline(0, color="#888", ls=":"); axs[1].set_title("Continuity residual (integrated)")
axs[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()

print(f"Residual mean ≈ {Res_global.mean():+.3e}   std ≈ {Res_global.std():.3e}")
print("Notes: (i) small nonzero residuals are expected from finite differencing & boundaries;")
print("       (ii) pump power enters only in CORE; guard removes energy off-ring; dissipation α≥0 everywhere.")
```

```python
# --- SWG/SCG AB-XIII — χ-wave momentum & radiation pressure -------------------
# Momentum density pχ = θ̇ ∇θ, pressure on ring arc from normal momentum flux.
# Builds on the AB-XII weak-link ring setup (self-contained here).

import numpy as np, matplotlib.pyplot as plt

# ---------------- geometry ----------------
H=W=180
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx,cy = W//2, H//2
X,Y = x-cx, y-cy
R = np.sqrt(X*X + Y*Y)
ang = np.arctan2(Y, X)

R_in, R_out = 44, 58
R_core      = 34
RING = (R>=R_in) & (R<=R_out)
CORE = (R<=R_core)

# normal (radial) unit vectors
nx_r = np.where(R>0, X/R, 0.0)
ny_r = np.where(R>0, Y/R, 0.0)

# weak link near +x
def wrap_pi(z): return ((z + np.pi)%(2*np.pi) - np.pi).astype(np.float32)
def lap(Z):  return (np.roll(Z,1,0)+np.roll(Z,-1,0)+np.roll(Z,1,1)+np.roll(Z,-1,1)-4*Z)
def grad(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def div(Fx, Fy):
    return 0.5*(np.roll(Fx,-1,1)-np.roll(Fx,1,1)) + 0.5*(np.roll(Fy,-1,0)-np.roll(Fy,1,0))

link_center = 0.0
link_halfw  = 8*np.pi/180
WEAK = RING & (np.abs(((ang-link_center)+np.pi)%(2*np.pi)-np.pi) <= link_halfw)

# boundary arc on which we'll measure pressure (a 40° sector around +x outer rim)
arc_w = 20*np.pi/180
ARC = (np.abs(((ang-link_center)+np.pi)%(2*np.pi)-np.pi) <= arc_w) & (R>=R_out-1.5) & (R<=R_out+1.5)

# ---------------- dynamics & drive (as AB-XII) ----------------
dt      = 0.10
steps   = 12000
alpha   = 0.035
kappa   = 1.45
lambda_strong = 0.050
lambda_weak   = 0.010
pin_core = 0.020
guard    = 0.004

Phi_dc = 0.0
Phi_ac = 2.0
Omega  = 2*np.pi*0.00113

lambda_map = np.where(WEAK, lambda_weak, lambda_strong).astype(np.float32)

theta = wrap_pi(0.0*ang)
thdot = np.zeros_like(theta)

rng = np.random.default_rng(13)

# logs
P_net_x, P_net_y = [], []          # ∫ pχ dA (domain)
F_arc = []                         # radiation-like force on arc
Flux_int = []                      # ∫ |Jχ| dA (magnitude of χ-flux)
snap_idx = np.linspace(int(0.55*steps), steps-1, 4, dtype=int)
snaps_pmag, snaps_Jx, snaps_Jy = [], [], []

for t in range(steps):
    # drive on core
    Phi_t = Phi_dc + Phi_ac*np.sin(Omega*t)
    tgt   = wrap_pi(Phi_t * ang)

    Lap = lap(theta)
    f_core  = np.zeros_like(theta); f_core[CORE] = -pin_core * wrap_pi(theta[CORE] - tgt[CORE])
    f_guard = np.zeros_like(theta); f_guard[~(RING|CORE)] = -guard*theta[~(RING|CORE)]
    f_all   = f_core + f_guard

    # θ̈ = κΔθ − λ sinθ + f_all − α θ̇
    thdd = kappa*Lap - lambda_map*np.sin(theta) + f_all - alpha*thdot

    # update (+ small thermal dither)
    thdot = thdot + dt*thdd + np.sqrt(dt)*4e-3*rng.standard_normal(thdot.shape)
    theta = wrap_pi(theta + dt*thdot)

    # fields
    gx, gy = grad(theta)

    # χ-energy flux (Poynting-like): Jχ = -κ θ̇ ∇θ
    Jx = -kappa * thdot * gx
    Jy = -kappa * thdot * gy
    Jmag = np.hypot(Jx, Jy)

    # χ-momentum density: pχ = θ̇ ∇θ
    px = thdot * gx
    py = thdot * gy

    # net momentum (domain integral) — tracks directionality of transport
    P_net_x.append(float(np.sum(px)))
    P_net_y.append(float(np.sum(py)))

    # radiation-like force on outer-arc: integrate normal momentum flux
    # Π_n ≈ (pχ · n) over the arc cells, scaled by cell "edge length" (~1 here)
    Pi_n = px*nx_r + py*ny_r
    F_arc.append(float(np.sum(Pi_n[ARC])))

    # total χ-flux magnitude (proxy for transport activity)
    Flux_int.append(float(np.sum(Jmag)))

    # snapshots
    if t in snap_idx:
        snaps_pmag.append(np.hypot(px, py))
        snaps_Jx.append(Jx.copy()); snaps_Jy.append(Jy.copy())

# arrays
P_net_x = np.array(P_net_x); P_net_y = np.array(P_net_y)
F_arc   = np.array(F_arc);   Flux_int = np.array(Flux_int)
tarr    = np.arange(steps)*dt

# ---------------- plots ----------------
fig,axs = plt.subplots(3,4, figsize=(14,8))
for k in range(4):
    im0 = axs[0,k].imshow(snaps_pmag[k], cmap="magma"); axs[0,k].set_title(f"|pχ| @ {snap_idx[k]}")
    axs[0,k].axis("off"); fig.colorbar(im0, ax=axs[0,k], fraction=0.046)

    axs[1,k].imshow(np.zeros_like(snaps_pmag[k]), cmap="gray_r", vmin=0, vmax=1); axs[1,k].axis("off")
    skip=6
    axs[1,k].quiver(x[::skip,::skip], y[::skip,::skip],
                    snaps_Jx[k][::skip,::skip], snaps_Jy[k][::skip,::skip],
                    color="turquoise", angles="xy", scale_units="xy", scale=1.0, width=0.002)
    axs[1,k].set_title("Jχ (arrows)")

    # highlight the pressure arc
    mask = np.zeros_like(snaps_pmag[k]); mask[ARC] = 1.0
    axs[2,k].imshow(mask, cmap="Reds"); axs[2,k].axis("off"); axs[2,k].set_title("pressure arc")

plt.tight_layout(); plt.show()

# time series: momentum, pressure, and flux-transport correlation
fig,axs = plt.subplots(2,2, figsize=(12,6))
axs[0,0].plot(tarr, P_net_x, lw=0.9, label="Px"); axs[0,0].plot(tarr, P_net_y, lw=0.9, label="Py")
axs[0,0].set_title("Net χ-momentum (domain)"); axs[0,0].legend(); axs[0,0].grid(alpha=0.3)

axs[0,1].plot(tarr, F_arc, lw=0.9, color="tab:red")
axs[0,1].set_title("Radiation-like force on outer arc"); axs[0,1].grid(alpha=0.3)

axs[1,0].plot(tarr, Flux_int/np.max(Flux_int), lw=0.9, label="∫|Jχ| dA (norm)")
axs[1,0].plot(tarr, (P_net_x**2+P_net_y**2)**0.5 / np.max((P_net_x**2+P_net_y**2)**0.5), lw=0.9, label="|Pχ| (norm)")
axs[1,0].set_title("Transport–momentum correlation"); axs[1,0].legend(); axs[1,0].grid(alpha=0.3)

# simple correlation metrics
from numpy import corrcoef
rho_P_F  = np.corrcoef(P_net_x, F_arc)[0,1]
rho_flux = np.corrcoef(Flux_int, (P_net_x**2+P_net_y**2)**0.5)[0,1]

axs[1,1].axis("off")
axs[1,1].text(0.02, 0.85, f"corr(Px, F_arc) ≈ {rho_P_F:+.3f}", fontsize=12)
axs[1,1].text(0.02, 0.65, f"corr(∫|Jχ|, |Pχ|) ≈ {rho_flux:+.3f}", fontsize=12)
axs[1,1].text(0.02, 0.45, "Interpretation:", fontsize=12, fontweight="bold")
axs[1,1].text(0.03, 0.30, "• χ-flux activity tracks momentum magnitude", fontsize=11)
axs[1,1].text(0.03, 0.18, "• normal momentum flux on arc = radiation-like pressure", fontsize=11)

plt.tight_layout(); plt.show()
```
