```python
# SWG/SCG Magnetism · Exp 1: single winding → "B-like" curl and invariants
# One cell, pure NumPy/Matplotlib. No external deps.
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
EPS = 1e-12

# --- grid & helpers ---
H=W=240; y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X = (x-cx).astype(np.float32); Y = (y-cy).astype(np.float32)
R = np.hypot(X,Y) + 1e-6
ang = np.arctan2(Y, X)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

def moving_mean2(Z, k=7):
    k = int(k)
    if k<=1: return Z
    v = Z
    for _ in range(2):
        v = (np.roll(v,1,0)+v+np.roll(v,-1,0))/3
        v = (np.roll(v,1,1)+v+np.roll(v,-1,1))/3
    return v

# --- imprint a loop: phase winding m=±1; amplitude ring seed ---
m = +1  # try -1 to flip "polarity"
theta = (m*ang).astype(np.float32)

A = (0.25*np.exp(-(R/22.0)**2) + 0.05*rng.standard_normal((H,W))).astype(np.float32)
A = np.clip(A, 0.0, None)

# memory χ as slow envelope
chi = moving_mean2(A, k=11).astype(np.float32)

# --- quick SWG-ish relax (envelope & phase diffusion + saturating nonlinearity) ---
dt=0.12; steps=120
def sat(u, alpha=0.20): return u/(1.0+alpha*(u**2))

for _ in range(steps):
    LA = lap2(A); Lth = lap2(theta)
    gthx,gthy = grad2(theta); gth2 = gthx*gthx + gthy*gthy
    # amplitude: diffusion - phase-curv penalty + weak nonlinearity to confine
    dA = 0.60*LA - 0.06*np.sqrt(gth2) - 0.02*A**3
    # phase: diffusion + weak coupling to amplitude gradient
    dth = 0.35*Lth + 0.25*(grad2(A)[0]*0 + grad2(A)[1]*0)  # minimal coupling
    # memory: EMA toward current amplitude
    chi = 0.995*chi + 0.005*A
    A   = np.clip(A + dt*sat(dA, alpha=0.15), 0, None)
    theta = (theta + dt*sat(dth, alpha=0.15) + np.pi)%(2*np.pi)-np.pi

# --- SCG invariants ---
# curvature face of memory (κ from χ), geometry floor components
kappa = np.abs(lap2(chi))/(np.abs(chi)+EPS)
L_local = 1.0/np.sqrt(kappa+EPS)
L_metric = np.percentile(L_local, 10)  # robust "metric" floor (p10)
L_dyn = 1.0
L_geom = np.sqrt(L_dyn**2 + L_metric**2)

# coherence entropy (amplitude weights), curvature variance, energy flux, dislocation density, order R
w = A/(np.sum(A)+EPS)
Hc = float(-np.sum(w*np.log(w+EPS)))
Kvar = float(np.var(kappa))
E = A*A
PhiE = float(np.mean((np.roll(E,-1,1)-E)/(np.abs(E)+EPS)))  # coarse proxy
dth = (np.roll(theta,-1,1)-theta + np.pi)%(2*np.pi)-np.pi
med = np.median(dth); mad = 1.4826*np.median(np.abs(dth-med))+EPS
rho_phi = float(np.mean(np.abs(dth-med) > (med + 2.0*mad)))
Rop = float(np.abs(np.mean(np.exp(1j*theta))))

# --- "B-like" field: treat J ≡ ∇θ, then B_z ≡ ∂x J_y - ∂y J_x ; vector plot of J and colormap of B_z
Jx,Jy = grad2(theta)
Bz = np.roll(Jy,-1,1)-Jy - (np.roll(Jx,-1,0)-Jx)

# --- visuals ---
fig,axs = plt.subplots(2,3,figsize=(12,7))
im0=axs[0,0].imshow(A,cmap='magma'); axs[0,0].set_title("Amplitude A")
fig.colorbar(im0,ax=axs[0,0],shrink=0.8)
im1=axs[0,1].imshow(theta,cmap='twilight'); axs[0,1].set_title("Phase θ (winding)")
fig.colorbar(im1,ax=axs[0,1],shrink=0.8)
im2=axs[0,2].imshow(Bz,cmap='coolwarm'); axs[0,2].set_title('“B-like” $B_z = \\partial_x J_y-\\partial_y J_x$')
fig.colorbar(im2,ax=axs[0,2],shrink=0.8)

skip=6
axs[1,0].quiver(x[::skip,::skip], y[::skip,::skip], Jx[::skip,::skip], Jy[::skip,::skip], color='k', scale=80)
axs[1,0].set_title("Current-like field J = ∇θ")

axs[1,1].hist(L_local.ravel(),bins=60,color='#555'); axs[1,1].axvline(L_metric,color='r',ls='--',label='L_metric p10')
axs[1,1].axvline(L_geom,color='g',ls='--',label='L_geom'); axs[1,1].legend(); axs[1,1].set_title("Coherence scales (floor)")

axs[1,2].axis('off')
axs[1,2].text(0.0,1.0,(
    f"H={Hc:.3f}\nσ²κ={Kvar:.3e}\nΦ_E≈{PhiE:.3e}\nρ_φ={rho_phi:.3f}\nR={Rop:.3f}\n"
    f"L_metric≈{L_metric:.3f}\nL_geom≈{L_geom:.3f}"
),va='top',family='monospace',fontsize=11)
plt.tight_layout(); plt.show()
```

```python
# SWG/SCG Magnetism · Exp 3: Meissner exclusion + single-quantum vortex
import numpy as np, matplotlib.pyplot as plt
EPS=1e-12
H=W=260; y,x=np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx); Y=(y-cy); R=np.hypot(X,Y)+1e-6

def grad2(Z):
    return 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1)), 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# seed: uniform weak phase tilt + central winding m=+1
theta = (0.002*X + 0.002*Y + np.arctan2(Y,X)).astype(np.float32)
A = np.clip(0.20*np.exp(-R**2/(2*(34.0**2))),0,None)

# define superconducting “island” (disk) with a small hole (annulus)
Rsc = 60.0; Rhole = 8.0
SC = (R < Rsc).astype(np.float32)
HOLE = (R < Rhole).astype(np.float32)
SCmask = np.logical_and(SC==1, HOLE==0)  # disk minus small central hole

# floor barrier: penalize phase gradients inside SCmask (Meissner)
def meissner_step(theta, A, lam_grad=2.0):
    Jx,Jy = grad2(theta)
    pen = lam_grad*SCmask*(Jx + Jy)  # simple isotropic penalty
    return 0.35*lap2(theta) - pen

# evolve
dt=0.10
for t in range(140):
    # amplitude: smooth & confine near island boundary (optional cosmetic)
    A = np.clip(A + dt*(0.45*lap2(A) - 0.02*A**3), 0, None)
    # phase: diffusion + Meissner suppression in SC
    theta = (theta + dt*meissner_step(theta, A) + np.pi)%(2*np.pi)-np.pi

# fields
Jx,Jy = grad2(theta)
Bz = np.roll(Jy,-1,1)-Jy - (np.roll(Jx,-1,0)-Jx)

fig,axs=plt.subplots(1,3,figsize=(12,4))
axs[0].imshow(SCmask, cmap='gray_r'); axs[0].set_title("SC region (white), hole at center")
axs[1].imshow(Bz,cmap='coolwarm'); axs[1].set_title('“B-like” $B_z$ — expelled, threads the hole')
skip=6
axs[2].quiver(x[::skip,::skip], y[::skip,::skip], Jx[::skip,::skip], Jy[::skip,::skip], color='k', scale=85)
axs[2].set_title("J field excluded by SC; a single vortex in the hole")
plt.tight_layout(); plt.show()
```

```python
# SWG/SCG Magnetism · Exp 4 — Quantization & Floor Validity for a Single Loop
# Self-contained; one cell; NumPy + Matplotlib only.

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(12)
EPS = 1e-12

# --- grid ---
H=W=260
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)
R = np.hypot(X,Y) + 1e-6
ang = np.arctan2(Y,X)

# --- helpers ---
def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

def moving_mean2(Z, k=7):
    k = int(k)
    if k<=1: return Z
    v = Z
    for _ in range(2):
        v = (np.roll(v,1,0)+v+np.roll(v,-1,0))/3
        v = (np.roll(v,1,1)+v+np.roll(v,-1,1))/3
    return v

def robust_vlims(Z, lo=2, hi=98):
    vmin = np.percentile(Z, lo); vmax = np.percentile(Z, hi)
    if vmax <= vmin:  # degenerate
        vmax = vmin + 1e-6
    return vmin, vmax

# --- imprint single loop (m = +1), amplitude ring seed ---
m = +1
theta = (m*ang).astype(np.float32)
A = (0.25*np.exp(-(R/22.0)**2) + 0.05*rng.standard_normal((H,W))).astype(np.float32)
A = np.clip(A, 0.0, None)
chi = moving_mean2(A, k=11).astype(np.float32)

# --- evolve (gentle SWG-ish relax) ---
dt=0.12; steps=140
def sat(u, alpha=0.20): return u/(1.0+alpha*(u**2))
for _ in range(steps):
    LA = lap2(A); Lth = lap2(theta)
    gthx,gthy = grad2(theta); gth2 = gthx*gthx + gthy*gthy
    dA = 0.60*LA - 0.06*np.sqrt(gth2) - 0.02*A**3
    dth = 0.35*Lth  # minimal coupling; amplitude coupling omitted on purpose
    chi = 0.995*chi + 0.005*A
    A   = np.clip(A + dt*sat(dA, alpha=0.15), 0, None)
    theta = (theta + dt*sat(dth, alpha=0.15) + np.pi)%(2*np.pi)-np.pi

# --- fields & floor metrics ---
Jx,Jy = grad2(theta)
Bz = np.roll(Jy,-1,1)-Jy - (np.roll(Jx,-1,0)-Jx)

kappa = np.abs(lap2(chi))/(np.abs(chi)+EPS)
L_local = 1.0/np.sqrt(kappa+EPS)
L_metric = np.percentile(L_local, 10)        # geometry metric floor (p10)
L_dyn = 1.0
L_geom = np.sqrt(L_dyn**2 + L_metric**2)

# Floor-Violation Rate (FVR): fraction where L_local < L_metric (i.e., below metric floor)
FVR = float(np.mean(L_local < L_metric))

# --- circulation vs radius (with unwrap on the path) ---
def circulation_at_radius(theta, r, cx, cy, samples=720):
    t = np.linspace(0, 2*np.pi, samples, endpoint=False)
    yy = (cy + r*np.sin(t)).astype(int)
    xx = (cx + r*np.cos(t)).astype(int)
    th_path = theta[yy, xx]
    th_unwrap = np.unwrap(th_path)  # unwrap along the loop
    d = np.diff(th_unwrap, append=th_unwrap[0])
    return float(np.sum(d))

# sample radii outward from the core; skip the singular center
radii = np.arange(8, min(H,W)//2 - 4, 2, dtype=int)
circs = np.array([circulation_at_radius(theta, int(r), cx, cy) for r in radii])
quanta = circs/(2*np.pi)

# estimate a plateau (robust) away from core & far boundary
inner = int(0.15*len(radii)); outer = int(0.80*len(radii))
plateau_est = float(np.median(quanta[inner:outer])) if outer>inner else float(np.median(quanta))

# --- visuals ---
fig = plt.figure(figsize=(12,8))
gs = fig.add_gridspec(2,3)

ax0 = fig.add_subplot(gs[0,0]); vm,vM = robust_vlims(A)
im0 = ax0.imshow(A, cmap='magma', vmin=vm, vmax=vM); ax0.set_title("Amplitude A")
fig.colorbar(im0, ax=ax0, shrink=0.8)

ax1 = fig.add_subplot(gs[0,1])
im1 = ax1.imshow(theta, cmap='twilight'); ax1.set_title("Phase θ (winding)")
fig.colorbar(im1, ax=ax1, shrink=0.8)

ax2 = fig.add_subplot(gs[0,2]); vb,vB = robust_vlims(Bz)
im2 = ax2.imshow(Bz, cmap='coolwarm', vmin=vb, vmax=vB); ax2.set_title('“B-like” $B_z$')
fig.colorbar(im2, ax=ax2, shrink=0.8)

ax3 = fig.add_subplot(gs[1,0])
skip=5
ax3.quiver(Jx[::skip,::skip], Jy[::skip,::skip], angles='xy', scale_units='xy', scale=1)
ax3.set_title("J = ∇θ (unit-scale quiver)")

ax4 = fig.add_subplot(gs[1,1])
ax4.plot(radii, quanta, lw=2)
ax4.axhline(1.0*m, ls='--', color='k', alpha=0.6, label='m target')
ax4.set_xlabel("radius (px)")
ax4.set_ylabel("circulation / (2π)")
ax4.set_title(f"Quantization vs radius · plateau ≈ {plateau_est:.2f} quanta")
ax4.legend()

ax5 = fig.add_subplot(gs[1,2])
# annular FVR: fraction below floor per ring
ann_FVR = []
for r in radii:
    rr = np.abs(R - r)
    band = rr < 1.0
    ann_FVR.append(np.mean(L_local[band] < L_metric))
ann_FVR = np.array(ann_FVR)
ax5.plot(radii, ann_FVR, lw=2)
ax5.set_xlabel("radius (px)")
ax5.set_ylabel("FVR (ring)")
ax5.set_title(f"Floor-Violation Rate · global FVR={FVR:.3f}")
plt.tight_layout(); plt.show()

print(f"Global FVR (L_local < L_metric): {FVR:.4f}")
print(f"Quantization plateau estimate: {plateau_est:.3f} quanta")
print(f"Bz range (robust 2–98%): {vb:.3e} .. {vB:.3e}  | full ptp={np.ptp(Bz):.3e}")
```

```python
# SWG/SCG Magnetism · Exp 4.2 — Circulating Vortex Stabilization (curl(J) feedback)
# Goal: lock a true quantized vortex (plateau ≈ ±1) via rotational feedback on phase.
# One self-contained cell. NumPy + Matplotlib only.

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
EPS = 1e-12

# --- grid ---
H=W=260
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)
R = np.hypot(X,Y) + 1e-6
ang = np.arctan2(Y,X)

# --- helpers ---
def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

def curl_from_J(Jx, Jy):
    # ∂x Jy - ∂y Jx
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

def robust_vlims(Z, lo=2, hi=98):
    vmin, vmax = np.percentile(Z,[lo,hi])
    if vmax<=vmin: vmax=vmin+1e-9
    return vmin, vmax

# --- seed: tighter amplitude core + slight smoothing; phase winding m=+1 ---
m = +1
theta = (m*ang).astype(np.float32)
# soften the branch cut a touch (two passes of neighbor average)
for _ in range(2):
    theta = (np.roll(theta,1,0)+theta+np.roll(theta,-1,0))/3
    theta = (np.roll(theta,1,1)+theta+np.roll(theta,-1,1))/3
theta = (theta + np.pi)%(2*np.pi) - np.pi

A = np.exp(-(R/10.0)**2).astype(np.float32)
A /= (np.max(A)+EPS)
A += 0.01*rng.standard_normal(A.shape).astype(np.float32)
A = np.clip(A,0,None)

# edge sponge to damp boundary artifacts
edge = np.minimum.reduce([x, y, W-1-x, H-1-y]).astype(np.float32)
sponge = np.clip((edge-10)/30, 0.0, 1.0)  # 0 near border, 1 in the interior

# --- dynamics: amplitude + phase with curl(J) feedback ---
dt = 0.08
steps = 220
Da, beta, gamma = 0.55, 0.12, 0.02        # amplitude: diffusion, phase-curv penalty, confinement
nu, omega = 0.18, 0.90                    # phase: diffusion, rotational feedback strength

def sat(u, a=0.18): return u/(1.0+a*u*u)

for _ in range(steps):
    # fields
    thx, thy = grad2(theta)
    gth2 = thx*thx + thy*thy
    # amplitude flow
    dA = Da*lap2(A) - beta*A*gth2 - gamma*A**3
    # current and rotational feedback
    Jx, Jy = (A*A)*thx, (A*A)*thy
    curlJ = curl_from_J(Jx, Jy)
    # phase flow: diffusion + curl(J)/(A^2) feedback (guarded by EPS), sponge to calm edges
    dth = nu*lap2(theta) + omega*curlJ/(A*A + 1e-6)
    # updates
    A = np.clip(A + dt*sat(dA,0.15), 0, None)
    theta = (theta + dt*sat(dth,0.15))*sponge + theta*(1.0 - sponge)
    theta = (theta + np.pi)%(2*np.pi) - np.pi

# --- diagnostics ---
# recompute J and B-like field after settle
thx, thy = grad2(theta)
Jx, Jy = (A*A)*thx, (A*A)*thy
Bz = curl_from_J(Jx, Jy)

# circulation vs radius (unwrap along path)
def circulation_at_radius(theta, r, cx, cy, samples=720):
    t = np.linspace(0, 2*np.pi, samples, endpoint=False)
    yy = (cy + r*np.sin(t)).astype(int)
    xx = (cx + r*np.cos(t)).astype(int)
    th = theta[yy, xx]
    th_unwrap = np.unwrap(th)  # unwrap around the loop
    return float(np.sum(np.diff(th_unwrap, append=th_unwrap[0])))

radii = np.arange(10, 120, 2, dtype=int)
circs = np.array([circulation_at_radius(theta, int(r), cx, cy) for r in radii])
quanta = circs/(2*np.pi)
plateau = float(np.median(quanta[15:80])) if len(quanta)>80 else float(np.median(quanta))

# --- visuals ---
fig,axs = plt.subplots(2,3,figsize=(12,8))
vm,vM = robust_vlims(A); im0 = axs[0,0].imshow(A, cmap='magma', vmin=vm, vmax=vM); axs[0,0].set_title("Amplitude A"); 
plt.colorbar(im0, ax=axs[0,0], shrink=0.8)
im1 = axs[0,1].imshow(theta, cmap='twilight'); axs[0,1].set_title("Phase θ (winding)"); plt.colorbar(im1, ax=axs[0,1], shrink=0.8)
vb,vB = robust_vlims(Bz); im2 = axs[0,2].imshow(Bz, cmap='coolwarm', vmin=vb, vmax=vB); axs[0,2].set_title('“B-like” $B_z$ = curl $J$'); 
plt.colorbar(im2, ax=axs[0,2], shrink=0.8)

skip=5
axs[1,0].quiver(Jx[::skip,::skip], Jy[::skip,::skip], angles='xy', scale_units='xy', scale=1)
axs[1,0].set_title("Current $J=A^2\\nabla\\theta$ (unit-scale quiver)")

axs[1,1].plot(radii, quanta, lw=2); axs[1,1].axhline(m, ls='--', color='k', alpha=0.65, label='target m=+1')
axs[1,1].set_xlabel("radius (px)"); axs[1,1].set_ylabel("circulation / $2\\pi$")
axs[1,1].set_title(f"Quantization plateau ≈ {plateau:.2f}"); axs[1,1].legend()

axs[1,2].hist(Bz.ravel(), bins=80, color='#666'); axs[1,2].set_title("Bz distribution")
plt.tight_layout(); plt.show()

print(f"Quantization plateau ≈ {plateau:.3f} quanta")
print(f"Bz robust range (2–98%): {vb:.3e} … {vB:.3e}; full ptp={np.ptp(Bz):.3e}")
```

```python
# SWG/SCG Magnetism · Exp 4.5 — XY vortex by energy descent (core pin + Neumann)
# Goal: obtain a stable m=±1 vortex with quantized circulation and clear J, B-like fields.
# One self-contained cell (NumPy + Matplotlib).

import numpy as np
import matplotlib.pyplot as plt

EPS = 1e-9

# --- grid & helpers (Neumann stencils: edge-mirror via padding) ---
H=W=260
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)
R = np.hypot(X,Y)+1e-6
ang = np.arctan2(Y,X).astype(np.float32)

def pad_edge(Z):
    return np.pad(Z, ((1,1),(1,1)), mode='edge')

def grad2_nb(Z):
    Zp = pad_edge(Z)
    gx = 0.5*(Zp[1:-1,2:] - Zp[1:-1,0:-2])
    gy = 0.5*(Zp[2:,1:-1] - Zp[0:-2,1:-1])
    return gx, gy

def lap2_nb(Z):
    Zp = pad_edge(Z)
    return (Zp[1:-1,2:] + Zp[1:-1,0:-2] + Zp[2:,1:-1] + Zp[0:-2,1:-1] - 4*Z)

def curl_from_J(Jx, Jy):
    Jxp, Jyp = pad_edge(Jx), pad_edge(Jy)
    dJydx = 0.5*(Jyp[1:-1,2:] - Jyp[1:-1,0:-2])
    dJxdy = 0.5*(Jxp[2:,1:-1] - Jxp[0:-2,1:-1])
    return dJydx - dJxdy

# --- XY energy with a core pin to enforce m=+1 at the origin ---
# Energy: E = ∫ [ (∇θ)^2 + κ_pin P (wrap(θ - θ_target))^2 ] dx
# We descend ∂E/∂θ to steady state.

m = +1
theta_target = (m*ang).astype(np.float32)

# core pin mask (small disk)
Rpin = 6.0
P = (R <= Rpin).astype(np.float32)

# initialize θ close to target + small noise (prevents getting stuck)
rng = np.random.default_rng(2)
theta = ((theta_target + 0.15*rng.standard_normal((H,W))).astype(np.float32) + np.pi)%(2*np.pi)-np.pi

# descent params
dt = 0.25
steps = 800
kappa_pin = 2.0  # strength of pin term

def wrap_pi(d):
    return (d + np.pi)%(2*np.pi) - np.pi

for t in range(steps):
    # gradient term
    Lth = lap2_nb(theta)
    # pin term: penalize deviation from target only in P region
    dphi = wrap_pi(theta - theta_target)
    pin_force = -kappa_pin * P * dphi
    # descent step
    theta = theta + dt*(Lth + pin_force)
    # keep in [-π,π) to avoid numeric blowup; our wrap is angle-safe
    theta = wrap_pi(theta)

# --- fields & quantization diagnostics ---
Jx, Jy = grad2_nb(theta)
Bz = curl_from_J(Jx, Jy)

# circulation vs radius with angular unwrapping along the loop
def circulation_at_radius(theta, r, cx, cy, samples=1024):
    t = np.linspace(0, 2*np.pi, samples, endpoint=False)
    yy = (cy + r*np.sin(t)).astype(int)
    xx = (cx + r*np.cos(t)).astype(int)
    th = theta[yy, xx]
    th_unwrap = np.unwrap(th)  # unwrap along the loop
    d = np.diff(th_unwrap, append=th_unwrap[0])
    return float(np.sum(d))

radii = np.arange(8, min(H,W)//2 - 8, 2, dtype=int)
circs = np.array([circulation_at_radius(theta, int(r), cx, cy) for r in radii], dtype=float)
quanta = circs/(2*np.pi)
# plateau (exclude tiny core and outer sponge-free boundary)
inner = int(0.12*len(radii)); outer = int(0.75*len(radii))
plateau = float(np.median(quanta[inner:outer])) if outer>inner else float(np.median(quanta))

# --- visuals ---
fig,axs = plt.subplots(2,3,figsize=(12,8))
im0=axs[0,0].imshow(theta, cmap='twilight'); axs[0,0].set_title("Phase θ (vortex)")
plt.colorbar(im0, ax=axs[0,0], shrink=0.8)

im1=axs[0,1].imshow(Bz, cmap='coolwarm', vmin=np.percentile(Bz,2), vmax=np.percentile(Bz,98))
axs[0,1].set_title('“B-like” curl J'); plt.colorbar(im1, ax=axs[0,1], shrink=0.8)

skip=5
axs[0,2].quiver(Jx[::skip,::skip], Jy[::skip,::skip], angles='xy', scale_units='xy', scale=1)
axs[0,2].set_title("Current J = ∇θ (unit-scale)")

axs[1,0].plot(radii, quanta, lw=2)
axs[1,0].axhline(m, ls='--', color='k', alpha=0.7, label='target m')
axs[1,0].set_xlabel("radius (px)"); axs[1,0].set_ylabel("circulation / 2π")
axs[1,0].set_title(f"Quantization plateau ≈ {plateau:.2f}"); axs[1,0].legend()

# pin diagnostics
axs[1,1].imshow(P, cmap='gray_r'); axs[1,1].set_title("Core pin mask (white)")
axs[1,1].axis('off')

# energy proxies (norms)
E_grad = np.mean(Jx*Jx + Jy*Jy)
axs[1,2].axis('off')
axs[1,2].text(0.0, 1.0,
              f"m={m:+d}\nE_grad≈{E_grad:.3e}\nplateau≈{plateau:.3f}",
              va='top', family='monospace', fontsize=11)
plt.tight_layout(); plt.show()

print(f"Quantization plateau ≈ {plateau:.3f} quanta")
```

```python
# SWG/SCG Magnetism · Exp 5 — Meissner exclusion with hole + robust flux quantization + SCG metrics
# One cell, NumPy + Matplotlib. Uses proper Meissner operator and your SCG floor/invariants.

import numpy as np, matplotlib.pyplot as plt
EPS = 1e-12
rng = np.random.default_rng(3)

# --- grid ---
H=W=260
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)
R = np.hypot(X,Y) + 1e-6
ang = np.arctan2(Y,X).astype(np.float32)

# --- helpers (periodic stencils are fine for this island-in-domain setup) ---
def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

def curl_from_J(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

def moving_mean2(Z, k=9):
    v = Z.copy()
    for _ in range(2):
        v = (np.roll(v,1,0)+v+np.roll(v,-1,0))/3
        v = (np.roll(v,1,1)+v+np.roll(v,-1,1))/3
    return v

def robust_vlims(Z, lo=2, hi=98):
    vmin, vmax = np.percentile(Z,[lo,hi]); 
    return vmin, (vmax if vmax>vmin else vmin+1e-9)

# --- seed field: gentle tilt + central winding; moderate envelope ---
theta = (0.003*X + 0.003*Y + ang).astype(np.float32)
A = (0.18*np.exp(-R**2/(2*(36.0**2))) + 0.01*rng.standard_normal((H,W))).astype(np.float32)
A = np.clip(A, 0, None)
chi = moving_mean2(A, k=11).astype(np.float32)

# --- SC island with a small hole (annulus) ---
Rsc = 60.0; Rhole = 9.0
SC = (R < Rsc).astype(np.float32)
HOLE = (R < Rhole).astype(np.float32)
SCmask = np.logical_and(SC==1, HOLE==0).astype(np.float32)  # island minus hole

# --- Proper Meissner operator:  -∇·(λ SC ∇θ)  (as used previously when you got ≈-1 quantum)
lam = 2.2
def meissner_operator(theta):
    Jx, Jy = grad2(theta)
    Jx *= SCmask; Jy *= SCmask
    divJ = 0.5*((np.roll(Jx,-1,1)-np.roll(Jx,1,1)) + (np.roll(Jy,-1,0)-np.roll(Jy,1,0)))
    return -lam*divJ

# --- evolve a bit
dt=0.10; steps=170
def sat(u,a=0.18): return u/(1.0+a*u*u)
for _ in range(steps):
    LA = lap2(A)
    # amplitude: smooth + gentle confinement
    dA = 0.45*LA - 0.02*A**3
    # phase: diffusion + Meissner suppression inside SC island
    dth = 0.33*lap2(theta) + meissner_operator(theta)
    # memory χ (slow)
    chi = 0.995*chi + 0.005*A
    A = np.clip(A + dt*sat(dA,0.15), 0, None)
    theta = (theta + dt*sat(dth,0.15) + np.pi)%(2*np.pi)-np.pi

# --- fields ---
thx, thy = grad2(theta)
Jx, Jy = thx, thy                 # J = ∇θ (phase current proxy)
Bz = curl_from_J(Jx, Jy)          # “B-like”

# --- robust winding around the hole via complex phase increments (closure ONCE) ---
def winding_complex(theta, r, cx, cy, samples=2048):
    t = np.linspace(0, 2*np.pi, samples, endpoint=False)
    yy = (cy + r*np.sin(t)).astype(int)
    xx = (cx + r*np.cos(t)).astype(int)
    z = np.exp(1j*theta[yy, xx])
    inc = np.angle(z[1:]*np.conj(z[:-1]))
    closure = np.angle(z[0]*np.conj(z[-1]))
    total = float(np.sum(inc) + closure)
    return total/(2*np.pi)

radii = np.arange(int(Rhole+4), int(Rsc-4), 2, dtype=int)  # circles that thread the hole but stay inside island
quanta = np.array([winding_complex(theta, int(r), cx, cy) for r in radii], dtype=float)
plateau = float(np.median(quanta)) if len(quanta)>0 else np.nan

# --- SCG invariants + geometry floor & FVR (global and inside SC island) ---
kappa = np.abs(lap2(chi))/(np.abs(chi)+EPS)
L_local = 1.0/np.sqrt(kappa+EPS)
L_metric = np.percentile(L_local, 10)       # metric floor (p10)
L_dyn = 1.0
L_geom = float(np.sqrt(L_dyn**2 + L_metric**2))
FVR_global = float(np.mean(L_local < L_metric))
FVR_SC     = float(np.mean(L_local[SCmask.astype(bool)] < L_metric)) if np.any(SCmask) else np.nan

# --- visuals ---
fig,axs = plt.subplots(2,3,figsize=(12,8))
axs[0,0].imshow(SCmask, cmap='gray_r'); axs[0,0].set_title("SC island (white) with hole (black)")
vb,vB = robust_vlims(Bz); im1 = axs[0,1].imshow(Bz, cmap='coolwarm', vmin=vb, vmax=vB)
axs[0,1].set_title('“B-like” $B_z$ — expelled; flux threads hole'); plt.colorbar(im1, ax=axs[0,1], shrink=0.8)
skip=5
axs[0,2].quiver(Jx[::skip,::skip], Jy[::skip,::skip], angles='xy', scale_units='xy', scale=1)
axs[0,2].set_title("$J=\\nabla\\theta$ (unit-scale)")

# winding curve
axs[1,0].plot(radii, quanta, lw=2)
axs[1,0].axhline(1.0, ls='--', color='k', alpha=0.5)
axs[1,0].axhline(-1.0, ls='--', color='k', alpha=0.3)
axs[1,0].set_xlabel("radius (px)"); axs[1,0].set_ylabel("winding (quanta)")
axs[1,0].set_title(f"Flux quantization around hole · plateau ≈ {plateau:.2f}")

# floor histogram + markers
axs[1,1].hist(L_local.ravel(), bins=60, color='#666')
axs[1,1].axvline(L_metric, color='r', ls='--', label='L_metric (p10)')
axs[1,1].axvline(L_geom,   color='g', ls='--', label='L_geom')
axs[1,1].legend(); axs[1,1].set_title("Coherence scales and geometry floor")

# metrics text
axs[1,2].axis('off')
axs[1,2].text(0.0, 1.0, (
    f"Plateau (hole) ≈ {plateau:.3f} quanta\n"
    f"L_metric ≈ {L_metric:.3f}\nL_geom ≈ {L_geom:.3f}\n"
    f"FVR_global = {FVR_global:.3f}\nFVR_SC = {FVR_SC:.3f}\n"
    f"Bz (2–98%) = [{vb:.2e}, {vB:.2e}]  |  ptp={np.ptp(Bz):.2e}"
), va='top', family='monospace', fontsize=11)

plt.tight_layout(); plt.show()
```

```python
# SWG/SCG Magnetism · Exp 5.3 — Supercurrent channel around hole (bulk Meissner + rim Dirichlet)
# Goal: allow a narrow annulus to carry circulation while bulk stays Meissner-suppressed.
# Uses 3 robust flux meters (complex increments, ∮∇θ·dl, Stokes ∫curlJ dA).

import numpy as np, matplotlib.pyplot as plt
EPS = 1e-12
rng = np.random.default_rng(5)

# --- grid ---
H=W=260
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)
R = np.hypot(X,Y) + 1e-6
ang = np.arctan2(Y,X).astype(np.float32)

# --- Neumann (edge-mirror) ops (no periodic wrap) ---
def pad_edge(Z): return np.pad(Z, ((1,1),(1,1)), mode='edge')
def grad2_nb(Z):
    Zp = pad_edge(Z)
    gx = 0.5*(Zp[1:-1,2:] - Zp[1:-1,0:-2])
    gy = 0.5*(Zp[2:,1:-1] - Zp[0:-2,1:-1])
    return gx, gy
def lap2_nb(Z):
    Zp = pad_edge(Z)
    return (Zp[1:-1,2:] + Zp[1:-1,0:-2] + Zp[2:,1:-1] + Zp[0:-2,1:-1] - 4*Z)
def div_nb(Jx, Jy):
    Jxp, Jyp = pad_edge(Jx), pad_edge(Jy)
    dJx = 0.5*(Jxp[1:-1,2:] - Jxp[1:-1,0:-2])
    dJy = 0.5*(Jyp[2:,1:-1] - Jyp[0:-2,1:-1])
    return dJx + dJy
def curl_from_J_nb(Jx, Jy):
    Jxp, Jyp = pad_edge(Jx), pad_edge(Jy)
    dJydx = 0.5*(Jyp[1:-1,2:] - Jyp[1:-1,0:-2])
    dJxdy = 0.5*(Jxp[2:,1:-1] - Jxp[0:-2,1:-1])
    return dJydx - dJxdy

def robust_vlims(Z, lo=2, hi=98):
    vmin, vmax = np.percentile(Z,[lo,hi])
    return vmin, (vmax if vmax>vmin else vmin+1e-9)

# --- Geometry: SC island with hole + supercurrent channel annulus ---
Rsc   = 60.0
Rhole =  9.0
chan_w = 8.0             # width of supercurrent channel
rim_w  = 2.0             # 2px clamp ring exactly at the hole

SC      = (R < Rsc).astype(np.float32)
HOLE    = (R < Rhole).astype(np.float32)
Rim     = np.abs(R - Rhole) <= rim_w
CHANNEL = np.logical_and(R >= Rhole+rim_w, R <= Rhole+rim_w+chan_w).astype(np.float32)
BULK    = np.logical_and(SC==1, np.logical_not(np.logical_or(HOLE==1, CHANNEL==1))).astype(np.float32)

# --- Fields & seed ---
m = -1  # target winding sign
theta = (0.002*X + 0.002*Y + 0.2*ang).astype(np.float32)  # slight tilt + tiny winding bias
A = (0.12*np.exp(-R**2/(2*(32.0**2))) + 0.01*rng.standard_normal((H,W))).astype(np.float32)
# Encourage current in the channel by raising A there (optional but helps visualization)
A = np.clip(A + 0.10*CHANNEL, 0, None)

# --- Meissner: suppress ∇θ only in BULK (not in CHANNEL) ---
lam = 2.2
def meissner_bulk(theta):
    gx, gy = grad2_nb(theta)
    gx *= BULK; gy *= BULK
    return -lam * div_nb(gx, gy)

# --- Evolution with rim clamp and channel free to carry current ---
dt=0.10; steps=220
def sat(u,a=0.18): return u/(1.0+a*u*u)
theta_target_rim = (m*ang).astype(np.float32)

for t in range(steps):
    # (cosmetic) amplitude smoothing
    A = np.clip(A + dt*(0.45*lap2_nb(A) - 0.02*A**3), 0, None)
    # phase dynamics: diffusion + BULK Meissner suppression
    dth = 0.33*lap2_nb(theta) + meissner_bulk(theta)
    theta = theta + dt*sat(dth,0.15)
    # clamp theta on the thin rim at the hole
    theta[Rim] = theta_target_rim[Rim]
    theta = (theta + np.pi)%(2*np.pi) - np.pi

# --- Fields & B-like ---
thx, thy = grad2_nb(theta)
Jx, Jy = thx, thy                     # for this core we use J = ∇θ
Bz = curl_from_J_nb(Jx, Jy)

# --- Robust flux meters in the channel band ---
def bilinear(Z, xf, yf):
    x0 = np.clip(np.floor(xf).astype(int), 0, W-2)
    y0 = np.clip(np.floor(yf).astype(int), 0, H-2)
    dx = xf - x0; dy = yf - y0
    z00 = Z[y0, x0];   z10 = Z[y0, x0+1]
    z01 = Z[y0+1, x0]; z11 = Z[y0+1, x0+1]
    return (z00*(1-dx)*(1-dy) + z10*dx*(1-dy) + z01*(1-dx)*dy + z11*dx*dy)

def winding_complex(theta, r, cx, cy, samples=4096):
    t = np.linspace(0, 2*np.pi, samples, endpoint=False)
    xf = cx + r*np.cos(t); yf = cy + r*np.sin(t)
    th = bilinear(theta, xf, yf)
    z = np.exp(1j*th)
    inc = np.angle(z[1:]*np.conj(z[:-1])); closure = np.angle(z[0]*np.conj(z[-1]))
    return float((np.sum(inc) + closure)/(2*np.pi))

def winding_grad(theta, r, cx, cy, samples=4096):
    t = np.linspace(0, 2*np.pi, samples, endpoint=False)
    xf = cx + r*np.cos(t); yf = cy + r*np.sin(t)
    gx = bilinear(thx, xf, yf); gy = bilinear(thy, xf, yf)
    tx, ty = -np.sin(t), np.cos(t); ds = r*(2*np.pi/samples)
    return float(np.sum((gx*tx + gy*ty)*ds)/(2*np.pi))

def winding_stokes(Bz, r, dr=1.5):
    yy, xx = np.mgrid[0:H,0:W]; rr = np.hypot(xx-cx, yy-cy)
    band = (rr >= (r-dr)) & (rr <= (r+dr))
    return float(np.sum(Bz[band])/(2*np.pi))

# Choose radii **inside the channel** only:
r_lo = int(Rhole + rim_w + 2)
r_hi = int(Rhole + rim_w + chan_w - 2)
radii = np.arange(r_lo, r_hi, 1, dtype=int)

q_c = np.array([winding_complex(theta, r, cx, cy) for r in radii])
q_g = np.array([winding_grad(theta,   r, cx, cy) for r in radii])
q_s = np.array([winding_stokes(Bz,    r, dr=1.5) for r in radii])

med_c, med_g, med_s = map(float, (np.median(q_c), np.median(q_g), np.median(q_s)))

# --- visuals ---
fig,axs = plt.subplots(2,3,figsize=(12,8))
axs[0,0].imshow(CHANNEL + 2*Rim + 3*BULK, cmap='viridis')
axs[0,0].set_title("Masks: channel (green), rim (yellow), bulk (blue)")
vb,vB = robust_vlims(Bz); im1 = axs[0,1].imshow(Bz, cmap='coolwarm', vmin=vb, vmax=vB)
axs[0,1].set_title('“B-like” $B_z$ (ring current visible)'); plt.colorbar(im1, ax=axs[0,1], shrink=0.8)
skip=5; axs[0,2].quiver(Jx[::skip,::skip], Jy[::skip,::skip], angles='xy', scale_units='xy', scale=1)
axs[0,2].set_title("$J=\\nabla\\theta$ — circulating in channel")

axs[1,0].plot(radii, q_c, label='complex'); axs[1,0].plot(radii, q_g, '--', label='∮∇θ·dl'); axs[1,0].plot(radii, q_s, ':', label='Stokes ∫curlJ dA')
axs[1,0].axhline(1.0, ls='--', color='k', alpha=0.4); axs[1,0].axhline(-1.0, ls='--', color='k', alpha=0.4)
axs[1,0].set_title(f"Medians in channel: {med_c:.2f}, {med_g:.2f}, {med_s:.2f}  (m={m:+d})"); axs[1,0].legend()
axs[1,0].set_xlabel("radius (px)"); axs[1,0].set_ylabel("winding (quanta)")

axs[1,1].imshow(Bz, cmap='coolwarm', vmin=vb, vmax=vB)
axs[1,1].add_patch(plt.Circle((cx,cy), r_lo, color='k', fill=False, lw=1))
axs[1,1].add_patch(plt.Circle((cx,cy), r_hi, color='k', fill=False, lw=1))
axs[1,1].set_title("Measurement band (inside channel)")

axs[1,2].axis('off')
axs[1,2].text(0.0,1.0, f"Plateaus ~ complex:{med_c:.3f}  grad:{med_g:.3f}  stokes:{med_s:.3f}\n"
                      f"Rhole={Rhole}, channel=[{r_lo},{r_hi}] px, m={m:+d}",
              va='top', family='monospace', fontsize=11)
plt.tight_layout(); plt.show()
```

```python
# SWG/SCG Magnetism · Exp 6 — Singularity-free UUD triad → coherence flow ("magnetism" without current)
# One cell. No phase singularities. Pure interference in a compressible medium.
# Key ideas:
#   - Build smooth loop forms as localized rotating compound waves (quadrature pair) with a small amplitude floor (ε) to avoid zeros.
#   - Define coherence-flow J from the complex field Ψ via J = Im(Ψ* · ∇Ψ)  (gauge-invariant, singularity-free with |Ψ|>0).
#   - Define B_like from curl of J. Report SCG-style metrics (entropy, variance of curvature proxy, floor-violation rate).

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(12)
EPS = 1e-6  # small floor to keep |Ψ|>0 everywhere

# --- grid ---
H=W=280
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)
R = np.hypot(X,Y).astype(np.float32)

# --- helpers (centered differences; periodic OK for this open scene) ---
def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)
def smooth(Z, n=2):
    V = Z.copy()
    for _ in range(n):
        V = (np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V = (np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V

# --- build one smooth loop form (no singularities)
# model: localized rotating "fin" via a quadrature pair (cos,sin) modulated by a Gaussian ring
def loop_form(x0, y0, sign=+1, radius=26.0, width=7.5, fin_ang=0.0, amp=1.0):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    # ring envelope (no zero: add a tiny floor)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    # single rotating fin ≈ cos(θ-θ0) component
    theta = np.arctan2(Yc, Xc)
    fin   = np.cos(theta - fin_ang)
    # quadrature rotating pair → complex wave; sign flips rotation sense
    base  = amp * env * (np.cos(fin) + 1j*sign*np.sin(fin))
    return base

# --- UUD triad placement (two "up", one "down" loop)
s = 32.0
centers = [(cx-s, cy), (cx+s, cy), (cx, cy+s*0.58)]
signs   = [+1, +1, -1]   # U, U, D
angles  = [0.0, 0.0, 0.0]

Psi = np.zeros((H,W), dtype=np.complex64)
for (x0,y0),sg,ang in zip(centers, signs, angles):
    Psi += loop_form(x0, y0, sign=sg, radius=26.0, width=8.0, fin_ang=ang, amp=1.0)

# small uniform floor to guarantee |Ψ|>0 (singularity-free)
Psi += EPS * (1.0 + 0.05*rng.standard_normal((H,W))).astype(np.complex64)

# --- coherence magnitude and phase-free current
# J = Im(Ψ* ∇Ψ)  (no arg/unwrap; smooth if |Ψ|>0). Normalize by |Ψ|^2 to get velocity-like field V.
Psix, Psiy = grad2(Psi)
Jx = np.imag(np.conj(Psi)*Psix)
Jy = np.imag(np.conj(Psi)*Psiy)
den = (np.abs(Psi)**2 + EPS)
Vx, Vy = Jx/den, Jy/den  # coherence-flow (velocity-like)

# magnetic-like proxy from coherence geometry
B_like = curl2(Jx, Jy)   # or curl of V; we’ll show curl(J) which emphasizes energy-bearing flow

# --- SCG-style metrics (coherence entropy, curvature proxy, floor-violation)
A = np.abs(Psi).astype(np.float32)
A_sm = smooth(A, n=2)
w = A_sm/(np.sum(A_sm)+1e-9)
Hc = float(-np.sum(w*np.log(w+1e-12)))   # coherence entropy

# curvature proxy κ from a smoothed "memory" χ = EMA(A); use laplacian magnitude / amplitude
def lap2(Z): 
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)
chi = smooth(A_sm, n=2)
kappa = np.abs(lap2(chi))/(np.abs(chi)+1e-9)
Kvar = float(np.var(kappa))

L_local = 1.0/np.sqrt(kappa+1e-9)
L_metric = float(np.percentile(L_local, 10))   # geometry floor (metric component)
L_geom = float(np.sqrt(1.0 + L_metric**2))
FVR = float(np.mean(L_local < L_metric))       # floor-violation rate

# --- visuals ---
fig,axs = plt.subplots(2,3,figsize=(12,8))

im0=axs[0,0].imshow(A_sm, cmap='magma'); axs[0,0].set_title("|Ψ| (coherence magnitude)")
plt.colorbar(im0, ax=axs[0,0], shrink=0.8)

skip=5
axs[0,1].quiver(x[::skip,::skip], y[::skip,::skip], Vx[::skip,::skip], Vy[::skip,::skip],
                angles='xy', scale_units='xy', scale=1)
axs[0,1].set_title("Coherence-flow V = J/|Ψ|² (unit-scale)")

vmin,vmax = np.percentile(B_like,[2,98]); vmax = max(vmax, vmin+1e-9)
im2=axs[0,2].imshow(B_like, cmap='coolwarm', vmin=vmin, vmax=vmax)
axs[0,2].set_title('“B-like” = curl J (from interference)'); plt.colorbar(im2, ax=axs[0,2], shrink=0.8)

axs[1,0].imshow(np.hypot(Vx,Vy), cmap='viridis'); axs[1,0].set_title("|V| (flow speed)")
axs[1,1].hist(L_local.ravel(), bins=60, color='#666')
axs[1,1].axvline(L_metric, color='r', ls='--', label='L_metric (p10)')
axs[1,1].axvline(L_geom,   color='g', ls='--', label='L_geom'); axs[1,1].legend()
axs[1,1].set_title("Coherence scales (geometry floor)")

axs[1,2].axis('off')
axs[1,2].text(0.0,1.0,(
    f"UUD triad (singularity-free)\n"
    f"H={Hc:.3f}  σ²κ={Kvar:.3e}  FVR={FVR:.3f}\n"
    f"L_metric≈{L_metric:.3f}  L_geom≈{L_geom:.3f}\n"
), va='top', family='monospace', fontsize=11)

plt.tight_layout(); plt.show()
```

```python
# SWG/SCG Magnetism · Exp 6.2 — Triad–Triad interaction (singularity-free UUD/DDU)
# One cell. Compare like vs opposite polarity. Outputs μ_z, midline B-energy, flow-overlap, SCG floor.

import numpy as np, matplotlib.pyplot as plt
rng = np.random.default_rng(101)
EPS = 1e-6

# --- grid & helpers ---
H=W=320
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)
def smooth(Z, n=2):
    V=Z.copy()
    for _ in range(n):
        V=(np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V=(np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V
def lap2(Z): 
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# smooth chiral loop (no zeros in |Ψ|)
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.0):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    th = np.arctan2(Yc, Xc)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr  = np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi = sign * twist_gain * fr * th
    return amp * env * np.exp(1j*phi)

def triad(P, signs, s=28.0, gain=3.0):
    (x0,y0) = P
    c = [(x0-s, y0), (x0+s, y0), (x0, y0+s*0.58)]
    Psi = 0j
    for (xx,yy),sg in zip(c, signs):
        Psi += loop_form_chiral(xx, yy, sign=sg, twist_gain=gain)
    return Psi

def metrics_from(Psi):
    Psix, Psiy = grad2(Psi)
    Jx = np.imag(np.conj(Psi)*Psix); Jy = np.imag(np.conj(Psi)*Psiy)
    A2 = (np.abs(Psi)**2 + EPS)
    Vx, Vy = Jx/A2, Jy/A2
    B = curl2(Jx, Jy)
    # magnetization-like moment about center
    mu = float(np.sum(X*Jy - Y*Jx))
    return Jx,Jy,Vx,Vy,B,mu

def scg_panel(A):
    A = A.astype(np.float32)
    chi = smooth(smooth(A,2),2)
    kappa = np.abs(lap2(chi))/(np.abs(chi)+1e-9)
    L_local = 1.0/np.sqrt(kappa+1e-9)
    L_metric = float(np.percentile(L_local,10))
    L_geom = float(np.sqrt(1.0 + L_metric**2))
    FVR = float(np.mean(L_local < L_metric))
    Hc = float(-np.sum((A/np.sum(A)+1e-12)*np.log(A/np.sum(A)+1e-12)))
    Kvar = float(np.var(kappa))
    return L_metric,L_geom,FVR,Hc,Kvar

def run_case(polarity2=+1, sep=84.0):
    # triad 1: UUD at left; triad 2: polarity2*(UUD) at right (i.e., UUD if +1, DDU if -1)
    P1 = (cx-sep/2, cy); P2 = (cx+sep/2, cy)
    t1 = [+1,+1,-1]
    t2 = [polarity2*+1, polarity2*+1, polarity2*-1]
    Psi = triad(P1, t1) + triad(P2, t2)
    Psi += EPS*(1.0 + 0.03*rng.standard_normal((H,W))).astype(np.complex64)
    Jx,Jy,Vx,Vy,B,mu = metrics_from(Psi)
    A = np.abs(Psi)
    L_metric,L_geom,FVR,Hc,Kvar = scg_panel(A)
    # interaction proxies
    mid = slice(int(cy-6), int(cy+6))
    B_mid_energy = float(np.mean(B[mid,:]**2))
    flow_overlap = float(np.mean(Vx[mid,:]**2 + Vy[mid,:]**2))  # stronger midline flow for attraction
    return dict(Psi=Psi,Jx=Jx,Jy=Jy,Vx=Vx,Vy=Vy,B=B,mu=mu,
                L_metric=L_metric,L_geom=L_geom,FVR=FVR,Hc=Hc,Kvar=Kvar,
                B_mid_energy=B_mid_energy, flow_overlap=flow_overlap)

case_like   = run_case(polarity2=+1)   # like–like
case_opp    = run_case(polarity2=-1)   # opposite

# --- visualize ---
def show_case(axs, case, title):
    A = np.abs(case['Psi'])
    im0=axs[0].imshow(A, cmap='magma'); axs[0].set_title(f"|Ψ| · {title}"); plt.colorbar(im0, ax=axs[0], shrink=0.8)
    vmin,vmax = np.percentile(case['B'],[2,98]); vmax = max(vmax, vmin+1e-9)
    im1=axs[1].imshow(case['B'], cmap='coolwarm', vmin=vmin, vmax=vmax); axs[1].set_title("B-like = curl J"); plt.colorbar(im1, ax=axs[1], shrink=0.8)
    skip=5
    axs[2].quiver(case['Vx'][::skip,::skip], case['Vy'][::skip,::skip], angles='xy', scale_units='xy', scale=0.14)
    axs[2].set_title("V = J/|Ψ|² (unit-scale)")
    axs[3].axis('off')
    axs[3].text(0.0,1.0,(
        f"μ_z≈{case['mu']:.3e}\n"
        f"B_mid⟨B²⟩≈{case['B_mid_energy']:.3e}\n"
        f"Flow_mid≈{case['flow_overlap']:.3e}\n"
        f"L_metric≈{case['L_metric']:.3f}  L_geom≈{case['L_geom']:.3f}\n"
        f"FVR={case['FVR']:.3f}  H={case['Hc']:.3f}  σ²κ={case['Kvar']:.3e}"
    ), va='top', family='monospace', fontsize=10)

fig,axs = plt.subplots(2,4,figsize=(12,7))
show_case(axs[0], case_like, "like–like (UUD + UUD)")
show_case(axs[1], case_opp,  "opposite (UUD + DDU)")
plt.tight_layout(); plt.show()

print("like–like  : μ_z = %.3e | mid ⟨B²⟩ = %.3e | flow_mid = %.3e" % (case_like['mu'], case_like['B_mid_energy'], case_like['flow_overlap']))
print("opposite   : μ_z = %.3e | mid ⟨B²⟩ = %.3e | flow_mid = %.3e" % (case_opp['mu'],  case_opp['B_mid_energy'],  case_opp['flow_overlap']))
```

```python
# SWG/SCG Magnetism · Exp 6.2b — Triad–Triad interaction curves vs separation (singularity-free)
# One cell: sweep separation, compare like (UUD+UUD) vs opposite (UUD+DDU) with contrastive metrics.

import numpy as np, matplotlib.pyplot as plt
rng = np.random.default_rng(202)
EPS = 1e-6

# --- grid & helpers (match 6.2) ---
H=W=320
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    th = np.arctan2(Yc, Xc)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr  = np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi = sign * twist_gain * fr * th
    return amp * env * np.exp(1j*phi)

def triad(P, signs, s=28.0, gain=3.2):
    (x0,y0) = P
    c = [(x0-s, y0), (x0+s, y0), (x0, y0+s*0.58)]
    Psi = 0j
    for (xx,yy),sg in zip(c, signs):
        Psi += loop_form_chiral(xx, yy, sign=sg, twist_gain=gain)
    return Psi

def fields(Psi):
    Psix, Psiy = grad2(Psi)
    Jx = np.imag(np.conj(Psi)*Psix); Jy = np.imag(np.conj(Psi)*Psiy)
    A2 = (np.abs(Psi)**2 + EPS)
    Vx, Vy = Jx/A2, Jy/A2
    B = curl2(Jx, Jy)
    return (Jx,Jy,Vx,Vy,B)

def midline_band(arr, h=6):
    return arr[int(cy-h):int(cy+h), :]

def run_pair(sep, polarity2):
    P1 = (cx-sep/2, cy); P2 = (cx+sep/2, cy)
    t1 = [+1,+1,-1]
    t2 = [polarity2*+1, polarity2*+1, polarity2*-1]
    Psi1 = triad(P1, t1); Psi2 = triad(P2, t2)
    Psi  = Psi1 + Psi2 + EPS*(1.0 + 0.02*rng.standard_normal((H,W))).astype(np.complex64)
    Jx,Jy,Vx,Vy,B = fields(Psi)
    # metrics
    B2_mid   = float(np.mean(midline_band(B)**2))
    flow_mid = float(np.mean(midline_band(Vx)**2 + midline_band(Vy)**2))
    # normalized cross-coherence on midline
    num = np.real(np.conj(Psi1)*Psi2)
    den = (np.abs(Psi1)*np.abs(Psi2) + EPS)
    xcoh = float(np.median(midline_band(num/den)))
    return B2_mid, flow_mid, xcoh

seps = np.array([56, 68, 80, 92, 104, 116, 128, 140], dtype=int)
like   = np.array([run_pair(d, +1) for d in seps])   # columns: B2_mid, flow_mid, xcoh
opp    = np.array([run_pair(d, -1) for d in seps])

dB2   = opp[:,0] - like[:,0]
dFlow = opp[:,1] - like[:,1]
dXcoh = opp[:,2] - like[:,2]

# --- plots ---
fig,axs = plt.subplots(1,3,figsize=(12,3.8))
axs[0].plot(seps, dB2, 'o-');   axs[0].set_title("Δ⟨B²⟩ (opp - like)")
axs[1].plot(seps, dFlow, 'o-'); axs[1].set_title("Δ flow_mid (opp - like)")
axs[2].plot(seps, dXcoh, 'o-'); axs[2].set_title("Δ cross-coherence (opp - like)")
for ax in axs:
    ax.axhline(0, color='#999', lw=1); ax.set_xlabel("separation (px)")
plt.tight_layout(); plt.show()

# tiny table
print("sep |  Δ⟨B²⟩      Δflow_mid    Δxcoh")
for d,db,df,dx in zip(seps,dB2,dFlow,dXcoh):
    print(f"{d:3d} | {db: .3e}  {df: .3e}  {dx: .3e}")
```

```python
# SWG/SCG Magnetism · Exp 6.2c — Per-triad polarity check + sign-sensitive interaction (singularity-free)
# One cell. Aligned (parallel) vs Anti-aligned (anti-parallel) triads; S_V, S_B, S_J vs separation.

import numpy as np, matplotlib.pyplot as plt
EPS = 1e-6
rng = np.random.default_rng(303)

# --- grid & helpers (match previous 6.x style) ---
H=W=320
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

# smooth, singularity-free chiral loop
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    th = np.arctan2(Yc, Xc)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr  = np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi = sign * twist_gain * fr * th
    return amp * env * np.exp(1j*phi)

def triad(P, signs, s=28.0, gain=3.2):
    (x0,y0) = P
    centers = [(x0-s, y0), (x0+s, y0), (x0, y0+s*0.58)]
    Psi = 0j
    for (xx,yy),sg in zip(centers, signs):
        Psi += loop_form_chiral(xx, yy, sign=sg, twist_gain=gain)
    return Psi

def fields_from(Psi):
    Psix, Psiy = grad2(Psi)
    Jx = np.imag(np.conj(Psi)*Psix)
    Jy = np.imag(np.conj(Psi)*Psiy)
    A2 = (np.abs(Psi)**2 + EPS)
    Vx, Vy = Jx/A2, Jy/A2
    B = curl2(Jx, Jy)
    return Jx,Jy,Vx,Vy,B

def moment_about(Psi, Px, Py):
    # μ_z = Σ ( (x-Px) Jy − (y-Py) Jx )
    Jx,Jy,_,_,_ = fields_from(Psi)
    Xr = (x-Px).astype(np.float32); Yr = (y-Py).astype(np.float32)
    return float(np.sum(Xr*Jy - Yr*Jx))

# midline band between triads (fixed height)
def mid_band(h=6):
    return slice(int(cy-h), int(cy+h)), slice(None)

def pair_case(sep, aligned=True):
    # left triad center P1, right triad center P2
    P1 = (cx-sep/2, cy); P2 = (cx+sep/2, cy)
    # base triad = UUD = [+1, +1, -1]
    t_base = [+1, +1, -1]
    t1 = t_base
    t2 = t_base if aligned else [-s for s in t_base]  # anti-aligned flips the chiral signs
    Psi1 = triad(P1, t1)
    Psi2 = triad(P2, t2)
    # small smooth floor
    floor = EPS*(1.0 + 0.02*rng.standard_normal((H,W))).astype(np.complex64)
    Psi1 = Psi1 + floor
    Psi2 = Psi2 + floor
    # per-triad moments about *their own* centers (sanity check)
    mu1 = moment_about(Psi1, *P1)
    mu2 = moment_about(Psi2, *P2)
    # separate fields for sign-sensitive observables
    J1x,J1y,V1x,V1y,B1 = fields_from(Psi1)
    J2x,J2y,V2x,V2y,B2 = fields_from(Psi2)
    # interaction observables on the midline band
    rr, cc = mid_band(h=6)
    S_V = float(np.mean(V1x[rr,cc]*V2x[rr,cc] + V1y[rr,cc]*V2y[rr,cc]))
    S_B = float(np.mean(B1[rr,cc]*B2[rr,cc]))
    S_J = float(np.mean(J1x[rr,cc]*J2x[rr,cc] + J1y[rr,cc]*J2y[rr,cc]))
    return dict(mu1=mu1, mu2=mu2, S_V=S_V, S_B=S_B, S_J=S_J)

# --- sweep separations for aligned vs anti-aligned ---
seps = np.array([56, 68, 80, 92, 104, 116, 128, 140], dtype=int)
aligned   = [pair_case(d, aligned=True)  for d in seps]
antialign = [pair_case(d, aligned=False) for d in seps]

# verify polarity: mu2 should ~equal mu1 for aligned, and ~−mu1 for anti-aligned
mu_chk = [(a['mu1'], a['mu2'], b['mu1'], b['mu2']) for a,b in zip(aligned, antialign)]

# build contrast curves: ΔS = anti-aligned − aligned (bridge > 0)
dSV = np.array([b['S_V'] - a['S_V'] for a,b in zip(aligned, antialign)])
dSB = np.array([b['S_B'] - a['S_B'] for a,b in zip(aligned, antialign)])
dSJ = np.array([b['S_J'] - a['S_J'] for a,b in zip(aligned, antialign)])

# --- plots ---
fig,axs = plt.subplots(1,3,figsize=(12,3.8))
axs[0].plot(seps, dSV, 'o-'); axs[0].axhline(0,color='#999'); axs[0].set_title("ΔS_V (anti − aligned)")
axs[1].plot(seps, dSB, 'o-'); axs[1].axhline(0,color='#999'); axs[1].set_title("ΔS_B (anti − aligned)")
axs[2].plot(seps, dSJ, 'o-'); axs[2].axhline(0,color='#999'); axs[2].set_title("ΔS_J (anti − aligned)")
for ax in axs: ax.set_xlabel("separation (px)")
plt.tight_layout(); plt.show()

# tiny summary table
print("sep |  μ1_algn     μ2_algn     μ1_anti     μ2_anti   |  ΔS_V        ΔS_B        ΔS_J")
for d,(m1a,m2a,m1n,m2n),dv,db,dj in zip(seps, mu_chk, dSV, dSB, dSJ):
    print(f"{d:3d} | {m1a: .3e} {m2a: .3e} {m1n: .3e} {m2n: .3e} | {dv: .3e} {db: .3e} {dj: .3e}")
```

```python
# SWG/SCG Magnetism · Exp 6.2d — Orientation × Separation sweep (singularity-free)
# Finds geometry where anti-aligned shows a clear midline "bridge" over aligned.

import numpy as np, matplotlib.pyplot as plt
EPS = 1e-6
rng = np.random.default_rng(404)

# --- grid ---
H=W=360
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X=(x-cx).astype(np.float32); Y=(y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

# smooth singularity-free chiral loop; angle 'rot' rotates the local triad layout
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    th = np.arctan2(Yc, Xc)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr  = np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi = sign * twist_gain * fr * th
    return amp * env * np.exp(1j*phi)

def rotate_pts(pts, origin, theta):
    ox, oy = origin
    c, s = np.cos(theta), np.sin(theta)
    out=[]
    for (px,py) in pts:
        dx, dy = px-ox, py-oy
        out.append((ox + c*dx - s*dy, oy + s*dx + c*dy))
    return out

def triad(center, signs, s=28.0, gain=3.2, rot=0.0):
    (x0,y0)=center
    base=[(x0-s, y0), (x0+s, y0), (x0, y0+s*0.58)]
    pts = rotate_pts(base, center, rot)
    Psi = 0j
    for (xx,yy),sg in zip(pts, signs):
        Psi += loop_form_chiral(xx, yy, sign=sg, amp=1.0, twist_gain=gain)
    return Psi

def fields(Psi):
    Psix, Psiy = grad2(Psi)
    Jx = np.imag(np.conj(Psi)*Psix); Jy = np.imag(np.conj(Psi)*Psiy)
    A2 = (np.abs(Psi)**2 + EPS)
    Vx, Vy = Jx/A2, Jy/A2
    B = curl2(Jx, Jy)
    return Jx,Jy,Vx,Vy,B

def adaptive_bridge_band(P1, P2, thickness=7):
    # band centered on the segment connecting triad centers, axis-aligned bbox mask + distance to segment
    (x1,y1),(x2,y2) = P1,P2
    # vector along segment
    vx, vy = x2-x1, y2-y1
    L = np.hypot(vx,vy)+1e-9
    ux, uy = vx/L, vy/L
    # vector from P1 to every pixel
    RX, RY = x-x1, y-y1
    # projection length along segment
    t = RX*ux + RY*uy
    # clamp to segment
    t_clamped = np.clip(t, 0, L)
    # closest point on segment
    CX = x1 + ux*t_clamped; CY = y1 + uy*t_clamped
    dist = np.hypot(x-CX, y-CY)
    return dist <= thickness

def pair_scores(sep, theta2):
    P1 = (cx-sep/2, cy); P2 = (cx+sep/2, cy)
    signs = [+1,+1,-1]  # UUD
    # aligned (parallel)
    Psi1_a = triad(P1, signs, rot=0.0)
    Psi2_a = triad(P2, signs, rot=theta2)
    # anti-aligned (anti-parallel chiralities, same rotation)
    Psi1_n = Psi1_a
    Psi2_n = triad(P2, [-s for s in signs], rot=theta2)
    # small floor to keep |Ψ|>0
    floor = EPS*(1.0 + 0.02*rng.standard_normal((H,W))).astype(np.complex64)
    Psi1_a += floor; Psi2_a += floor; Psi1_n += floor; Psi2_n += floor

    J1x,J1y,V1x,V1y,B1 = fields(Psi1_a)
    J2x,J2y,V2x,V2y,B2 = fields(Psi2_a)
    K1x,K1y,W1x,W1y,D1 = fields(Psi1_n)
    K2x,K2y,W2x,W2y,D2 = fields(Psi2_n)

    band = adaptive_bridge_band(P1,P2, thickness=7)

    def mean_band(arr): return float(np.mean(arr[band]))

    # aligned scores
    S_V_a = mean_band(V1x*V2x + V1y*V2y)
    S_B_a = mean_band(B1*B2)
    S_J_a = mean_band(J1x*J2x + J1y*J2y)
    # anti-aligned scores
    S_V_n = mean_band(W1x*W2x + W1y*W2y)
    S_B_n = mean_band(D1*D2)
    S_J_n = mean_band(K1x*K2x + K1y*K2y)

    return (S_V_n - S_V_a, S_B_n - S_B_a, S_J_n - S_J_a)  # Δ = anti − aligned

# sweep
seps   = np.array([56, 68, 80, 92, 104, 116, 128, 140], dtype=int)
angles = np.deg2rad(np.linspace(-60, +60, 11))  # rotate second triad
dSV = np.zeros((len(angles), len(seps)))
dSB = np.zeros_like(dSV); dSJ = np.zeros_like(dSV)

for i,th in enumerate(angles):
    for j,d in enumerate(seps):
        dSV[i,j], dSB[i,j], dSJ[i,j] = pair_scores(d, th)

# visualize heatmaps
fig,axs = plt.subplots(1,3,figsize=(13,3.8))
im0=axs[0].imshow(dSV, cmap='coolwarm', origin='lower',
                  extent=[seps.min(), seps.max(), np.rad2deg(angles.min()), np.rad2deg(angles.max())],
                  aspect='auto'); axs[0].set_title("ΔS_V (anti − aligned)")
plt.colorbar(im0, ax=axs[0], shrink=0.8)
im1=axs[1].imshow(dSB, cmap='coolwarm', origin='lower',
                  extent=[seps.min(), seps.max(), np.rad2deg(angles.min()), np.rad2deg(angles.max())],
                  aspect='auto'); axs[1].set_title("ΔS_B (anti − aligned)")
plt.colorbar(im1, ax=axs[1], shrink=0.8)
im2=axs[2].imshow(dSJ, cmap='coolwarm', origin='lower',
                  extent=[seps.min(), seps.max(), np.rad2deg(angles.min()), np.rad2deg(angles.max())],
                  aspect='auto'); axs[2].set_title("ΔS_J (anti − aligned)")
plt.colorbar(im2, ax=axs[2], shrink=0.8)
for ax in axs:
    ax.set_xlabel("separation (px)"); ax.set_ylabel("rotation of triad 2 (deg)")
plt.tight_layout(); plt.show()

# print top-k geometry configurations where anti > aligned
k=6
flat_idx = np.dstack(np.unravel_index(np.argsort(dSV.ravel())[::-1], dSV.shape))[0]  # by ΔS_V
print("Top by ΔS_V (anti − aligned):")
for r,c in flat_idx[:k]:
    print(f"sep={seps[c]:3d}px  rot={np.rad2deg(angles[r]):+5.1f}°  ΔS_V={dSV[r,c]: .3e}  ΔS_B={dSB[r,c]: .3e}  ΔS_J={dSJ[r,c]: .3e}")
```

```python
# SWG/SCG Magnetism · Exp 7.0 — Small-patch M(b) sweep (singularity-free, reversible baseline)
# Tiled UUD triads; smooth chirality; sweep a tiny global bias 'b' and read M, ⟨B²⟩, SCG metrics.

import numpy as np, matplotlib.pyplot as plt

EPS = 1e-6
rng = np.random.default_rng(707)

# --- domain & helpers (non-singular, periodic diffs are fine for this patch) ---
H=W=520
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X = (x-cx).astype(np.float32); Y = (y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

def smooth(Z, n=2):
    V = Z.copy()
    for _ in range(n):
        V = (np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V = (np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V

def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# --- singularity-free chiral loop; bias b adds a smooth extra twist in the ring (same sign everywhere) ---
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2, bias=0.0):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    th = np.arctan2(Yc, Xc)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3     # tiny floor → |Ψ|>0
    fr  = np.exp(-((r-radius)**2)/(2*(1.2*width)**2))      # radial window for twist/bias
    phi = (sign*twist_gain + bias) * fr * th               # smooth, single-valued twist
    return amp * env * np.exp(1j*phi)

def rotate_pts(pts, origin, theta):
    ox,oy = origin
    c,s = np.cos(theta), np.sin(theta)
    out=[]
    for (px,py) in pts:
        dx,dy = px-ox, py-oy
        out.append((ox + c*dx - s*dy, oy + s*dx + c*dy))
    return out

def triad(center, signs, s=28.0, gain=3.2, rot=0.0, bias=0.0):
    (x0,y0) = center
    base=[(x0-s, y0), (x0+s, y0), (x0, y0+s*0.58)]
    pts = rotate_pts(base, center, rot)
    Psi = 0j
    for (xx,yy),sg in zip(pts, signs):
        Psi += loop_form_chiral(xx, yy, sign=sg, twist_gain=gain, bias=bias)
    return Psi

def fields_from(Psi):
    Psix, Psiy = grad2(Psi)
    Jx = np.imag(np.conj(Psi)*Psix); Jy = np.imag(np.conj(Psi)*Psiy)
    A2 = (np.abs(Psi)**2 + EPS)
    Vx, Vy = Jx/A2, Jy/A2
    B  = curl2(Jx, Jy)
    return Jx, Jy, Vx, Vy, B

def scg_metrics(A):
    A = A.astype(np.float32)
    chi = smooth(smooth(A,2),2)
    kappa = np.abs(lap2(chi))/(np.abs(chi)+1e-9)
    L_local = 1.0/np.sqrt(kappa+1e-9)
    L_metric = float(np.percentile(L_local, 10))
    L_geom   = float(np.sqrt(1.0 + L_metric**2))
    FVR      = float(np.mean(L_local < L_metric))
    Hc = float(-np.sum((A/np.sum(A)+1e-12)*np.log(A/np.sum(A)+1e-12)))
    Kvar = float(np.var(kappa))
    return L_metric, L_geom, FVR, Hc, Kvar

# --- lattice of triads (UUD) at validated spacing/orientation ---
sep = 68.0                       # from 6.2e best geometry
rot2 = np.deg2rad(60.0)          # orientation used in the sweep
signs = [+1,+1,-1]               # UUD

# tile centers
nx, ny = 4, 3
gx0, gy0 = cx - (nx-1)*sep/2, cy - (ny-1)*sep/2
centers=[]
for j in range(ny):
    for i in range(nx):
        centers.append((gx0 + i*sep, gy0 + j*sep))

def build_patch(bias):
    Psi = 0j
    for k,(px,py) in enumerate(centers):
        # alternate a small rotation pattern to break symmetry (checkerboard)
        rot = rot2 if (k%2==0) else -rot2
        Psi += triad((px,py), signs, rot=rot, bias=bias)
    # tiny global floor to keep |Ψ|>0 strictly
    Psi += EPS*(1.0 + 0.02*rng.standard_normal((H,W))).astype(np.complex64)
    return Psi

def measure_state(Psi):
    Jx,Jy,Vx,Vy,B = fields_from(Psi)
    A = np.abs(Psi)
    # magnetization-like moment about patch center
    mu = float(np.sum(X*Jy - Y*Jx))
    B2 = float(np.mean(B*B))
    L_metric, L_geom, FVR, Hc, Kvar = scg_metrics(A)
    return dict(mu=mu, B2=B2, L_metric=L_metric, L_geom=L_geom, FVR=FVR, Hc=Hc, Kvar=Kvar,
                Jx=Jx, Jy=Jy, Vx=Vx, Vy=Vy, B=B, A=A)

# --- bias sweep (forward and back) ---
bmax = 0.6
grid = np.linspace(-bmax, bmax, 17)
sweep = np.concatenate([grid, grid[::-1][1:]])  # go and return (reversible baseline)

records=[]
for b in sweep:
    Psi = build_patch(bias=b)
    rec = measure_state(Psi)
    rec['b'] = float(b)
    records.append(rec)

# collect curves
b_vals = np.array([r['b'] for r in records])
M_vals = np.array([r['mu'] for r in records])
B2_vals= np.array([r['B2'] for r in records])
FVR_vals=np.array([r['FVR'] for r in records])
Lmet_vals=np.array([r['L_metric'] for r in records])
Lgeo_vals=np.array([r['L_geom'] for r in records])

# --- plots: M(b), ⟨B²⟩(b), SCG metrics; plus a snapshot panel at b≈0 and b≈±bmax ---
fig,axs = plt.subplots(2,3,figsize=(13,8))

axs[0,0].plot(b_vals, M_vals, '-o', ms=3)
axs[0,0].set_title("M(b) = μ_z vs bias (reversible baseline)")
axs[0,0].set_xlabel("bias b"); axs[0,0].set_ylabel("μ_z (arb)")

axs[0,1].plot(b_vals, B2_vals, '-o', ms=3); axs[0,1].set_title("⟨B²⟩ vs b"); axs[0,1].set_xlabel("bias b"); axs[0,1].set_ylabel("⟨B²⟩")

axs[0,2].plot(b_vals, Lmet_vals, label='L_metric (p10)')
axs[0,2].plot(b_vals, Lgeo_vals, label='L_geom')
axs[0,2].plot(b_vals, FVR_vals, label='FVR')
axs[0,2].legend(); axs[0,2].set_title("SCG metrics vs b"); axs[0,2].set_xlabel("bias b")

# choose frames near b≈0 and |b|≈bmax for context
idx0 = np.argmin(np.abs(b_vals-0))
idxP = np.argmax(b_vals)  # +bmax
idxN = np.argmin(b_vals)  # -bmax

def show_state(axs_row, rec, title):
    A = rec['A']; B = rec['B']; Vx, Vy = rec['Vx'], rec['Vy']
    im0=axs_row[0].imshow(A, cmap='magma'); axs_row[0].set_title(f"|Ψ| · {title}"); plt.colorbar(im0, ax=axs_row[0], shrink=0.7)
    vmin,vmax = np.percentile(B,[2,98]); vmax = max(vmax, vmin+1e-9)
    im1=axs_row[1].imshow(B, cmap='coolwarm', vmin=vmin, vmax=vmax); axs_row[1].set_title("B-like = curl J"); plt.colorbar(im1, ax=axs_row[1], shrink=0.7)
    skip=6
    axs_row[2].quiver(Vx[::skip,::skip], Vy[::skip,::skip], angles='xy', scale_units='xy', scale=0.16)
    axs_row[2].set_title("V field (unit-scale)")

show_state(axs[1], records[idx0], f"b≈{b_vals[idx0]:+.2f}")
plt.tight_layout(); plt.show()

print(f"Summary at b=0:   μ_z={M_vals[idx0]:.3e}  ⟨B²⟩={B2_vals[idx0]:.3e}  FVR={FVR_vals[idx0]:.3f}  "
      f"L_metric={Lmet_vals[idx0]:.3f}  L_geom={Lgeo_vals[idx0]:.3f}")
print(f"Summary at +bmax: μ_z={M_vals[idxP]:.3e}  ⟨B²⟩={B2_vals[idxP]:.3e}  FVR={FVR_vals[idxP]:.3f}  "
      f"L_metric={Lmet_vals[idxP]:.3f}  L_geom={Lgeo_vals[idxP]:.3f}")
print(f"Summary at -bmax: μ_z={M_vals[idxN]:.3e}  ⟨B²⟩={B2_vals[idxN]:.3e}  FVR={FVR_vals[idxN]:.3f}  "
      f"L_metric={Lmet_vals[idxN]:.3f}  L_geom={Lgeo_vals[idxN]:.3f}")
```

```python
# SWG/SCG Magnetism · Exp 7.1 — Hysteresis with χ-memory (singularity-free)
# Adds a slow memory χ that biases chirality to create pinning & path-dependence.

import numpy as np, matplotlib.pyplot as plt

EPS = 1e-6
rng = np.random.default_rng(711)

# --- domain & helpers (same as 7.0) ---
H=W=520
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X = (x-cx).astype(np.float32); Y = (y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy

def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)

def smooth(Z, n=2):
    V = Z.copy()
    for _ in range(n):
        V = (np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V = (np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V

def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# --- loop & triad builders; *bias_eff* injects both external bias b and memory χ ---
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2, bias_eff=0.0):
    Xc = X - (x0-cx); Yc = Y - (y0-cy)
    r  = np.hypot(Xc, Yc)
    th = np.arctan2(Yc, Xc)
    env = np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3      # |Ψ|>0
    fr  = np.exp(-((r-radius)**2)/(2*(1.2*width)**2))       # ring window
    phi = (sign*twist_gain + bias_eff) * fr * th            # smooth, single-valued twist
    return amp * env * np.exp(1j*phi)

def rotate_pts(pts, origin, theta):
    ox,oy = origin; c,s = np.cos(theta), np.sin(theta)
    out=[]
    for (px,py) in pts:
        dx,dy = px-ox, py-oy
        out.append((ox + c*dx - s*dy, oy + s*dx + c*dy))
    return out

def triad(center, signs, s=28.0, gain=3.2, rot=0.0, bias_eff=0.0):
    (x0,y0) = center
    base=[(x0-s, y0), (x0+s, y0), (x0, y0+s*0.58)]
    pts = rotate_pts(base, center, rot)
    Psi = 0j
    for (xx,yy),sg in zip(pts, signs):
        Psi += loop_form_chiral(xx, yy, sign=sg, twist_gain=gain, bias_eff=bias_eff)
    return Psi

def fields_from(Psi):
    Psix, Psiy = grad2(Psi)
    Jx = np.imag(np.conj(Psi)*Psix); Jy = np.imag(np.conj(Psi)*Psiy)
    A2 = (np.abs(Psi)**2 + EPS)
    Vx, Vy = Jx/A2, Jy/A2
    B  = curl2(Jx, Jy)
    return Jx, Jy, Vx, Vy, B

def scg_metrics(A):
    A = A.astype(np.float32)
    chi = smooth(smooth(A,2),2)
    kappa = np.abs(lap2(chi))/(np.abs(chi)+1e-9)
    L_local = 1.0/np.sqrt(kappa+1e-9)
    L_metric = float(np.percentile(L_local, 10))
    L_geom   = float(np.sqrt(1.0 + L_metric**2))
    FVR      = float(np.mean(L_local < L_metric))
    return L_metric, L_geom, FVR

# --- patch geometry (as in 7.0) ---
sep = 68.0
rot2 = np.deg2rad(60.0)
signs = [+1,+1,-1]
nx, ny = 4, 3
gx0, gy0 = cx - (nx-1)*sep/2, cy - (ny-1)*sep/2
centers=[(gx0 + i*sep, gy0 + j*sep) for j in range(ny) for i in range(nx)]

# --- χ-memory parameters ---
eta = 0.10   # memory update rate (smaller → more hysteresis)
k_mem = 0.80 # memory feedback gain into effective twist
R_gain = 0.65# how strongly curl(J) imprints into χ (through tanh)
chi_mem = np.zeros((H,W), np.float32)

def build_with_memory(b):
    # effective bias is external bias + memory projection (windowed + limited)
    chi_win = smooth(chi_mem, n=2)
    bias_eff_field = k_mem * np.tanh(R_gain * chi_win) + b
    # sample a single scalar bias per loop by local averaging around each center (cheap & smooth)
    Psi = 0j
    for k,(px,py) in enumerate(centers):
        # sample bias_eff near this triad center
        ix,iy = int(round(px)), int(round(py))
        patch = bias_eff_field[max(iy-4,0):iy+5, max(ix-4,0):ix+5]
        b_eff = float(np.mean(patch))
        rot =  rot2 if (k%2==0) else -rot2
        Psi += triad((px,py), signs, rot=rot, bias_eff=b_eff)
    Psi += EPS*(1.0 + 0.02*rng.standard_normal((H,W))).astype(np.complex64)
    return Psi

def measure(Psi):
    Jx,Jy,Vx,Vy,B = fields_from(Psi)
    A = np.abs(Psi)
    mu = float(np.sum(X*Jy - Y*Jx))
    B2 = float(np.mean(B*B))
    Lm, Lg, FVR = scg_metrics(A)
    return dict(mu=mu, B2=B2, Lm=Lm, Lg=Lg, FVR=FVR, Jx=Jx, Jy=Jy, Vx=Vx, Vy=Vy, B=B, A=A)

def update_memory(B_like):
    global chi_mem
    # χ integrates the rotational density (smoothed), slow time-constant
    imprint = smooth(B_like, n=2)
    chi_mem = (1.0-eta)*chi_mem + eta*imprint

# --- sweep forward and back with memory-coupled states ---
bmax = 0.6
grid = np.linspace(-bmax, bmax, 17)
sweep_fwd = grid
sweep_bwd = grid[::-1]

rec_fwd, rec_bwd = [], []

# forward branch
for b in sweep_fwd:
    Psi = build_with_memory(b)
    rec = measure(Psi); rec['b']=float(b); rec_fwd.append(rec)
    update_memory(rec['B'])  # update χ after measuring

# backward branch (memory carries over)
for b in sweep_bwd:
    Psi = build_with_memory(b)
    rec = measure(Psi); rec['b']=float(b); rec_bwd.append(rec)
    update_memory(rec['B'])

# curves
bf = np.array([r['b']  for r in rec_fwd]);  Mf = np.array([r['mu'] for r in rec_fwd])
bb = np.array([r['b']  for r in rec_bwd]);  Mb = np.array([r['mu'] for r in rec_bwd])
B2f= np.array([r['B2'] for r in rec_fwd]);  B2b= np.array([r['B2'] for r in rec_bwd])
FVRf=np.array([r['FVR']for r in rec_fwd]);  FVRb=np.array([r['FVR']for r in rec_bwd])

# remanence (M at b≈0 on return) & coercive biases (b where M crosses 0)
def interp_cross(b, M):
    # simple zero-crossing estimate
    idx = np.where(np.sign(M[:-1])!=np.sign(M[1:]))[0]
    if len(idx)==0: return np.nan
    i=idx[0]; t = -M[i]/(M[i+1]-M[i]+1e-12)
    return float(b[i] + t*(b[i+1]-b[i]))
bHc_f = interp_cross(bf, Mf)
bHc_b = interp_cross(bb, Mb)

# find M at b closest to 0 on each branch
i0_f = np.argmin(np.abs(bf)); Mrem_f = float(Mf[i0_f])
i0_b = np.argmin(np.abs(bb)); Mrem_b = float(Mb[i0_b])

# --- plots ---
fig,axs = plt.subplots(2,3,figsize=(13,8))
axs[0,0].plot(bf, Mf, '-o', ms=3, label='forward')
axs[0,0].plot(bb, Mb, '-o', ms=3, label='backward')
axs[0,0].axhline(0,color='#777',lw=1)
axs[0,0].set_title("M–H curve with χ-memory"); axs[0,0].set_xlabel("bias b"); axs[0,0].set_ylabel("μ_z"); axs[0,0].legend()

axs[0,1].plot(bf, B2f, '-o', ms=3, label='⟨B²⟩ fwd')
axs[0,1].plot(bb, B2b, '-o', ms=3, label='⟨B²⟩ bwd'); axs[0,1].legend()
axs[0,1].set_title("⟨B²⟩ vs b (with memory)"); axs[0,1].set_xlabel("bias b"); axs[0,1].set_ylabel("⟨B²⟩")

axs[0,2].plot(bf, FVRf, label='FVR fwd'); axs[0,2].plot(bb, FVRb, label='FVR bwd'); axs[0,2].legend()
axs[0,2].set_title("FVR vs b (floor respected)"); axs[0,2].set_xlabel("bias b")

# snapshots near extremes & zero on backward branch
idxP = np.argmax(bb); idxN = np.argmin(bb); idx0 = i0_b
for col, idx, title in zip(range(3), [idxN, idx0, idxP], ["b≈-bmax (bwd)", "b≈0 (bwd)", "b≈+bmax (bwd)"]):
    A = rec_bwd[idx]['A']; B = rec_bwd[idx]['B']; Vx,Vy = rec_bwd[idx]['Vx'],rec_bwd[idx]['Vy']
    im0=axs[1,col].imshow(B, cmap='coolwarm', vmin=np.percentile(B,2), vmax=np.percentile(B,98))
    axs[1,col].set_title(title+" · B-like"); plt.colorbar(im0, ax=axs[1,col], shrink=0.7)

plt.tight_layout(); plt.show()

print(f"Remanence (M@b≈0): forward {Mrem_f:.3e} | backward {Mrem_b:.3e}")
print(f"Coercive biases (b where M crosses 0): forward {bHc_f:.3f} | backward {bHc_b:.3f}")
```

```python
# SWG/SCG Magnetism · Exp 7.2 — Hysteresis with χ-memory + sparse pinning (singularity-free)
# Adds a weak, random pinning field that resists local chirality changes; sweeps b and shows M–H with coercivity.

import numpy as np, matplotlib.pyplot as plt
EPS = 1e-6
rng = np.random.default_rng(712)

# --- domain & helpers (same as 7.1) ---
H=W=520
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X = (x-cx).astype(np.float32); Y = (y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def curl2(Jx,Jy): return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)
def smooth(Z, n=2):
    V=Z.copy()
    for _ in range(n):
        V=(np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V=(np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V
def lap2(Z): 
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# loop & triad builders; bias_eff now includes memory + pinning
def loop_form_chiral(x0,y0,sign=+1,radius=24.0,width=7.5,amp=1.0,twist_gain=3.2,bias_eff=0.0):
    Xc=X-(x0-cx); Yc=Y-(y0-cy); r=np.hypot(Xc,Yc); th=np.arctan2(Yc,Xc)
    env=np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr =np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi=(sign*twist_gain + bias_eff)*fr*th
    return amp*env*np.exp(1j*phi)

def rotate_pts(pts, origin, theta):
    ox,oy=origin; c,s=np.cos(theta),np.sin(theta); out=[]
    for px,py in pts:
        dx,dy=px-ox,py-oy
        out.append((ox+c*dx-s*dy, oy+s*dx+c*dy))
    return out

def triad(center, signs, s=28.0, gain=3.2, rot=0.0, bias_eff=0.0):
    (x0,y0)=center; base=[(x0-s,y0),(x0+s,y0),(x0,y0+s*0.58)]
    pts=rotate_pts(base, center, rot)
    Psi=0j
    for (xx,yy),sg in zip(pts, signs):
        Psi+=loop_form_chiral(xx,yy,sign=sg,twist_gain=gain,bias_eff=bias_eff)
    return Psi

def fields_from(Psi):
    Psix,Psiy=grad2(Psi)
    Jx=np.imag(np.conj(Psi)*Psix); Jy=np.imag(np.conj(Psi)*Psiy)
    A2=(np.abs(Psi)**2+EPS); Vx,Jy_over=A2,A2
    Vx=Jx/A2; Vy=Jy/A2
    B=curl2(Jx,Jy)
    return Jx,Jy,Vx,Vy,B

def scg_metrics(A):
    A=A.astype(np.float32)
    chi=smooth(smooth(A,2),2)
    kappa=np.abs(lap2(chi))/(np.abs(chi)+1e-9)
    L_local=1.0/np.sqrt(kappa+1e-9)
    L_metric=float(np.percentile(L_local,10))
    L_geom=float(np.sqrt(1.0+L_metric**2))
    FVR=float(np.mean(L_local<L_metric))
    return L_metric,L_geom,FVR

# patch geometry (same as 7.0/7.1)
sep=68.0; rot2=np.deg2rad(60.0); signs=[+1,+1,-1]
nx,ny=4,3
gx0,gy0=cx-(nx-1)*sep/2, cy-(ny-1)*sep/2
centers=[(gx0+i*sep, gy0+j*sep) for j in range(ny) for i in range(nx)]

# --- χ-memory + sparse pinning ---
eta   = 0.10   # memory update rate
k_mem = 0.90   # memory gain
R_gain= 0.70   # curl->memory sensitivity
chi_mem = np.zeros((H,W), np.float32)

# pinning field: sparse gaussian bumps that resist twist change (acts like local threshold)
pin = np.zeros((H,W), np.float32)
num_pins = 18
for _ in range(num_pins):
    px,py = rng.integers(40,W-40), rng.integers(40,H-40)
    sx,sy = rng.uniform(6,12), rng.uniform(6,12)
    gx = np.exp(-((x-px)**2)/(2*sx*sx))
    gy = np.exp(-((y-py)**2)/(2*sy*sy))
    pin += 0.8*gx*gy
pin = smooth(pin, n=3) / (np.max(pin)+1e-9)  # 0..1

pin_gain = 0.35   # strength with which pinning modulates effective bias (resists changes)

def build_with_memory_and_pinning(b, prev_sign=+1):
    # memory projection
    chi_win = smooth(chi_mem, n=2)
    mem_field = k_mem * np.tanh(R_gain * chi_win)
    # pinning opposes motion: subtract previous direction scaled by pin
    # We don't track domain history; approximate by opposing current bias sign.
    pin_field = -pin_gain * np.sign(b) * pin
    bias_eff_field = mem_field + b + pin_field

    Psi=0j
    for k,(px,py) in enumerate(centers):
        ix,iy = int(round(px)), int(round(py))
        patch = bias_eff_field[max(iy-4,0):iy+5, max(ix-4,0):ix+5]
        b_eff = float(np.mean(patch))
        rot = rot2 if (k%2==0) else -rot2
        Psi += triad((px,py), signs, rot=rot, bias_eff=b_eff)
    Psi += EPS*(1.0+0.02*rng.standard_normal((H,W))).astype(np.complex64)
    return Psi

def measure(Psi):
    Jx,Jy,Vx,Vy,B = fields_from(Psi)
    A=np.abs(Psi)
    mu = float(np.sum(X*Jy - Y*Jx))
    B2 = float(np.mean(B*B))
    Lm,Lg,FVR = scg_metrics(A)
    return dict(mu=mu,B2=B2,Lm=Lm,Lg=Lg,FVR=FVR,B=B,A=A)

def update_memory(B_like):
    global chi_mem
    imprint = smooth(B_like, n=2)
    chi_mem = (1.0-eta)*chi_mem + eta*imprint

# sweep with memory + pinning
bmax=0.6
grid=np.linspace(-bmax, bmax, 17)
bf=grid; bb=grid[::-1]
rec_f, rec_b = [], []

for b in bf:
    Psi=build_with_memory_and_pinning(b)
    r=measure(Psi); r['b']=float(b); rec_f.append(r)
    update_memory(r['B'])

for b in bb:
    Psi=build_with_memory_and_pinning(b)
    r=measure(Psi); r['b']=float(b); rec_b.append(r)
    update_memory(r['B'])

# curves & coercivity, remanence
def extract(rec):
    b=np.array([r['b'] for r in rec])
    M=np.array([r['mu'] for r in rec])
    B2=np.array([r['B2'] for r in rec]); F=np.array([r['FVR'] for r in rec])
    return b,M,B2,F
bf,Mf,B2f,Ff = extract(rec_f); bb,Mb,B2b,Fb = extract(rec_b)

def zero_cross(b,M):
    idx=np.where(np.sign(M[:-1])!=np.sign(M[1:]))[0]
    if len(idx)==0: return np.nan
    i=idx[0]; t = -M[i]/(M[i+1]-M[i]+1e-12)
    return float(b[i] + t*(b[i+1]-b[i]))

bHc_f, bHc_b = zero_cross(bf,Mf), zero_cross(bb,Mb)
Mrem_f = float(Mf[np.argmin(np.abs(bf))])
Mrem_b = float(Mb[np.argmin(np.abs(bb))])

# differential (Barkhausen-like)
def diff_curve(b,M):
    return np.gradient(M, b)
dMf, dMb = diff_curve(bf,Mf), diff_curve(bb,Mb)

# plots
fig,axs=plt.subplots(2,3,figsize=(13,8))
axs[0,0].plot(bf,Mf,'-o',ms=3,label='forward'); axs[0,0].plot(bb,Mb,'-o',ms=3,label='backward')
axs[0,0].axhline(0,color='#888'); axs[0,0].set_title("M–H with χ-memory + pinning"); axs[0,0].set_xlabel("bias b"); axs[0,0].set_ylabel("μ_z"); axs[0,0].legend()

axs[0,1].plot(bf,dMf,'-o',ms=3,label='dM/db fwd'); axs[0,1].plot(bb,dMb,'-o',ms=3,label='dM/db bwd'); axs[0,1].legend()
axs[0,1].set_title("dM/db (Barkhausen-like)"); axs[0,1].set_xlabel("bias b")

axs[0,2].plot(bf,Ff,label='FVR fwd'); axs[0,2].plot(bb,Fb,label='FVR bwd'); axs[0,2].legend()
axs[0,2].set_title("FVR vs b"); axs[0,2].set_xlabel("bias b")

# snapshots
def show_state(ax, rec, title):
    B=rec['B']; vmin,vmax=np.percentile(B,[2,98]); vmax=max(vmax,vmin+1e-9)
    im=ax.imshow(B,cmap='coolwarm',vmin=vmin,vmax=vmax); ax.set_title(title); plt.colorbar(im,ax=ax,shrink=0.7)

show_state(axs[1,0], rec_f[0],  "start fwd (−bmax)")
show_state(axs[1,1], rec_f[len(rec_f)//2], "mid fwd (~0)")
show_state(axs[1,2], rec_f[-1], "end fwd (+bmax)")
plt.tight_layout(); plt.show()

print(f"Remanence M@b≈0: fwd {Mrem_f:.3e} | bwd {Mrem_b:.3e}")
print(f"Coercive bias b_Hc: fwd {bHc_f:.3f} | bwd {bHc_b:.3f}")
```

```python
# SWG/SCG Magnetism · Exp 7.3 — Balanced (UUD+DDU) patch with χ-memory + pinning (singularity-free)
# Goal: unbiased M≈0 at b≈0 so the M–H loop crosses zero → finite coercive bias.

import numpy as np, matplotlib.pyplot as plt
EPS = 1e-6
rng = np.random.default_rng(713)

# --- domain & helpers (as before) ---
H=W=520
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X = (x-cx).astype(np.float32); Y = (y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def curl2(Jx, Jy):
    return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)
def smooth(Z, n=2):
    V=Z.copy()
    for _ in range(n):
        V=(np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V=(np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V
def lap2(Z):
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# smooth, singularity-free loop with effective twist (base + bias)
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2, bias_eff=0.0):
    Xc=X-(x0-cx); Yc=Y-(y0-cy)
    r=np.hypot(Xc,Yc); th=np.arctan2(Yc,Xc)
    env=np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr =np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi=(sign*twist_gain + bias_eff)*fr*th
    return amp*env*np.exp(1j*phi)

def rotate_pts(pts, origin, theta):
    ox,oy=origin; c,s=np.cos(theta),np.sin(theta); out=[]
    for px,py in pts:
        dx,dy=px-ox,py-oy
        out.append((ox+c*dx-s*dy, oy+s*dx+c*dy))
    return out

def triad(center, signs, s=28.0, gain=3.2, rot=0.0, bias_eff=0.0):
    (x0,y0)=center
    base=[(x0-s,y0),(x0+s,y0),(x0,y0+s*0.58)]
    pts=rotate_pts(base, center, rot)
    Psi=0j
    for (xx,yy),sg in zip(pts, signs):
        Psi+=loop_form_chiral(xx,yy,sign=sg,twist_gain=gain,bias_eff=bias_eff)
    return Psi

def fields_from(Psi):
    Psix,Psiy=grad2(Psi)
    Jx=np.imag(np.conj(Psi)*Psix); Jy=np.imag(np.conj(Psi)*Psiy)
    A2=(np.abs(Psi)**2+EPS)
    Vx,Vy=Jx/A2, Jy/A2
    B=curl2(Jx,Jy)
    return Jx,Jy,Vx,Vy,B

def scg_metrics(A):
    A=A.astype(np.float32)
    chi=smooth(smooth(A,2),2)
    kappa=np.abs(lap2(chi))/(np.abs(chi)+1e-9)
    L_local=1.0/np.sqrt(kappa+1e-9)
    L_metric=float(np.percentile(L_local,10))
    L_geom=float(np.sqrt(1.0+L_metric**2))
    FVR=float(np.mean(L_local<L_metric))
    return L_metric,L_geom,FVR

# --- balanced patch geometry (half UUD, half DDU), spacing/orientation from 6.2e ---
sep=68.0; rot2=np.deg2rad(60.0)
nx,ny=4,3
gx0,gy0=cx-(nx-1)*sep/2, cy-(ny-1)*sep/2
centers=[(gx0+i*sep, gy0+j*sep) for j in range(ny) for i in range(nx)]

# assignment: checkerboard of UUD and DDU (sign-flipped)
signs_UUD=[+1,+1,-1]; signs_DDU=[-1,-1,+1]

assign=[]
for j in range(ny):
    for i in range(nx):
        assign.append(signs_UUD if ((i+j)%2==0) else signs_DDU)

# --- χ-memory + sparse pinning (same idea as 7.2; slightly stronger to ensure crossing) ---
eta   = 0.08   # slower memory
k_mem = 1.00   # memory gain
R_gain= 0.75   # curl -> memory sensitivity
chi_mem = np.zeros((H,W), np.float32)

pin = np.zeros((H,W), np.float32)
for _ in range(22):
    px,py = rng.integers(40,W-40), rng.integers(40,H-40)
    sx,sy = rng.uniform(6,12), rng.uniform(6,12)
    pin += 0.9*np.exp(-((x-px)**2)/(2*sx*sx))*np.exp(-((y-py)**2)/(2*sy*sy))
pin = smooth(pin, n=3) / (np.max(pin)+1e-9)
pin_gain = 0.45

def build_with_memory_and_pinning(b):
    chi_win = smooth(chi_mem, n=2)
    mem_field = k_mem * np.tanh(R_gain * chi_win)
    pin_field = -pin_gain * np.sign(b) * pin
    bias_eff_field = mem_field + b + pin_field

    Psi=0j
    for k,((px,py),sgs) in enumerate(zip(centers, assign)):
        ix,iy = int(round(px)), int(round(py))
        patch = bias_eff_field[max(iy-4,0):iy+5, max(ix-4,0):ix+5]
        b_eff = float(np.mean(patch))
        rot = rot2 if (k%2==0) else -rot2
        Psi += triad((px,py), sgs, rot=rot, bias_eff=b_eff)
    Psi += EPS*(1.0+0.02*rng.standard_normal((H,W))).astype(np.complex64)
    return Psi

def measure(Psi):
    Jx,Jy,Vx,Vy,B = fields_from(Psi)
    A=np.abs(Psi)
    mu = float(np.sum(X*Jy - Y*Jx))       # net moment about patch center
    B2 = float(np.mean(B*B))
    Lm,Lg,FVR = scg_metrics(A)
    return dict(mu=mu,B2=B2,Lm=Lm,Lg=Lg,FVR=FVR,B=B,A=A)

def update_memory(B_like):
    global chi_mem
    chi_mem = (1.0-eta)*chi_mem + eta*smooth(B_like, n=2)

# bias sweep
bmax=0.7
grid=np.linspace(-bmax,bmax,19)
bf=grid; bb=grid[::-1]
rec_f,rec_b=[],[]

for b in bf:
    Psi=build_with_memory_and_pinning(b)
    r=measure(Psi); r['b']=float(b); rec_f.append(r)
    update_memory(r['B'])
for b in bb:
    Psi=build_with_memory_and_pinning(b)
    r=measure(Psi); r['b']=float(b); rec_b.append(r)
    update_memory(r['B'])

# curves and coercivity
def extract(rec):
    b=np.array([r['b'] for r in rec]); M=np.array([r['mu'] for r in rec])
    B2=np.array([r['B2'] for r in rec]); F=np.array([r['FVR'] for r in rec])
    return b,M,B2,F
bf,Mf,B2f,Ff = extract(rec_f); bb,Mb,B2b,Fb = extract(rec_b)

def zero_cross(b,M):
    idx=np.where(np.sign(M[:-1])!=np.sign(M[1:]))[0]
    if len(idx)==0: return np.nan
    i=idx[0]; t = -M[i]/(M[i+1]-M[i]+1e-12)
    return float(b[i] + t*(b[i+1]-b[i]))
bHc_f, bHc_b = zero_cross(bf,Mf), zero_cross(bb,Mb)

i0_f = np.argmin(np.abs(bf)); i0_b = np.argmin(np.abs(bb))
Mrem_f, Mrem_b = float(Mf[i0_f]), float(Mb[i0_b])

# plot
fig,axs=plt.subplots(2,3,figsize=(13,8))
axs[0,0].plot(bf,Mf,'-o',ms=3,label='forward'); axs[0,0].plot(bb,Mb,'-o',ms=3,label='backward')
axs[0,0].axhline(0,color='#777'); axs[0,0].set_title("Balanced patch M–H (χ-memory + pinning)")
axs[0,0].set_xlabel("bias b"); axs[0,0].set_ylabel("μ_z"); axs[0,0].legend()
axs[0,1].plot(bf,B2f,'-o',ms=3,label='⟨B²⟩ fwd'); axs[0,1].plot(bb,B2b,'-o',ms=3,label='⟨B²⟩ bwd'); axs[0,1].legend(); axs[0,1].set_title("⟨B²⟩ vs b")
axs[0,2].plot(bf,Ff,label='FVR fwd'); axs[0,2].plot(bb,Fb,label='FVR bwd'); axs[0,2].legend(); axs[0,2].set_title("FVR vs b")

def showB(ax, rec, title):
    B=rec['B']; vmin,vmax=np.percentile(B,[2,98]); vmax=max(vmax,vmin+1e-9)
    im=ax.imshow(B,cmap='coolwarm',vmin=vmin,vmax=vmax); ax.set_title(title); plt.colorbar(im,ax=ax,shrink=0.7)
showB(axs[1,0], rec_f[0], "start fwd (−bmax)")
showB(axs[1,1], rec_f[len(rec_f)//2], "mid fwd (~0)")
showB(axs[1,2], rec_f[-1], "end fwd (+bmax)")
plt.tight_layout(); plt.show()

print(f"Remanence M@b≈0: forward {Mrem_f:.3e} | backward {Mrem_b:.3e}")
print(f"Coercive bias b_Hc: forward {bHc_f:.3f} | backward {bHc_b:.3f}")
```

```python
# SWG/SCG Magnetism · Exp 7.4 — Domain-wall drift under a tiny bias gradient (χ-memory + pinning)
# Measures wall position & velocity vs bias; still singularity-free.

import numpy as np, matplotlib.pyplot as plt

EPS = 1e-6
rng = np.random.default_rng(714)

# --- grid & helpers (non-singular) ---
H=W=520
y,x = np.mgrid[0:H,0:W].astype(np.float32)
cx, cy = W//2, H//2
X = (x-cx).astype(np.float32); Y = (y-cy).astype(np.float32)

def grad2(Z):
    gx = 0.5*(np.roll(Z,-1,1)-np.roll(Z,1,1))
    gy = 0.5*(np.roll(Z,-1,0)-np.roll(Z,1,0))
    return gx, gy
def curl2(Jx,Jy): return (np.roll(Jy,-1,1)-Jy) - (np.roll(Jx,-1,0)-Jx)
def smooth(Z, n=2):
    V=Z.copy()
    for _ in range(n):
        V=(np.roll(V,1,0)+V+np.roll(V,-1,0))/3
        V=(np.roll(V,1,1)+V+np.roll(V,-1,1))/3
    return V
def lap2(Z): 
    return (np.roll(Z,-1,1)+np.roll(Z,1,1)+np.roll(Z,-1,0)+np.roll(Z,1,0)-4*Z)

# --- loop/triad builders with effective twist (base + memory + local bias) ---
def loop_form_chiral(x0, y0, sign=+1, radius=24.0, width=7.5, amp=1.0, twist_gain=3.2, bias_eff=0.0):
    Xc=X-(x0-cx); Yc=Y-(y0-cy); r=np.hypot(Xc,Yc); th=np.arctan2(Yc,Xc)
    env=np.exp(-((r-radius)**2)/(2*width**2)) + 1e-3
    fr =np.exp(-((r-radius)**2)/(2*(1.2*width)**2))
    phi=(sign*twist_gain + bias_eff)*fr*th
    return amp*env*np.exp(1j*phi)

def rotate_pts(pts, origin, theta):
    ox,oy=origin; c,s=np.cos(theta),np.sin(theta); out=[]
    for px,py in pts:
        dx,dy=px-ox,py-oy
        out.append((ox+c*dx-s*dy, oy+s*dx+c*dy))
    return out

def triad(center, signs, s=28.0, gain=3.2, rot=0.0, bias_eff=0.0):
    (x0,y0)=center
    base=[(x0-s,y0),(x0+s,y0),(x0,y0+s*0.58)]
    pts=rotate_pts(base, center, rot)
    Psi=0j
    for (xx,yy),sg in zip(pts, signs):
        Psi+=loop_form_chiral(xx,yy,sign=sg,twist_gain=gain,bias_eff=bias_eff)
    return Psi

def fields_from(Psi):
    Psix,Psiy=grad2(Psi)
    Jx=np.imag(np.conj(Psi)*Psix); Jy=np.imag(np.conj(Psi)*Psiy)
    A2=(np.abs(Psi)**2+EPS); Vx,Vy=Jx/A2, Jy/A2
    B=curl2(Jx,Jy)
    return Jx,Jy,Vx,Vy,B

# --- balanced patch (UUD+DDU), spacing/orientation from best geometry ---
sep=68.0; rot2=np.deg2rad(60.0)
nx,ny=6,3
gx0,gy0=cx-(nx-1)*sep/2, cy-(ny-1)*sep/2
centers=[(gx0+i*sep, gy0+j*sep) for j in range(ny) for i in range(nx)]
signs_UUD=[+1,+1,-1]; signs_DDU=[-1,-1,+1]
assign=[signs_UUD if ((i+j)%2==0) else signs_DDU
        for j in range(ny) for i in range(nx)]

# --- χ-memory + sparse pinning (as in 7.3) ---
eta, k_mem, R_gain = 0.08, 1.00, 0.75
chi_mem = np.zeros((H,W), np.float32)
pin = np.zeros((H,W), np.float32)
for _ in range(26):
    px,py = rng.integers(40,W-40), rng.integers(40,H-40)
    sx,sy = rng.uniform(6,12), rng.uniform(6,12)
    pin += 0.9*np.exp(-((x-px)**2)/(2*sx*sx))*np.exp(-((y-py)**2)/(2*sy*sy))
pin = smooth(pin, n=3) / (np.max(pin)+1e-9)
pin_gain = 0.45

# --- small bias gradient (drives walls left→right): b(x) = b0 + g*(x/W)
def build_state(b0, grad):
    chi_win = smooth(chi_mem, n=2)
    mem_field = k_mem*np.tanh(R_gain*chi_win)
    bias_field = mem_field + b0 + grad*(x/W) - pin_gain*np.sign(b0)*pin
    Psi=0j
    for k,((px,py),sgs) in enumerate(zip(centers, assign)):
        ix,iy = int(round(px)), int(round(py))
        patch = bias_field[max(iy-4,0):iy+5, max(ix-4,0):ix+5]
        b_eff = float(np.mean(patch))
        rot = rot2 if (k%2==0) else -rot2
        Psi += triad((px,py), sgs, rot=rot, bias_eff=b_eff)
    Psi += EPS*(1.0+0.02*rng.standard_normal((H,W))).astype(np.complex64)
    return Psi

def update_memory(B_like):
    global chi_mem
    chi_mem = (1.0-eta)*chi_mem + eta*smooth(B_like, n=2)

# --- measure wall position: column-mean sign of B-like (center band), then center of mass of transition ---
def wall_position(B, band_half=40):
    row_lo, row_hi = int(cy-band_half), int(cy+band_half)
    col_profile = np.mean(B[row_lo:row_hi,:], axis=0)
    # normalize to [-1,1] and compute "center" of sign change
    s = np.tanh(col_profile / (np.std(col_profile)+1e-9))
    idx = np.arange(W)
    # center of mass weighted by positive slope (detect boundary)
    ds = np.gradient(s)
    w = np.maximum(ds, 0)
    if w.sum()<1e-9: return np.nan
    return float(np.sum(idx*w)/np.sum(w))

# --- drive: fix a small gradient, sweep b0 from -bmax to +bmax with memory carry-over ---
bmax, grad = 0.6, 0.12
grid = np.linspace(-bmax, bmax, 19)

wall_x, mus, bvals = [], [], []
for b0 in grid:
    Psi = build_state(b0, grad)
    Jx,Jy,Vx,Vy,B = fields_from(Psi)
    mus.append(float(np.sum(X*Jy - Y*Jx)))
    bvals.append(float(b0))
    wall_x.append(wall_position(B))
    update_memory(B)

wall_x = np.array(wall_x); mus=np.array(mus); bvals=np.array(bvals)

# --- plots ---
fig,axs = plt.subplots(2,3,figsize=(13,8))
axs[0,0].plot(bvals, mus, '-o', ms=3); axs[0,0].axhline(0,color='#777',lw=1)
axs[0,0].set_title("M vs bias (with gradient)"); axs[0,0].set_xlabel("b0"); axs[0,0].set_ylabel("μ_z")

axs[0,1].plot(bvals, wall_x, '-o', ms=3)
axs[0,1].set_title("Domain-wall position vs bias"); axs[0,1].set_xlabel("b0"); axs[0,1].set_ylabel("x_wall (px)")

# show start/mid/end B-like
def showB(ax, B, title):
    vmin,vmax=np.percentile(B,[2,98]); vmax=max(vmax,vmin+1e-9)
    im=ax.imshow(B,cmap='coolwarm',vmin=vmin,vmax=vmax); ax.set_title(title); plt.colorbar(im,ax=ax,shrink=0.7)

Psi0 = build_state(-bmax, grad); _,_,_,_,B0 = fields_from(Psi0)
Psiz = build_state(0.0,   grad); _,_,_,_,Bz = fields_from(Psiz)
Psip = build_state(+bmax, grad); _,_,_,_,Bp = fields_from(Psip)
showB(axs[1,0], B0, "B-like @ b0≈−bmax")
showB(axs[1,1], Bz, "B-like @ b0≈0")
showB(axs[1,2], Bp, "B-like @ b0≈+bmax")

# quick text
axs[0,2].axis('off')
axs[0,2].text(0,1,f"grad={grad:.3f}\nΔx_wall≈{np.nanmax(wall_x)-np.nanmin(wall_x):.1f} px",
              va='top', family='monospace', fontsize=11)

plt.tight_layout(); plt.show()
```
