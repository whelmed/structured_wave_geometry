# 1. Ray Equations in SWG

Under geometric-optics scaling, the SWG eikonal law is:

$$
\theta_t^2 = c_\chi^2 \, |\nabla\theta|^2 ,
\qquad
c_\chi(x,y) = \frac{1}{\sqrt{\chi(x,y)}} .
$$

Define momentum:

$$
k = \nabla\theta .
$$

The Hamiltonian is:

$$
H(x,k) = c_\chi(x) \, |k|.
$$

Hamilton’s equations:

$$
\dot{x} = c_\chi \frac{k}{|k|},
\qquad
\dot{k} = - |k| \, \nabla c_\chi .
$$

Rays follow null geodesics of the SWG metric:

$$
g_{ij}(x) = \chi(x) \, \delta_{ij}.
$$

---

# 2. Gauge-Coupled Rays

With a geometric gauge field $A_{\mathrm{geom}}(x,y)$, define:

$$
k_{\mathrm{eff}} = k - A_{\mathrm{geom}} .
$$

Gauge-coupled Hamiltonian:

$$
H(x,k) = c_\chi(x) \, |k_{\mathrm{eff}}|.
$$

Ray equations:

$$
\dot{x} = c_\chi \frac{k_{\mathrm{eff}}}{|k_{\mathrm{eff}}|},
\qquad
\dot{k} = - |k_{\mathrm{eff}}| \, \nabla c_\chi .
$$

Only $\chi$ influences curvature and SCG diagnostics.  
$A_{\mathrm{geom}}$ alters direction through $k_{\mathrm{eff}}$ but does not affect the geometry floor.

# 3. SCG Geometry Floor Diagnostics

Curvature proxy:

$$
\kappa(x,y) = \frac{|\nabla^2 \chi|}{|\chi| + \varepsilon} .
$$

Local geometric scale:

$$
L_{\mathrm{local}}(x,y) = \frac{1}{\sqrt{\kappa(x,y) + \varepsilon}} .
$$

Global metric floor (10th percentile):

$$
L_{\mathrm{metric}} = \mathrm{percentile}_{10}(L_{\mathrm{local}}),
\qquad
L_{\mathrm{dyn}} = 1,
\qquad
L_{\mathrm{geom}} = \sqrt{L_{\mathrm{metric}}^2 + L_{\mathrm{dyn}}^2}.
$$

Floor violation occurs when:

$$
L_{\mathrm{local}}(s) < L_{\mathrm{geom}} .
$$

The floor-violation rate (FVR) is the fraction of ray samples where this holds.

Flat $\chi$ gives $\kappa = 0 \Rightarrow L_{\mathrm{local}} \to \infty$ and thus $L_{\mathrm{metric}}$ becomes large (typically $\sim 3\times10^4$).  
This produces FVR = 1.0, which is not collapse, only degeneracy of the floor in flat media.

```python
# SWG Ray Mechanics · Core Utilities
# ----------------------------------
# Dependencies: NumPy, Matplotlib

import numpy as np
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Tuple, Callable

np.set_printoptions(precision=4, suppress=True)

# Small epsilon
EPS = 1e-9

# --------- Grid helpers ---------
@dataclass
class Grid2D:
    nx: int
    ny: int
    dx: float = 1.0
    dy: float = 1.0

    def coords(self) -> Tuple[np.ndarray, np.ndarray]:
        x = (np.arange(self.nx) - 0.5 * (self.nx - 1)) * self.dx
        y = (np.arange(self.ny) - 0.5 * (self.ny - 1)) * self.dy
        X, Y = np.meshgrid(x, y, indexing="xy")
        return X, Y


# --------- Finite differences (periodic) ---------
def grad2_centered(Z: np.ndarray, dx: float, dy: float) -> Tuple[np.ndarray, np.ndarray]:
    gx = (np.roll(Z, -1, axis=1) - np.roll(Z, 1, axis=1)) / (2.0 * dx)
    gy = (np.roll(Z, -1, axis=0) - np.roll(Z, 1, axis=0)) / (2.0 * dy)
    return gx, gy


def lap2_centered(Z: np.ndarray, dx: float, dy: float) -> np.ndarray:
    return (
        np.roll(Z, -1, 1) + np.roll(Z, 1, 1)
        + np.roll(Z, -1, 0) + np.roll(Z, 1, 0)
        - 4.0 * Z
    ) / (dx * dx)  # assumes dx == dy; we keep dx here for simplicity


# --------- Bilinear sampling on periodic grid ---------
def bilinear_sample(Z: np.ndarray, x: np.ndarray, y: np.ndarray,
                    grid: Grid2D) -> np.ndarray:
    """
    Sample Z[y,x] with bilinear interpolation.
    x, y are in the same coordinate system as Grid2D.coords().
    Periodic boundary conditions are used.
    """
    Xg, Yg = Grid2D(grid.nx, grid.ny, grid.dx, grid.dy).coords()
    dx, dy = grid.dx, grid.dy

    # Normalize to index space
    ix = (x - Xg[0, 0]) / dx
    iy = (y - Yg[0, 0]) / dy

    ix0 = np.floor(ix).astype(int)
    iy0 = np.floor(iy).astype(int)
    ix1 = ix0 + 1
    iy1 = iy0 + 1

    # Periodic wrap
    ix0 %= grid.nx
    ix1 %= grid.nx
    iy0 %= grid.ny
    iy1 %= grid.ny

    tx = ix - ix0
    ty = iy - iy0

    f00 = Z[iy0, ix0]
    f10 = Z[iy0, ix1]
    f01 = Z[iy1, ix0]
    f11 = Z[iy1, ix1]

    return (
        (1 - tx) * (1 - ty) * f00
        + tx * (1 - ty) * f10
        + (1 - tx) * ty * f01
        + tx * ty * f11
    )


# --------- Geometry floor proxies (χ-based) ---------
def local_kappa_from_chi(chi: np.ndarray, grid: Grid2D) -> np.ndarray:
    """Curvature proxy κ = |∇²χ| / (|χ|+eps)."""
    lap_chi = lap2_centered(chi, grid.dx, grid.dy)
    return np.abs(lap_chi) / (np.abs(chi) + EPS)


def geometry_floor_from_chi(chi: np.ndarray, grid: Grid2D) -> Tuple[np.ndarray, float, float]:
    """
    Compute per-pixel local scale L_local = 1/sqrt(κ),
    and metric floor L_metric as 10th percentile of L_local.
    Returns (L_local, L_metric, L_geom) with L_geom using a simple
    global dynamic component L_dyn = 1.
    """
    kappa = local_kappa_from_chi(chi, grid)
    L_local = 1.0 / np.sqrt(kappa + EPS)
    L_metric = float(np.percentile(L_local, 10.0))
    L_dyn = 1.0
    L_geom = float(np.sqrt(L_metric * L_metric + L_dyn * L_dyn))
    return L_local, L_metric, L_geom
```

```python
# SWG Ray Mechanics · Medium Definitions
# --------------------------------------

def chi_uniform(grid: Grid2D, chi0: float = 1.0) -> np.ndarray:
    """Uniform medium: χ = const."""
    return np.full((grid.ny, grid.nx), chi0, dtype=np.float32)


def chi_lens(grid: Grid2D, chi_bg: float = 1.0,
             chi_core: float = 2.0,
             r0: float = 20.0,
             sigma: float = 10.0) -> np.ndarray:
    """
    Simple 2D lens-like medium:
    χ(r) = chi_bg + (chi_core - chi_bg) * exp(-r^2 / (2 sigma^2)).
    Rays see lower speed (higher n) near the center.
    """
    X, Y = grid.coords()
    R = np.sqrt(X * X + Y * Y)
    chi = chi_bg + (chi_core - chi_bg) * np.exp(-R * R / (2.0 * sigma * sigma))
    return chi.astype(np.float32)


def chi_channel(grid: Grid2D, chi_bg: float = 1.0,
                chi_channel_val: float = 0.5,
                width: float = 5.0) -> np.ndarray:
    """
    Channel / waveguide medium:
    χ lower in a horizontal band to create a fast channel.
    """
    X, Y = grid.coords()
    mask = np.abs(Y) < width
    chi = np.full_like(X, chi_bg, dtype=np.float32)
    chi[mask] = chi_channel_val
    return chi

def chi_channel_smooth(grid: Grid2D,
                       chi_bg: float = 2.0,
                       chi_channel_val: float = 0.6,
                       width: float = 15.0,
                       sigma: float = 8.0) -> np.ndarray:
    X, Y = grid.coords()
    # Gaussian dip in χ across y (channel centered at y=0)
    chi = chi_bg - (chi_bg - chi_channel_val) * np.exp(-(Y**2)/(2*sigma**2))
    return chi.astype(np.float32)

def chi_wall_smooth(grid: Grid2D,
                    chi_bg: float = 1.0,
                    chi_peak: float = 4.0,
                    x0: float = 0.0,
                    sigma: float = 10.0) -> np.ndarray:
    Xw, Yw = grid.coords()
    chi = chi_bg + (chi_peak - chi_bg) * np.exp(-(Xw - x0)**2 / (2*sigma**2))
    return chi.astype(np.float32)
```

```python
# SWG Ray Mechanics · Hamiltonian Ray Integrator
# ----------------------------------------------

@dataclass
class RayState:
    x: float
    y: float
    kx: float
    ky: float


def c_chi_from_chi(chi: np.ndarray) -> np.ndarray:
    """
    Effective speed field c_chi = 1 / sqrt(chi).
    chi must be strictly positive.
    """
    return 1.0 / np.sqrt(np.maximum(chi, EPS))


def sample_c_and_grad_c(x: np.ndarray, y: np.ndarray,
                        chi: np.ndarray,
                        grid: Grid2D) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Sample c_chi(x,y) and its spatial gradients using bilinear interpolation
    from the grid fields.
    """
    cfield = c_chi_from_chi(chi)
    gx_chi, gy_chi = grad2_centered(chi, grid.dx, grid.dy)
    # gradient of c wrt x,y:
    # c = chi^{-1/2} => ∇c = -(1/2) chi^{-3/2} ∇χ
    pref = -0.5 * chi ** (-1.5)
    cx_field = pref * gx_chi
    cy_field = pref * gy_chi

    c_vals = bilinear_sample(cfield, x, y, grid)
    cx_vals = bilinear_sample(cx_field, x, y, grid)
    cy_vals = bilinear_sample(cy_field, x, y, grid)
    return c_vals, cx_vals, cy_vals


def hamiltonian_rhs(ray: RayState, chi: np.ndarray, grid: Grid2D) -> Tuple[np.ndarray, np.ndarray, np.ndarray, np.ndarray]:
    """
    Right-hand side for ray Hamiltonian system:
    H = c_chi(x) |k|.

    xdot = ∂H/∂k = c_chi * k / |k|
    kdot = -∂H/∂x = -|k| ∇ c_chi
    """
    x = np.array([ray.x])
    y = np.array([ray.y])
    kx = np.array([ray.kx])
    ky = np.array([ray.ky])

    k_mag = np.sqrt(kx * kx + ky * ky) + EPS
    c_vals, cx_vals, cy_vals = sample_c_and_grad_c(x, y, chi, grid)

    xdot = c_vals * kx / k_mag
    ydot = c_vals * ky / k_mag
    kx_dot = -k_mag * cx_vals
    ky_dot = -k_mag * cy_vals

    return xdot[0], ydot[0], kx_dot[0], ky_dot[0]


def step_ray_euler(ray: RayState, chi: np.ndarray, grid: Grid2D, ds: float) -> RayState:
    """Simple Euler step for ray state."""
    dx, dy, dkx, dky = hamiltonian_rhs(ray, chi, grid)
    return RayState(
        x=ray.x + ds * dx,
        y=ray.y + ds * dy,
        kx=ray.kx + ds * dkx,
        ky=ray.ky + ds * dky,
    )


def integrate_ray(ray0: RayState, chi: np.ndarray, grid: Grid2D,
                  ds: float = 0.5, n_steps: int = 400) -> Tuple[np.ndarray, np.ndarray]:
    """
    Integrate one ray using simple Euler integration.
    Returns (xs, ys) arrays of length n_steps+1.
    """
    xs = np.empty(n_steps + 1)
    ys = np.empty(n_steps + 1)
    ray = ray0

    xs[0] = ray.x
    ys[0] = ray.y

    for i in range(1, n_steps + 1):
        ray = step_ray_euler(ray, chi, grid, ds)
        xs[i] = ray.x
        ys[i] = ray.y

    return xs, ys
```

```python
# SWG Ray Mechanics · Along-Ray Diagnostics
# -----------------------------------------

def sample_Lgeom_along_ray(xs: np.ndarray, ys: np.ndarray,
                            chi: np.ndarray,
                            grid: Grid2D) -> Tuple[np.ndarray, float, float]:
    """
    Sample local geometry floor proxies along a ray:
    - L_local from κ(χ) via 1/sqrt(κ)
    - L_metric as global p10 floor
    - L_geom combining a simple L_dyn=1 with L_metric
    """
    L_local, L_metric, L_geom_global = geometry_floor_from_chi(chi, grid)
    # sample L_local at ray positions
    L_samples = bilinear_sample(L_local, xs, ys, grid)
    return L_samples, L_metric, L_geom_global


def floor_violation_rate(ray_L_local: np.ndarray, L_geom_global: float) -> float:
    """
    Fraction of points along the ray where L_local < L_geom_global.
    """
    return float(np.mean(ray_L_local < L_geom_global))
```

```python
# Experiment 1 — Straight Rays in Uniform Medium
# ----------------------------------------------

# Grid
grid = Grid2D(nx=200, ny=200, dx=1.0, dy=1.0)
X, Y = grid.coords()

# Uniform χ
chi_uni = chi_uniform(grid, chi0=1.0)

# Ray initial conditions: a fan of rays launched from left edge
n_rays = 11
x0 = -grid.nx * grid.dx * 0.45
y0_center = 0.0
angles = np.linspace(-0.4, 0.4, n_rays)

rays = []
for ang in angles:
    kx = np.cos(ang)
    ky = np.sin(ang)
    rays.append(RayState(x=x0, y=y0_center, kx=kx, ky=ky))

# Integrate
ds = 1.0
n_steps = 260
all_xs, all_ys = [], []
for r in rays:
    xs, ys = integrate_ray(r, chi_uni, grid, ds=ds, n_steps=n_steps)
    all_xs.append(xs)
    all_ys.append(ys)

# Plot
plt.figure(figsize=(6, 6))
plt.title("Experiment 1 — Straight Rays in Uniform Medium")
plt.xlabel("x")
plt.ylabel("y")
for xs, ys in zip(all_xs, all_ys):
    plt.plot(xs, ys, lw=1.0)
plt.axhline(0.0, color="k", lw=0.5, ls="--")
plt.gca().set_aspect("equal", "box")
plt.tight_layout()
plt.show()

# Geometry floor along central ray
xs_mid, ys_mid = all_xs[len(rays)//2], all_ys[len(rays)//2]
L_samples, L_metric, L_geom = sample_Lgeom_along_ray(xs_mid, ys_mid, chi_uni, grid)
FVR = floor_violation_rate(L_samples, L_geom)

print("Uniform medium:")
print(f"  L_metric = {L_metric:.3f}, L_geom = {L_geom:.3f}")
print(f"  Ray floor-violation rate (FVR) = {FVR:.3f}")
```

```python
# ================================================================
# Experiment 6 — Ray–Ray Interaction Through χ-Memory Coupling
# ================================================================

# This experiment demonstrates how rays influence the medium χ(x,y),
# and how the resulting curvature bends other rays. This is the SWG
# analogue of coherent “gravity” or coherence-mediated forces.

# Grid
grid_R = Grid2D(nx=260, ny=260, dx=1.0, dy=1.0)
XR, YR = grid_R.coords()

# Initial medium: uniform χ
chi_R = np.ones_like(XR, dtype=np.float32)

# Rays: two beams with opposite angles
rays_R = [
    RayState(x=-90, y=-40, kx=1.0, ky=0.3),
    RayState(x=-90, y=+40, kx=1.0, ky=-0.3)
]

# Parameters
ds = 1.0
n_steps = 300

# χ update coefficients
D_chi = 0.08      # diffusion (smoothing)
gamma_chi = 0.02  # memory “pump” from ray amplitude
sigma_smooth = 4  # smoothing width for splat

# Ray footprint amplitude
def ray_splat(chi, x, y, amp=0.8, sigma=6.0):
    """
    Add a Gaussian footprint to χ at ray position (x,y).
    """
    Xg, Yg = grid_R.coords()
    dx = Xg - x
    dy = Yg - y
    w = np.exp(-(dx*dx + dy*dy)/(2*sigma*sigma))
    return chi + amp * w

# Simple χ relaxation / update step
def update_chi(chi):
    # diffusion
    lap = lap2_centered(chi, grid_R.dx, grid_R.dy)
    chi_new = chi + D_chi * lap - gamma_chi * (chi - np.mean(chi))
    return chi_new

# Storage for visualization
all_xs_R = []
all_ys_R = []
all_chi_R = []

# Main loop
cur_rays = rays_R
for step in range(n_steps):

    # Integrate all rays one step
    new_rays = []
    for r in cur_rays:
        new_r = step_ray_euler(r, chi_R, grid_R, ds)
        new_rays.append(new_r)

    # Splat each ray into χ
    chi_tmp = chi_R.copy()
    for r in new_rays:
        chi_tmp = ray_splat(chi_tmp, r.x, r.y, amp=0.5, sigma=5.0)

    # Relax χ
    chi_tmp = update_chi(chi_tmp)

    # Keep χ strictly positive
    chi_tmp = np.clip(chi_tmp, 0.15, None)

    # Update rays and χ
    cur_rays = new_rays
    chi_R = chi_tmp

    # Store outputs
    xs_step = np.array([r.x for r in cur_rays])
    ys_step = np.array([r.y for r in cur_rays])
    all_xs_R.append(xs_step)
    all_ys_R.append(ys_step)
    if step % 20 == 0:
        all_chi_R.append(chi_R.copy())

# ================================================================
# Plot trajectories and final χ-field
# ================================================================

plt.figure(figsize=(6,6))
plt.title("Experiment 6 — Ray–Ray Interaction via Dynamic χ-Field")
plt.imshow(chi_R, origin="lower",
           extent=[XR.min(), XR.max(), YR.min(), YR.max()],
           cmap="Greys", alpha=0.5)

xs_arr = np.array(all_xs_R)
ys_arr = np.array(all_ys_R)

for i in range(len(rays_R)):
    plt.plot(xs_arr[:, i], ys_arr[:, i], lw=2)

plt.xlabel("x")
plt.ylabel("y")
plt.gca().set_aspect("equal", "box")
plt.tight_layout()
plt.show()


# ================================================================
# Geometry-floor diagnostics along each ray
# ================================================================

for i in range(len(rays_R)):
    L_samples_R, L_metric_R, L_geom_R = sample_Lgeom_along_ray(
        xs_arr[:, i], ys_arr[:, i], chi_R, grid_R)
    FVR_R = floor_violation_rate(L_samples_R, L_geom_R)

    print(f"Ray {i}: L_metric={L_metric_R:.3f}, L_geom={L_geom_R:.3f}, FVR={FVR_R:.3f}")

    s_axis = np.arange(L_samples_R.size) * ds
    plt.figure(figsize=(7,3))
    plt.plot(s_axis, L_samples_R, label="L_local(s)")
    plt.axhline(L_geom_R, color="r", ls="--", label="L_geom")
    plt.title(f"Ray {i} — Geometry Floor Along Trajectory")
    plt.xlabel("ray parameter s")
    plt.ylabel("local scale")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# ================================================================
# Gauge-Coupled Rays · Utilities and Hamiltonian
# ================================================================

# Geometric gauge field A_geom(x,y) is represented as two grids:
#   A_x_field(y,x), A_y_field(y,x)
# Rays follow a modified momentum: k_eff = k - A_geom

def sample_A_geom(x: np.ndarray, y: np.ndarray,
                  Ax_field: np.ndarray, Ay_field: np.ndarray,
                  grid: Grid2D) -> Tuple[np.ndarray, np.ndarray]:
    """
    Bilinear sample of A_geom at positions (x,y).
    Ax_field, Ay_field are shape (ny, nx), matching grid coords.
    """
    Ax_vals = bilinear_sample(Ax_field, x, y, grid)
    Ay_vals = bilinear_sample(Ay_field, x, y, grid)
    return Ax_vals, Ay_vals


def hamiltonian_rhs_gauge(ray: RayState,
                          chi: np.ndarray,
                          Ax_field: np.ndarray, Ay_field: np.ndarray,
                          grid: Grid2D) -> Tuple[float, float, float, float]:
    """
    Gauge-coupled Hamiltonian system:
        H = c_chi(x) |k_eff|,   k_eff = k - A_geom

    For simplicity we treat A_geom as static and include it only in k_eff.
    The metric curvature (via c_chi) still drives k̇ as before.

    Equations:
        ẋ = c_chi * k_eff / |k_eff|
        k̇ = -|k_eff| ∇ c_chi
    """

    x = np.array([ray.x], dtype=float)
    y = np.array([ray.y], dtype=float)
    kx = np.array([ray.kx], dtype=float)
    ky = np.array([ray.ky], dtype=float)

    # Sample speed and its gradient
    c_vals, cx_vals, cy_vals = sample_c_and_grad_c(x, y, chi, grid)

    # Sample gauge field
    Ax_vals, Ay_vals = sample_A_geom(x, y, Ax_field, Ay_field, grid)

    # Effective momentum
    k_eff_x = kx - Ax_vals
    k_eff_y = ky - Ay_vals
    k_eff_mag = np.sqrt(k_eff_x*k_eff_x + k_eff_y*k_eff_y) + EPS

    # Ray equations
    xdot = c_vals * k_eff_x / k_eff_mag
    ydot = c_vals * k_eff_y / k_eff_mag

    # Metric-driven bending (gauge-free for simplicity)
    kx_dot = -k_eff_mag * cx_vals
    ky_dot = -k_eff_mag * cy_vals

    return xdot[0], ydot[0], kx_dot[0], ky_dot[0]


def step_ray_euler_gauge(ray: RayState, chi: np.ndarray,
                         Ax_field: np.ndarray, Ay_field: np.ndarray,
                         grid: Grid2D, ds: float) -> RayState:
    """
    Single Euler step for a gauge-coupled ray.
    """
    dx, dy, dkx, dky = hamiltonian_rhs_gauge(ray, chi, Ax_field, Ay_field, grid)
    return RayState(
        x=ray.x + ds*dx,
        y=ray.y + ds*dy,
        kx=ray.kx + ds*dkx,
        ky=ray.ky + ds*dky
    )


def integrate_ray_gauge(ray0: RayState, chi: np.ndarray,
                        Ax_field: np.ndarray, Ay_field: np.ndarray,
                        grid: Grid2D, ds: float = 0.5,
                        n_steps: int = 400) -> Tuple[np.ndarray, np.ndarray]:
    """
    Integrate one gauge-coupled ray; returns arrays xs, ys.
    """
    xs = np.empty(n_steps + 1, dtype=float)
    ys = np.empty(n_steps + 1, dtype=float)
    ray = ray0

    xs[0] = ray.x
    ys[0] = ray.y

    for i in range(1, n_steps + 1):
        ray = step_ray_euler_gauge(ray, chi, Ax_field, Ay_field, grid, ds)
        xs[i] = ray.x
        ys[i] = ray.y

    return xs, ys
```

```python
# ================================================================
# Experiment 9 — Hybrid Gauge–Metric Rays (χ + A_geom Together)
# ================================================================
# In this experiment, rays propagate through BOTH:
#   1) a metric curvature χ-field (lensing, focusing, scattering), and
#   2) a vortex-like gauge field A_geom (Lorentz-like steering).
#
# This produces rich hybrid trajectories where gauge and metric
# effects can reinforce, compete, or cancel depending on geometry.

# -------------------------------------------------
# Grid
# -------------------------------------------------
grid_H = Grid2D(nx=260, ny=260, dx=1.0, dy=1.0)
XH, YH = grid_H.coords()

# -------------------------------------------------
# Metric curvature: smooth χ-lens + offset
# -------------------------------------------------
def chi_lens_smooth(grid, chi_bg=1.0, chi_core=3.0,
                    x0=-30.0, y0=0.0, sigma=40.0):
    X, Y = grid.coords()
    dx = X - x0
    dy = Y - y0
    R2 = dx*dx + dy*dy
    chi = chi_bg + (chi_core - chi_bg) * np.exp(-R2/(2*sigma*sigma))
    return chi.astype(np.float32)

# Metric curvature field (offset to the left)
from scipy.ndimage import gaussian_filter
chi_H = gaussian_filter(chi_H, sigma=2.0).astype(np.float32)


# -------------------------------------------------
# Gauge field: vortex-like A_geom centered at the origin
# -------------------------------------------------
A0 = 45.0
r_core = 15.0

R2_H = XH*XH + YH*YH
den_H = R2_H + r_core*r_core

Ax_H = (-A0 * YH / den_H).astype(np.float32)
Ay_H = ( A0 * XH / den_H).astype(np.float32)

# -------------------------------------------------
# Ray bundle setup
# -------------------------------------------------
n_rays = 11
x0 = -grid_H.nx * grid_H.dx * 0.45
y_positions = np.linspace(-80.0, 80.0, n_rays)

# Aim slightly upward to encourage hybrid scattering
k_norm = np.sqrt(1.0 + (0.15)**2)
kx0 = 1.0 / k_norm
ky0 = 0.15 / k_norm

rays_H = [RayState(x=x0, y=y0, kx=kx0, ky=ky0) for y0 in y_positions]

ds = 1.0
n_steps = 420

all_xs_H, all_ys_H = [], []
for r in rays_H:
    xs, ys = integrate_ray_gauge(r,
                                 chi_H,
                                 Ax_H, Ay_H,
                                 grid_H,
                                 ds=ds,
                                 n_steps=n_steps)
    all_xs_H.append(xs)
    all_ys_H.append(ys)

# -------------------------------------------------
# Visualization
# -------------------------------------------------
plt.figure(figsize=(6,6))
plt.title("Experiment 9 — Hybrid Gauge–Metric Rays (χ + A_geom)")

# Background: metric curvature + gauge magnitude
A_mag_H = np.sqrt(Ax_H*Ax_H + Ay_H*Ay_H)
plt.imshow(chi_H + 0.25*A_mag_H, origin="lower",
           extent=[XH.min(), XH.max(), YH.min(), YH.max()],
           cmap="Greys", alpha=0.45)

for xs, ys in zip(all_xs_H, all_ys_H):
    plt.plot(xs, ys, lw=1.3)

plt.xlabel("x")
plt.ylabel("y")
plt.gca().set_aspect("equal", "box")
plt.tight_layout()
plt.show()

# -------------------------------------------------
# Geometry floor along the central ray
# (Floor depends only on χ, not gauge.)
# -------------------------------------------------
idx_mid = len(rays_H)//2
xs_mid_H = all_xs_H[idx_mid]
ys_mid_H = all_ys_H[idx_mid]

L_samp_H, L_metric_H, L_geom_H = sample_Lgeom_along_ray(xs_mid_H,
                                                        ys_mid_H,
                                                        chi_H,
                                                        grid_H)
FVR_H = floor_violation_rate(L_samp_H, L_geom_H)

print("Hybrid gauge-metric ray (mid):")
print(f"  L_metric = {L_metric_H:.3f}, L_geom = {L_geom_H:.3f}, FVR = {FVR_H:.3f}")

s_axis = np.arange(L_samp_H.size) * ds
plt.figure(figsize=(7,3))
plt.plot(s_axis, L_samp_H, label="L_local(s)")
plt.axhline(L_geom_H, color="r", ls="--", label="L_geom")
plt.title("Mid Ray — Geometry Floor Along Trajectory (Hybrid Case)")
plt.xlabel("ray parameter s")
plt.ylabel("local scale")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ================================================================
# Experiment 10 — Ray Caustics in χ-Lens (Metric Rays)
# ================================================================
# A dense fan of rays is launched from a point source and propagated
# through a smooth χ-lens. Neighboring rays focus and cross,
# forming caustics: regions of high ray density where the bundle
# effectively "folds". We visualize both the rays and the caustic
# loci, and compute the geometry floor along the central ray.

# -------------------------------------------------
# Grid and χ-lens
# -------------------------------------------------
grid_C = Grid2D(nx=260, ny=260, dx=1.0, dy=1.0)
XC, YC = grid_C.coords()

# Smooth χ-lens centered near origin
chi_Caustic = chi_lens_smooth(
    grid_C,
    chi_bg=1.0,
    chi_core=3.5,
    x0=0.0,
    y0=0.0,
    sigma=45.0
)

# Optional: smooth χ to remove any Laplacian artifacts
from scipy.ndimage import gaussian_filter
chi_Caustic = gaussian_filter(chi_Caustic, sigma=1.5).astype(np.float32)

# -------------------------------------------------
# Ray fan from a point source
# -------------------------------------------------
n_rays = 41
x_src = -100.0
y_src = 0.0

# Angles spanning a modest cone
theta_min = -0.45
theta_max =  0.45
thetas0 = np.linspace(theta_min, theta_max, n_rays)

rays_Caustic = []
for th in thetas0:
    kx0 = np.cos(th)
    ky0 = np.sin(th)
    rays_Caustic.append(RayState(x=x_src, y=y_src, kx=kx0, ky=ky0))

ds = 1.0
n_steps = 360

# Integrate all rays
all_xs_Ca, all_ys_Ca = [], []
for r in rays_Caustic:
    xs, ys = integrate_ray(r, chi_Caustic, grid_C, ds=ds, n_steps=n_steps)
    all_xs_Ca.append(xs)
    all_ys_Ca.append(ys)

all_xs_Ca = np.array(all_xs_Ca)  # shape (n_rays, n_steps+1)
all_ys_Ca = np.array(all_ys_Ca)

# -------------------------------------------------
# Caustic detection: where neighboring rays nearly intersect
# -------------------------------------------------
# We use a simple geometric proxy: where the separation between
# neighboring rays in y (at the same step) is very small. This
# approximates locations where the bundle Jacobian det(∂x/∂θ0)
# approaches zero.

# For each step, compute |y_i - y_{i+1}|
sep_y = np.abs(np.diff(all_ys_Ca, axis=0))  # shape (n_rays-1, n_steps+1)

# Threshold: small fraction of typical separation
# We use a quantile-based threshold per-step to be robust.
caustic_mask = np.zeros_like(sep_y, dtype=bool)
for t_idx in range(sep_y.shape[1]):
    col = sep_y[:, t_idx]
    q = np.quantile(col, 0.15)  # 15th percentile
    caustic_mask[:, t_idx] = col <= q

# Now build caustic points at midpoints between rays where mask is True
caustic_x = []
caustic_y = []
for i in range(sep_y.shape[0]):
    for t_idx in np.where(caustic_mask[i])[0]:
        x_mid = 0.5 * (all_xs_Ca[i, t_idx] + all_xs_Ca[i+1, t_idx])
        y_mid = 0.5 * (all_ys_Ca[i, t_idx] + all_ys_Ca[i+1, t_idx])
        caustic_x.append(x_mid)
        caustic_y.append(y_mid)

caustic_x = np.array(caustic_x)
caustic_y = np.array(caustic_y)

# -------------------------------------------------
# Visualization: rays + caustic loci
# -------------------------------------------------
plt.figure(figsize=(6,6))
plt.title("Experiment 10 — Ray Caustics in χ-Lens")

plt.imshow(chi_Caustic, origin="lower",
           extent=[XC.min(), XC.max(), YC.min(), YC.max()],
           cmap="Greys", alpha=0.35)

for i in range(n_rays):
    plt.plot(all_xs_Ca[i], all_ys_Ca[i], lw=0.9)

# Overlay caustic points
if caustic_x.size > 0:
    plt.scatter(caustic_x, caustic_y, s=8, c="red", alpha=0.8,
                label="caustic loci")

plt.xlabel("x")
plt.ylabel("y")
plt.gca().set_aspect("equal", "box")
if caustic_x.size > 0:
    plt.legend(loc="upper right")
plt.tight_layout()
plt.show()

# -------------------------------------------------
# Geometry floor along central ray in the fan
# -------------------------------------------------
idx_mid = n_rays // 2
xs_mid_Ca = all_xs_Ca[idx_mid]
ys_mid_Ca = all_ys_Ca[idx_mid]

L_samp_Ca, L_metric_Ca, L_geom_Ca = sample_Lgeom_along_ray(
    xs_mid_Ca, ys_mid_Ca, chi_Caustic, grid_C
)
FVR_Ca = floor_violation_rate(L_samp_Ca, L_geom_Ca)

print("Caustic central ray:")
print(f"  L_metric = {L_metric_Ca:.3f}, L_geom = {L_geom_Ca:.3f}, FVR = {FVR_Ca:.3f}")

s_axis = np.arange(L_samp_Ca.size) * ds
plt.figure(figsize=(7,3))
plt.plot(s_axis, L_samp_Ca, label="L_local(s)")
plt.axhline(L_geom_Ca, color="r", ls="--", label="L_geom")
plt.title("Central Ray — Geometry Floor Along Trajectory (Caustic Case)")
plt.xlabel("ray parameter s")
plt.ylabel("local scale")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ================================================================
# Experiment 11 — Ray Bundle Jacobian Collapse in χ-Lens
# ================================================================
# We launch a fan of rays parameterized by initial angle θ_0, propagate
# them through a χ-lens, and compute the Jacobian
#
#   J(θ_0, t) = ∂y(θ_0, t) / ∂θ_0
#
# using central finite differences. Caustics occur when J → 0:
# neighboring rays converge and the bundle "folds".
#
# We visualize:
#  1) Ray trajectories
#  2) Caustic loci on top of the rays (zero-J points)
#  3) A (θ_0, t) heatmap of log10|J| with caustic bands.

# -------------------------------------------------
# Grid and χ-lens (reuse lens shape from Exp 10)
# -------------------------------------------------
grid_J = Grid2D(nx=260, ny=260, dx=1.0, dy=1.0)
XJ, YJ = grid_J.coords()

chi_J = chi_lens_smooth(
    grid_J,
    chi_bg=1.0,
    chi_core=3.5,
    x0=0.0,
    y0=0.0,
    sigma=45.0
)

from scipy.ndimage import gaussian_filter
chi_J = gaussian_filter(chi_J, sigma=1.5).astype(np.float32)

# -------------------------------------------------
# Ray fan parameterized by θ_0
# -------------------------------------------------
n_rays = 51                   # more rays → smoother Jacobian
x_src = -100.0
y_src = 0.0

theta_min = -0.50
theta_max =  0.50
theta0 = np.linspace(theta_min, theta_max, n_rays)   # bundle parameter

rays_J = []
for th in theta0:
    kx0 = np.cos(th)
    ky0 = np.sin(th)
    rays_J.append(RayState(x=x_src, y=y_src, kx=kx0, ky=ky0))

ds = 1.0
n_steps = 360

all_xs_J, all_ys_J = [], []
for r in rays_J:
    xs, ys = integrate_ray(r, chi_J, grid_J, ds=ds, n_steps=n_steps)
    all_xs_J.append(xs)
    all_ys_J.append(ys)

all_xs_J = np.array(all_xs_J)  # shape (n_rays, n_steps+1)
all_ys_J = np.array(all_ys_J)

# -------------------------------------------------
# Jacobian J(θ_0, t) = ∂y(θ_0,t)/∂θ_0 via central differences
# -------------------------------------------------
# θ_0 is uniform, so Δθ = constant
dtheta = theta0[1] - theta0[0]

# central diff in θ_0; forward/backward at edges
J_y = np.zeros_like(all_ys_J, dtype=float)  # shape (n_rays, n_steps+1)

# central for interior
J_y[1:-1, :] = (all_ys_J[2:, :] - all_ys_J[0:-2, :]) / (2*dtheta)
# one-sided at edges
J_y[0,  :] = (all_ys_J[1, :] - all_ys_J[0,  :]) / dtheta
J_y[-1, :] = (all_ys_J[-1, :] - all_ys_J[-2, :]) / dtheta

# -------------------------------------------------
# Caustic detection: where J_y changes sign and |J_y| small
# -------------------------------------------------
eps_J = 1e-3  # small threshold to avoid numerical noise

caustic_x_J = []
caustic_y_J = []

for i in range(1, n_rays-1):
    for t_idx in range(1, n_steps):
        # check sign change across θ_0
        J_left  = J_y[i-1, t_idx]
        J_mid   = J_y[i,   t_idx]
        J_right = J_y[i+1, t_idx]

        if np.sign(J_left) != np.sign(J_right) and np.abs(J_mid) < eps_J:
            # record approximate caustic point at this ray/time
            caustic_x_J.append(all_xs_J[i, t_idx])
            caustic_y_J.append(all_ys_J[i, t_idx])

caustic_x_J = np.array(caustic_x_J)
caustic_y_J = np.array(caustic_y_J)

# -------------------------------------------------
# Visualization 1: rays + Jacobian caustics
# -------------------------------------------------
plt.figure(figsize=(6,6))
plt.title("Experiment 11 — Ray Bundle Jacobian Collapse in χ-Lens")

plt.imshow(chi_J, origin="lower",
           extent=[XJ.min(), XJ.max(), YJ.min(), YJ.max()],
           cmap="Greys", alpha=0.35)

for i in range(n_rays):
    plt.plot(all_xs_J[i], all_ys_J[i], lw=0.9)

if caustic_x_J.size > 0:
    plt.scatter(caustic_x_J, caustic_y_J, s=8, c="red", alpha=0.9,
                label="Jacobian caustics (J_y ≈ 0)")

plt.xlabel("x")
plt.ylabel("y")
plt.gca().set_aspect("equal", "box")
if caustic_x_J.size > 0:
    plt.legend(loc="upper right")
plt.tight_layout()
plt.show()

# -------------------------------------------------
# Visualization 2: log10|J_y| heatmap over (θ_0, t)
# -------------------------------------------------
# Time axis
t_axis = np.arange(n_steps+1) * ds

# Avoid log(0)
J_abs = np.abs(J_y) + 1e-12
logJ = np.log10(J_abs)

plt.figure(figsize=(8,4))
plt.title("log10|∂y/∂θ_0| — Bundle Jacobian Over (θ_0, t)")
plt.imshow(logJ,
           extent=[t_axis.min(), t_axis.max(), theta0.min(), theta0.max()],
           aspect="auto", origin="lower", cmap="viridis")
plt.colorbar(label="log10|∂y/∂θ_0|")
plt.xlabel("ray parameter s (time)")
plt.ylabel("initial angle θ_0")
plt.tight_layout()
plt.show()

# -------------------------------------------------
# Geometry floor along central ray (optional SCG check)
# -------------------------------------------------
idx_mid = n_rays // 2
xs_mid_J = all_xs_J[idx_mid]
ys_mid_J = all_ys_J[idx_mid]

L_samp_J, L_metric_J, L_geom_J = sample_Lgeom_along_ray(
    xs_mid_J, ys_mid_J, chi_J, grid_J
)
FVR_J = floor_violation_rate(L_samp_J, L_geom_J)

print("Jacobian central ray:")
print(f"  L_metric = {L_metric_J:.3f}, L_geom = {L_geom_J:.3f}, FVR = {FVR_J:.3f}")

s_axis = np.arange(L_samp_J.size) * ds
plt.figure(figsize=(7,3))
plt.plot(s_axis, L_samp_J, label="L_local(s)")
plt.axhline(L_geom_J, color="r", ls="--", label="L_geom")
plt.title("Central Ray — Geometry Floor Along Trajectory (Jacobian Case)")
plt.xlabel("ray parameter s")
plt.ylabel("local scale")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ================================================================
# Experiment 13 — Catastrophe Caustics in SWG Ray Mechanics
# ================================================================
# We construct χ fields that produce canonical catastrophe optics:
#   - Cusp (A3)
#   - Swallowtail (A4)
# Using SWG ray propagation and nearest-neighbor caustic detection.

# ---------------------------------------------------------------
#  Setup
# ---------------------------------------------------------------
grid_cat = Grid2D(nx=260, ny=260, dx=1.0, dy=1.0)
XCt, YCt = grid_cat.coords()

from scipy.ndimage import gaussian_filter

def smooth(X, sigma=2.0):
    return gaussian_filter(X, sigma=sigma).astype(np.float32)

# ---------------------------------------------------------------
#  1. Cusp χ-Field (A3)
# ---------------------------------------------------------------
# A cusp arises when curvature concentrates along a curved ridge.
# We use a stretched Gaussian ridge, then smooth to avoid artifacts.

def chi_cusp_field(grid, chi_bg=1.0, chi_h=3.0, x0=-20, y0=0,
                   sigma_x=45, sigma_y=18, skew=0.03):
    X, Y = grid.coords()
    Xs = X - x0
    Ys = Y - y0 - skew*(X - x0)   # tilt to force cusp
    R = (Xs**2)/(2*sigma_x**2) + (Ys**2)/(2*sigma_y**2)
    chi = chi_bg + (chi_h - chi_bg)*np.exp(-R)
    return smooth(chi, sigma=2.0)

chi_cusp = chi_cusp_field(grid_cat)

# ---------------------------------------------------------------
#  2. Swallowtail χ-Field (A4)
# ---------------------------------------------------------------
# A swallowtail requires a 3-fold curvature structure.
# Build three overlapping elongated lobes with tuned strengths.

def chi_swallowtail_field(grid, chi_bg=1.0, amps=(2.8, 2.2, 2.0),
                          centers=((-40,0), (-10,15), (-10,-15)),
                          sigma=25):
    X, Y = grid.coords()
    chi = np.ones_like(X, dtype=np.float32)*chi_bg
    for (cx, cy), amp in zip(centers, amps):
        dx = X - cx
        dy = Y - cy
        chi += amp*np.exp(-(dx*dx + dy*dy)/(2*sigma*sigma))
    return smooth(chi, sigma=3.0)

chi_swallowtail = chi_swallowtail_field(grid_cat)

# ---------------------------------------------------------------
# Ray fan generator
# ---------------------------------------------------------------
def generate_ray_fan(n_rays=51, x_src=-110, y_src=0,
                     th_min=-0.55, th_max=0.55):
    thetas = np.linspace(th_min, th_max, n_rays)
    rays = []
    for th in thetas:
        kx = np.cos(th)
        ky = np.sin(th)
        rays.append(RayState(x=x_src, y=y_src, kx=kx, ky=ky))
    return rays, thetas

# ---------------------------------------------------------------
# Caustic detector via nearest-neighbor separation
# ---------------------------------------------------------------
def detect_caustics(all_xs, all_ys, quant=0.10):
    sep_y = np.abs(np.diff(all_ys, axis=0))
    mask = np.zeros_like(sep_y, dtype=bool)

    for t in range(sep_y.shape[1]):
        col = sep_y[:, t]
        q = np.quantile(col, quant)
        mask[:, t] = col <= q

    cx, cy = [], []
    for i in range(sep_y.shape[0]):
        for t in np.where(mask[i])[0]:
            x_mid = 0.5*(all_xs[i,t] + all_xs[i+1,t])
            y_mid = 0.5*(all_ys[i,t] + all_ys[i+1,t])
            cx.append(x_mid)
            cy.append(y_mid)
    return np.array(cx), np.array(cy)

# ---------------------------------------------------------------
# Ray propagation through a χ-field
# ---------------------------------------------------------------
def run_catastrophe_sim(chi_field):
    rays, thetas = generate_ray_fan()
    ds = 1.0
    steps = 380
    all_xs, all_ys = [], []
    for r in rays:
        xs, ys = integrate_ray(r, chi_field, grid_cat, ds=ds, n_steps=steps)
        all_xs.append(xs)
        all_ys.append(ys)
    return np.array(all_xs), np.array(all_ys)

# ---------------------------------------------------------------
# Run cusp simulation
# ---------------------------------------------------------------
xs_cusp, ys_cusp = run_catastrophe_sim(chi_cusp)
cx_cusp, cy_cusp = detect_caustics(xs_cusp, ys_cusp, quant=0.12)

plt.figure(figsize=(6,6))
plt.title("Experiment 13 — Cusp Caustics (A3)")
plt.imshow(chi_cusp, origin="lower",
           extent=[XCt.min(), XCt.max(), YCt.min(), YCt.max()],
           cmap="Greys", alpha=0.35)
for i in range(xs_cusp.shape[0]):
    plt.plot(xs_cusp[i], ys_cusp[i], lw=0.8)
plt.scatter(cx_cusp, cy_cusp, s=8, c="red", label="cusp caustics")
plt.legend()
plt.gca().set_aspect("equal","box")
plt.tight_layout()
plt.show()

# ---------------------------------------------------------------
# Run swallowtail simulation
# ---------------------------------------------------------------
xs_sw, ys_sw = run_catastrophe_sim(chi_swallowtail)
cx_sw, cy_sw = detect_caustics(xs_sw, ys_sw, quant=0.08)

plt.figure(figsize=(6,6))
plt.title("Experiment 13 — Swallowtail Caustics (A4)")
plt.imshow(chi_swallowtail, origin="lower",
           extent=[XCt.min(), XCt.max(), YCt.min(), YCt.max()],
           cmap="Greys", alpha=0.35)
for i in range(xs_sw.shape[0]):
    plt.plot(xs_sw[i], ys_sw[i], lw=0.8)
plt.scatter(cx_sw, cy_sw, s=8, c="red", label="swallowtail caustics")
plt.legend()
plt.gca().set_aspect("equal","box")
plt.tight_layout()
plt.show()
```

```python
# ================================================================
# Experiment 14 — Hyperbolic Umbilic (D4) Caustics in SWG Ray Geometry
# ================================================================
# This catastrophe requires a 2D control-space χ field.
# We construct χ(x,y) with cubic cross-terms that produce
# the canonical hyperbolic-umbilic curvature geometry.

# ---------------------------------------------------------------
# Grid
# ---------------------------------------------------------------
grid_HU = Grid2D(nx=260, ny=260, dx=1.0, dy=1.0)
XHU, YHU = grid_HU.coords()

from scipy.ndimage import gaussian_filter

# ---------------------------------------------------------------
# D4 Hyperbolic Umbilic χ-field
# ---------------------------------------------------------------
# A minimal canonical form for the hyperbolic umbilic potential is:
#     V(x,y) = x^3 + y^3 + a x y
# We'll embed an SWG version inside χ by exponentiating a smoothed potential.

def chi_hyperbolic_umbilic(grid, chi_bg=1.0, chi_amp=3.5,
                           a=0.012, scale=0.015, sigma=3.0):
    X, Y = grid.coords()

    # Canonical D4-like potential
    V = (X*X*X + Y*Y*Y) * scale + a * X * Y

    # Convert to χ as a positive curvature metric
    chi = chi_bg + chi_amp * np.exp(-V*V)

    # Smooth to prevent Laplacian noise artifacts
    chi = gaussian_filter(chi, sigma=sigma)

    # Ensure positivity
    return np.clip(chi, 0.25, None).astype(np.float32)

chi_HU = chi_hyperbolic_umbilic(grid_HU)
chi_HU = gaussian_filter(chi_HU, sigma=3.5).astype(np.float32)


# ---------------------------------------------------------------
# Ray fan from a point source
# ---------------------------------------------------------------
n_rays = 51
x_src = -110.0
y_src = 0.0

theta_min = -0.65
theta_max =  0.65
theta0_HU = np.linspace(theta_min, theta_max, n_rays)

rays_HU = []
for th in theta0_HU:
    rays_HU.append(
        RayState(x=x_src, y=y_src, kx=np.cos(th), ky=np.sin(th))
    )

ds = 1.0
n_steps = 380

all_xs_HU = []
all_ys_HU = []
for r in rays_HU:
    xs, ys = integrate_ray(r, chi_HU, grid_HU, ds=ds, n_steps=n_steps)
    all_xs_HU.append(xs)
    all_ys_HU.append(ys)

all_xs_HU = np.array(all_xs_HU)
all_ys_HU = np.array(all_ys_HU)

# ---------------------------------------------------------------
# Caustic detection (nearest-neighbor separation)
# ---------------------------------------------------------------
sep_y_HU = np.abs(np.diff(all_ys_HU, axis=0))
mask_HU = np.zeros_like(sep_y_HU, dtype=bool)

for t_idx in range(sep_y_HU.shape[1]):
    col = sep_y_HU[:, t_idx]
    q = np.quantile(col, 0.10)  # collapse threshold
    mask_HU[:, t_idx] = col <= q

cx_HU = []
cy_HU = []
for i in range(sep_y_HU.shape[0]):
    for t_idx in np.where(mask_HU[i])[0]:
        cx_HU.append(0.5*(all_xs_HU[i,t_idx] + all_xs_HU[i+1,t_idx]))
        cy_HU.append(0.5*(all_ys_HU[i,t_idx] + all_ys_HU[i+1,t_idx]))

cx_HU = np.array(cx_HU)
cy_HU = np.array(cy_HU)

# ---------------------------------------------------------------
# Visualization
# ---------------------------------------------------------------
plt.figure(figsize=(6,6))
plt.title("Experiment 14 — Hyperbolic Umbilic Caustics (D4)")

plt.imshow(chi_HU, origin="lower",
           extent=[XHU.min(), XHU.max(), YHU.min(), YHU.max()],
           cmap="Greys", alpha=0.35)

for i in range(n_rays):
    plt.plot(all_xs_HU[i], all_ys_HU[i], lw=0.8)

if cx_HU.size > 0:
    plt.scatter(cx_HU, cy_HU, s=8, c="red", label="D4 caustic loci")

plt.xlabel("x")
plt.ylabel("y")
plt.gca().set_aspect("equal","box")
plt.legend()
plt.tight_layout()
plt.show()

# ---------------------------------------------------------------
# Geometry floor along the central ray
# ---------------------------------------------------------------
idx_mid = n_rays // 2
xs_mid_HU = all_xs_HU[idx_mid]
ys_mid_HU = all_ys_HU[idx_mid]

L_samp_HU, L_metric_HU, L_geom_HU = sample_Lgeom_along_ray(
    xs_mid_HU, ys_mid_HU, chi_HU, grid_HU
)
FVR_HU = floor_violation_rate(L_samp_HU, L_geom_HU)

print("D4 catastrophe central ray:")
print(f"  L_metric = {L_metric_HU:.3f}, L_geom = {L_geom_HU:.3f}, FVR = {FVR_HU:.3f}")

s_axis = np.arange(L_samp_HU.size) * ds
plt.figure(figsize=(7,3))
plt.plot(s_axis, L_samp_HU, label="L_local(s)")
plt.axhline(L_geom_HU, color="r", ls="--", label="L_geom")
plt.title("Central Ray — Geometry Floor (Hyperbolic Umbilic)")
plt.xlabel("ray parameter s")
plt.ylabel("local scale")
plt.legend()
plt.tight_layout()
plt.show()
```
