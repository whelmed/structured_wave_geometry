$$
\Gamma = \oint_\gamma \nabla\theta\cdot d\ell = 2\pi n, \qquad n\in\mathbb{Z}.

identical windings repel. This is the geometric origin of **charge, force, and composite structure**  
in this framework.

```python
# Cell 0 — Utilities (imports, numerics, invariants, visualization)
# Idiomatic, dependency-light; pure NumPy + Matplotlib; FFT-based Gaussian blur.
from __future__ import annotations
import numpy as np
import numpy.fft as nfft
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Tuple, Dict

# ---------------------------
# Reproducibility
# ---------------------------
RNG = np.random.default_rng(42)

# ---------------------------
# Grid helpers
# ---------------------------
@dataclass
class Grid:
    n: int
    dx: float = 1.0

    def coords(self) -> Tuple[np.ndarray, np.ndarray]:
        ax = (np.arange(self.n) - 0.5 * (self.n - 1)) * self.dx
        X, Y = np.meshgrid(ax, ax, indexing="xy")
        return X, Y

# ---------------------------
# FFT Gaussian blur (no external deps)
# ---------------------------
def gaussian_blur_fft(img: np.ndarray, sigma: float) -> np.ndarray:
    """Separable Gaussian via frequency-domain transfer; sigma in pixels."""
    if sigma <= 0:
        return img
    n, m = img.shape
    ky = nfft.fftfreq(n)[:, None]
    kx = nfft.fftfreq(m)[None, :]
    k2 = kx * kx + ky * ky
    H = np.exp(-2.0 * (np.pi ** 2) * (sigma ** 2) * k2)
    F = nfft.fft2(img)
    return np.real(nfft.ifft2(F * H))

# ---------------------------
# Derivatives & geometry
# ---------------------------
def grad(f: np.ndarray, dx: float) -> Tuple[np.ndarray, np.ndarray]:
    """Centered finite differences with periodic boundary (good for rings)."""
    gy, gx = np.gradient(f, dx, edge_order=2)
    return gx, gy  # return in (x, y) order for consistency

def laplacian(f: np.ndarray, dx: float) -> np.ndarray:
    gx, gy = grad(f, dx)
    gxx, _  = grad(gx, dx)
    _, gyy  = grad(gy, dx)
    return gxx + gyy

def curlz(Jx: np.ndarray, Jy: np.ndarray, dx: float) -> np.ndarray:
    """2D scalar curl: dJ_y/dx - dJ_x/dy."""
    dJy_dx, _ = grad(Jy, dx)
    _, dJx_dy = grad(Jx, dx)
    return dJy_dx - dJx_dy

# ---------------------------
# Phase tools
# ---------------------------
def unwrap_theta_path(theta_path: np.ndarray) -> np.ndarray:
    return np.unwrap(theta_path)

def sample_ring_values(field: np.ndarray, cx: float, cy: float, r: float, k: int = 2048) -> np.ndarray:
    """Sample 'field' along a ring of radius r centered at (cx, cy)."""
    n = field.shape[0]
    t = np.linspace(0.0, 2.0 * np.pi, k, endpoint=False)
    x = cx + r * np.cos(t)
    y = cy + r * np.sin(t)
    # periodic indexing
    xi = np.mod(np.round(x).astype(int), n)
    yi = np.mod(np.round(y).astype(int), n)
    return field[yi, xi]

def circulation_quanta(theta: np.ndarray, center: Tuple[float, float], r: float, k: int = 4096) -> float:
    """Return winding number n = (1/2π) ∮ ∇θ ⋅ dl, estimated by unwrap of θ along the ring."""
    cx, cy = center
    th = sample_ring_values(theta, cx, cy, r, k=k)
    th_u = unwrap_theta_path(th)
    total_phase = th_u[-1] - th_u[0]
    return float(total_phase / (2.0 * np.pi))

# ---------------------------
# Topological defects (plaquette winding)
# ---------------------------
def count_phase_defects(theta: np.ndarray) -> int:
    """
    Count 2π windings over 2x2 plaquettes (simple, robust).
    Returns signed count (sum of local windings).
    """
    t = np.pad(theta, ((0,1),(0,1)), mode='wrap')
    # plaquette corners
    a = t[:-1, :-1]
    b = t[:-1, 1:]
    c = t[1:, 1:]
    d = t[1:, :-1]
    # accumulate wrapped differences around the loop
    def wrap_diff(x, y):  # principal value in (-π, π]
        d = x - y
        return (d + np.pi) % (2*np.pi) - np.pi
    s = wrap_diff(b, a) + wrap_diff(c, b) + wrap_diff(d, c) + wrap_diff(a, d)
    # Each ~±2π is a defect
    return int(np.round(np.sum(s / (2.0 * np.pi))))

# ---------------------------
# Invariants & diagnostics
# ---------------------------
@dataclass
class Invariants:
    H_coh: float          # coherence entropy proxy
    kappa_var: float      # curvature variance (χ)
    Phi_E: float          # energy-flux proxy (|A^2 ∇θ|^2 mean)
    rho_phi: float        # defect density proxy (|defects| / area)
    R_global: float       # global order: mean(A)/std(A+eps)
    L_metric: float       # metric floor (p10 of 1/|∇χ|; clipped)
    L_geom: float         # combined floor (here == L_metric; dyn term optional)
    FVR: float            # floor violation rate proxy

def p10(x: np.ndarray) -> float:
    return float(np.percentile(x, 10.0))

def invariants(A: np.ndarray, theta: np.ndarray, chi: np.ndarray, dx: float) -> Invariants:
    eps = 1e-8
    # entropy-like measure on normalized amplitude
    Apos = np.clip(A, 0.0, None)
    w = Apos / (np.mean(Apos) + eps)
    H = float(np.mean(np.log1p(w**2)))

    # curvature of chi
    kappa = laplacian(chi, dx)
    kappa_var = float(np.var(kappa))

    # transport-aware current & energy flux proxy
    gx, gy = grad(theta, dx)
    Jx = (Apos**2) * gx
    Jy = (Apos**2) * gy
    Phi_E = float(np.mean(Jx*Jx + Jy*Jy))

    # topological defect density
    n_def = abs(count_phase_defects(theta))
    rho_phi = n_def / (A.size + 0.0)

    # global order
    Astd = float(np.std(Apos) + eps)
    Rg = float(np.mean(Apos) / Astd)

    # metric floor (local length scale from |∇χ|)
    cx, cy = grad(chi, dx)
    gmag = np.sqrt(cx*cx + cy*cy) + eps
    L_local = 1.0 / gmag
    L_metric = max(1e-3, p10(L_local))
    L_geom  = L_metric  # hook: could combine with a dyn term

    # floor violation proxy (where |∇θ| too large relative to floor)
    gth = np.sqrt(gx*gx + gy*gy)
    viol = gth > (1.0 / (L_geom + eps))
    FVR = float(np.mean(viol))

    return Invariants(H, kappa_var, Phi_E, rho_phi, Rg, L_metric, L_geom, FVR)

# ---------------------------
# Visualization
# ---------------------------
def robust_limits(arr: np.ndarray, lo: float = 2.0, hi: float = 98.0) -> Tuple[float, float]:
    vmin = float(np.percentile(arr, lo))
    vmax = float(np.percentile(arr, hi))
    if np.isclose(vmin, vmax):
        span = max(1e-6, np.ptp(arr))  # per project note: np.ptp(array)
        vmin, vmax = float(np.min(arr)), float(np.min(arr) + span)
    return vmin, vmax

def panel(A: np.ndarray, theta: np.ndarray, Jx: np.ndarray, Jy: np.ndarray,
          Bz: np.ndarray, inv: Invariants, title: str = "") -> None:
    n = A.shape[0]
    fig, axes = plt.subplots(2, 3, figsize=(13, 8), constrained_layout=True)
    axA, axT, axB, axJ, axH, axM = axes.ravel()

    vA = robust_limits(A)
    vB = robust_limits(Bz)

    im0 = axA.imshow(A, cmap="viridis", vmin=vA[0], vmax=vA[1])
    axA.set_title("Amplitude A")
    fig.colorbar(im0, ax=axA, fraction=0.046, pad=0.04)

    im1 = axT.imshow(np.mod(theta + np.pi, 2*np.pi) - np.pi, cmap="twilight", vmin=-np.pi, vmax=np.pi)
    axT.set_title("Phase θ (wrapped)")
    fig.colorbar(im1, ax=axT, fraction=0.046, pad=0.04)

    im2 = axB.imshow(Bz, cmap="coolwarm", vmin=vB[0], vmax=vB[1])
    axB.set_title("Curl proxy Bz = ∂x Jy − ∂y Jx")
    fig.colorbar(im2, ax=axB, fraction=0.046, pad=0.04)

    # Quiver (downsample for clarity)
    step = max(1, n // 24)
    y, x = np.mgrid[0:n:step, 0:n:step]
    axJ.quiver(x, y, Jx[::step, ::step], Jy[::step, ::step], pivot="mid", scale=50)
    axJ.set_xlim(0, n-1); axJ.set_ylim(n-1, 0)
    axJ.set_title("Current J = A²∇θ (quiver)")

    # Invariants text block
    txt = (
        f"H_coh: {inv.H_coh:.4f}\n"
        f"κ_var: {inv.kappa_var:.4e}\n"
        f"Φ_E:   {inv.Phi_E:.4e}\n"
        f"ρ_φ:   {inv.rho_phi:.4e}\n"
        f"R:     {inv.R_global:.3f}\n"
        f"L_metric: {inv.L_metric:.3f}\n"
        f"L_geom:  {inv.L_geom:.3f}\n"
        f"FVR:     {inv.FVR:.3f}"
    )
    axH.axis("off")
    axH.text(0.02, 0.98, txt, va="top", ha="left", family="monospace", fontsize=11)

    # Histogram of Bz for shape insight
    axM.hist(Bz.ravel(), bins=64)
    axM.set_title("Bz distribution")
    axM.set_xlabel("Bz"); axM.set_ylabel("count")

    if title:
        fig.suptitle(title, fontsize=14)
    plt.show()

# ---------------------------
# Convenience: build J & Bz, run diagnostics
# ---------------------------
def diagnostics(A: np.ndarray, theta: np.ndarray, chi: np.ndarray, dx: float,
                title: str = "") -> Dict[str, float]:
    gx, gy = grad(theta, dx)
    Jx = (A**2) * gx
    Jy = (A**2) * gy
    Bz = curlz(Jx, Jy, dx)
    inv = invariants(A, theta, chi, dx)
    panel(A, theta, Jx, Jy, Bz, inv, title=title)
    return {
        "H_coh": inv.H_coh,
        "kappa_var": inv.kappa_var,
        "Phi_E": inv.Phi_E,
        "rho_phi": inv.rho_phi,
        "R": inv.R_global,
        "L_metric": inv.L_metric,
        "L_geom": inv.L_geom,
        "FVR": inv.FVR,
    }
```

```python
# Cell 1 — Experiment 1: Single Loop Form (quark analog)
# Static characterization of a single-winding loop primitive.

# --- Grid & coordinates
G = Grid(n=256, dx=1.0)
X, Y = G.coords()
R = np.hypot(X, Y)

# --- Loop seed parameters
m_wind = +1                 # topological winding
r0, w = 32.0, 6.0           # ring radius and width (pixels)
A_peak = 1.0                # amplitude peak
noise_phase = 0.03          # small phase noise to avoid degenerate gradients

# --- Fields
# Amplitude: Gaussian annulus
A = A_peak * np.exp(-((R - r0) ** 2) / (2.0 * (w ** 2)))

# Phase: m * atan2 with tiny noise, then wrap to (-π, π]
theta = m_wind * np.arctan2(Y, X) + RNG.normal(scale=noise_phase, size=R.shape)
theta = (theta + np.pi) % (2.0 * np.pi) - np.pi

# Medium/metric χ: smoothed version of amplitude (proxy for local stiffness)
chi = gaussian_blur_fft(A, sigma=3.0)

# --- Diagnostics
stats = diagnostics(A, theta, chi, dx=G.dx, title="Exp 1 — Single Loop Form (m=+1): static characterization")

# --- Circulation audit across radii
cx = cy = 0.5 * (G.n - 1)
radii = np.array([12, 20, 28, 36, 44, 52, 64], dtype=float)
quanta = []
for r in radii:
    n = circulation_quanta(theta, center=(cx, cy), r=r, k=4096)
    quanta.append(n)

# Friendly printout
print("Circulation audit (n ≈ integer):")
for r, n in zip(radii, quanta):
    print(f"  r={r:5.1f} px  ->  n = {n: .3f}")

# --- Quick assertions for sanity (non-fatal if they fail; just informative)
approx_one_inside = any(abs(n - 1.0) < 0.1 for (r, n) in zip(radii, quanta) if r < r0 + 1.5*w)
if not approx_one_inside:
    print("[Note] Circulation not ~1 near the ring; consider slightly narrower ring or lower phase noise.")

# Store for downstream experiments (allowed by spec)
A_exp1, theta_exp1, chi_exp1 = A.copy(), theta.copy(), chi.copy()
```

```python
# Cell 2 — Experiment 2a: Two-Loop Interaction (like–like, m1=+1, m2=+1)
# Build two single-winding loops, combine as complex fields ψ = A e^{iθ}, and analyze interaction.

# --- Grid (reuse Cell 1 size for continuity)
G2 = G  # downstream cells can rely on this
X, Y = G2.coords()
R = np.hypot(X, Y)

# --- Parameters
m1, m2 = +1, +1
r0, w = 28.0, 6.5          # ring radius & width (slightly tighter than Exp 1)
sep = 60.0                 # center-to-center separation (pixels)
A_peak = 1.0
phase_noise = 0.02

# Centers along x-axis
cx = cy = 0.5 * (G2.n - 1)
c1 = (cx - sep/2.0, cy)
c2 = (cx + sep/2.0, cy)

# --- Helper: annular amplitude and phase from a given center & winding
def ring_seed(X, Y, center, r0, w, m, A_peak=1.0, noise=0.0):
    x0, y0 = center
    Rloc = np.hypot(X - x0, Y - y0)
    A = A_peak * np.exp(-((Rloc - r0) ** 2) / (2.0 * (w ** 2)))
    theta = m * np.arctan2(Y - y0, X - x0)
    if noise > 0:
        theta = theta + RNG.normal(scale=noise, size=theta.shape)
    # wrap to (-π, π]
    theta = (theta + np.pi) % (2.0 * np.pi) - np.pi
    return A, theta

# --- Build individual loops
A1, th1 = ring_seed(X, Y, c1, r0, w, m1, A_peak=A_peak, noise=phase_noise)
A2, th2 = ring_seed(X, Y, c2, r0, w, m2, A_peak=A_peak, noise=phase_noise)

# --- Complex superposition and extraction of combined fields
psi1 = A1 * np.exp(1j * th1)
psi2 = A2 * np.exp(1j * th2)
psi  = psi1 + psi2

A_2ll = np.abs(psi)
theta_2ll = np.angle(psi)
theta_2ll = (theta_2ll + np.pi) % (2.0 * np.pi) - np.pi  # ensure (-π, π]

# Medium/metric: smoothed amplitude surrogate
chi_2ll = gaussian_blur_fft(A_2ll, sigma=3.0)

# --- Diagnostics panel
stats_2ll = diagnostics(A_2ll, theta_2ll, chi_2ll, dx=G2.dx,
                        title="Exp 2a — Two-Loop Interaction (like–like: m1=+1, m2=+1)")

# --- Interaction proxies
# 1) Midline ⟨Bz²⟩ along the axis between the two cores (repulsion -> higher edge lobes, low central band)
#    We'll recompute Bz from the current built inside diagnostics for clarity.
gx, gy = grad(theta_2ll, G2.dx)
Jx = (A_2ll**2) * gx
Jy = (A_2ll**2) * gy
Bz = curlz(Jx, Jy, G2.dx)

y_mid = int(round(cy))
x_left  = int(round(c1[0] + r0/2))
x_right = int(round(c2[0] - r0/2))
x_slice = slice(max(0, x_left), min(G2.n, x_right))
midline_Bz2_mean = float(np.mean(Bz[y_mid, x_slice] ** 2)) if x_left < x_right else float(np.mean(Bz ** 2))

# 2) Amplitude saddle between loops (repulsion -> lower saddle amplitude)
A_saddle = float(np.mean(A_2ll[y_mid-2:y_mid+3, x_slice])) if x_left < x_right else float(np.mean(A_2ll))

# --- Circulation audit around each core (small rings)
r_small = 14.0
n1 = circulation_quanta(theta_2ll, center=c1, r=r_small, k=4096)
n2 = circulation_quanta(theta_2ll, center=c2, r=r_small, k=4096)

print("Two-loop like–like diagnostics:")
print(f"  Circulation near core 1 (r={r_small:.1f}): n ≈ {n1:.3f}")
print(f"  Circulation near core 2 (r={r_small:.1f}): n ≈ {n2:.3f}")
print(f"  Midline ⟨Bz²⟩ between cores: {midline_Bz2_mean:.4e}")
print(f"  Amplitude saddle Ā between cores: {A_saddle:.4f}")

# --- Store for downstream comparisons (e.g., vs unlike–like in Exp 2b)
exp2a = {
    "A": A_2ll, "theta": theta_2ll, "chi": chi_2ll,
    "centers": (c1, c2), "params": {"m1": m1, "m2": m2, "sep": sep, "r0": r0, "w": w},
    "metrics": {
        "n1": n1, "n2": n2,
        "midline_Bz2_mean": midline_Bz2_mean,
        "A_saddle": A_saddle,
        **stats_2ll
    }
}
```

```python
# Cell 4 — Experiment 3: Triad Formation (UUD/DDU analog) — binding & stability under mild evolution
# Three single-winding loops: two of one chirality, one of the opposite.
# Mild deterministic evolution: diffusion + curl-driven rotation feedback + amplitude localization by |∇θ|².
# Reuses Cell 0 utilities.

# ---------------------------
# Grid & centers (equilateral triangle)
# ---------------------------
G4 = G
X, Y = G4.coords()
n = G4.n
cx = cy = 0.5 * (n - 1)

# Triangle geometry
sep = 56.0                # side length center-to-center (pixels)
h   = np.sqrt(3) * 0.5 * sep
cA  = (cx - 0.5*sep, cy + h/3.0)   # vertex 1
cB  = (cx + 0.5*sep, cy + h/3.0)   # vertex 2
cC  = (cx,             cy - 2*h/3.0)  # vertex 3 (top)

# ---------------------------
# Loop parameters & chirality assignment
# ---------------------------
r0, w  = 24.0, 6.0
A_peak = 1.0
phase_noise = 0.02

# UUD by default (two +1, one -1). Toggle to DDU by flipping signs.
mA, mB, mC = +1, +1, -1    # UUD
# mA, mB, mC = -1, -1, +1  # DDU (alternative)

def ring_seed(X, Y, center, r0, w, m, A_peak=1.0, noise=0.0):
    x0, y0 = center
    Rloc = np.hypot(X - x0, Y - y0)
    A = A_peak * np.exp(-((Rloc - r0) ** 2) / (2.0 * (w ** 2)))
    theta = m * np.arctan2(Y - y0, X - x0)
    if noise > 0:
        theta = theta + RNG.normal(scale=noise, size=theta.shape)
    theta = (theta + np.pi) % (2.0 * np.pi) - np.pi
    return A, theta

# Individual loops
A_A, th_A = ring_seed(X, Y, cA, r0, w, mA, A_peak=A_peak, noise=phase_noise)
A_B, th_B = ring_seed(X, Y, cB, r0, w, mB, A_peak=A_peak, noise=phase_noise)
A_C, th_C = ring_seed(X, Y, cC, r0, w, mC, A_peak=A_peak, noise=phase_noise)

# Superpose complex fields and extract (A0, θ0)
psi0 = A_A*np.exp(1j*th_A) + A_B*np.exp(1j*th_B) + A_C*np.exp(1j*th_C)
A    = np.abs(psi0)
theta = np.angle(psi0); theta = (theta + np.pi) % (2.0*np.pi) - np.pi
chi   = gaussian_blur_fft(A, sigma=2.5)

# ---------------------------
# Evolution parameters (mild, floor-aware)
# ---------------------------
T   = 320            # steps
dt  = 0.12
Da  = 0.20           # amplitude diffusion
Dth = 0.18           # phase diffusion
eta = 0.55           # curl-driven rotation coupling (–eta * Bz term)
cJ  = 0.12           # amplitude localization by |∇θ|^2
chi_sigma = 2.2      # χ refresh smoothing
FVR_clip = 0.08      # simple guard: if FVR exceeds this, soften eta a bit

# Monitoring
t_samples = [0, 40, 120, 220, T-1]
PhiE_hist, FVR_hist, R_hist = [], [], []

def evolve_step(A, theta, chi):
    # gradients & current
    gx, gy = grad(theta, G4.dx)
    Jx = (A**2) * gx
    Jy = (A**2) * gy
    Bz = curlz(Jx, Jy, G4.dx)

    # diffusion (θ) + curl-driven rotation feedback
    theta_new = theta + dt*( Dth*laplacian(theta, G4.dx) - eta * Bz )
    # wrap back to (-π, π]
    theta_new = (theta_new + np.pi) % (2.0*np.pi) - np.pi

    # amplitude diffusion + localization by |∇θ|^2
    g2 = gx*gx + gy*gy
    A_new = A + dt*( Da*laplacian(A, G4.dx) - cJ * g2 * A )
    A_new = np.clip(A_new, 0.0, None)

    # refresh χ from A (acts as a moving metric/stiffness field)
    chi_new = gaussian_blur_fft(A_new, sigma=chi_sigma)

    return A_new, theta_new, chi_new

# Initial diagnostics (t=0)
stats0 = diagnostics(A, theta, chi, dx=G4.dx, title="Exp 3 — Triad (UUD) — initial")
PhiE_hist.append(stats0["Phi_E"]); FVR_hist.append(stats0["FVR"]); R_hist.append(stats0["R"])

# ---------------------------
# Time evolution
# ---------------------------
for t in range(1, T+1):
    # tiny adaptive softening if floor is violated persistently
    inv = invariants(A, theta, chi, G4.dx)
    soft_eta = eta * (0.85 if inv.FVR > FVR_clip else 1.0)

    # one step with possibly softened eta (without side effects on global eta)
    gx, gy = grad(theta, G4.dx)
    Jx = (A**2) * gx
    Jy = (A**2) * gy
    Bz = curlz(Jx, Jy, G4.dx)

    theta = theta + dt*( Dth*laplacian(theta, G4.dx) - soft_eta * Bz )
    theta = (theta + np.pi) % (2.0*np.pi) - np.pi

    g2 = gx*gx + gy*gy
    A = A + dt*( Da*laplacian(A, G4.dx) - cJ * g2 * A )
    A = np.clip(A, 0.0, None)

    chi = gaussian_blur_fft(A, sigma=chi_sigma)

    if t in t_samples:
        st = invariants(A, theta, chi, G4.dx)
        PhiE_hist.append(st.Phi_E)
        FVR_hist.append(st.FVR)
        R_hist.append(st.R_global)


# Final diagnostics
statsF = diagnostics(A, theta, chi, dx=G4.dx, title="Exp 3 — Triad (UUD) — final")

# ---------------------------
# Circulation audits (cores + outer ring)
# ---------------------------
r_small = 12.0
nA = circulation_quanta(theta, center=cA, r=r_small, k=4096)
nB = circulation_quanta(theta, center=cB, r=r_small, k=4096)
nC = circulation_quanta(theta, center=cC, r=r_small, k=4096)
# Outer ring enclosing all three
r_outer = sep    # generous
n_outer = circulation_quanta(theta, center=(cx, cy), r=r_outer, k=4096)

# ---------------------------
# Bridge metrics (amplitude and Bz along triangle edges)
# ---------------------------
def edge_slice(p, q, width=3):
    # axis-aligned band around the midpoint line via nearest grid (simple & robust)
    x0, y0 = p; x1, y1 = q
    xm, ym = (x0 + x1)/2.0, (y0 + y1)/2.0
    xm_i, ym_i = int(round(xm)), int(round(ym))
    # form a small box around the midpoint; rotate-free proxy for a narrow band
    w = int(width)
    xs = slice(max(0, xm_i - w), min(n, xm_i + w + 1))
    ys = slice(max(0, ym_i - w), min(n, ym_i + w + 1))
    return ys, xs

pairs = [("AB", cA, cB), ("BC", cB, cC), ("CA", cC, cA)]
bridge = {}
# Recompute Bz once
gx, gy = grad(theta, G4.dx)
Jx = (A**2) * gx
Jy = (A**2) * gy
Bz = curlz(Jx, Jy, G4.dx)

for name, p, q in pairs:
    ys, xs = edge_slice(p, q, width=3)
    bridge[name] = {
        "A_mean": float(np.mean(A[ys, xs])),
        "Bz2_mean": float(np.mean(Bz[ys, xs]**2)),
    }

# ---------------------------
# Print summary
# ---------------------------
print("Triad (UUD) — circulation quanta:")
print(f"  n_A @ r={r_small:.1f}: {nA:+.3f}  (center {cA})")
print(f"  n_B @ r={r_small:.1f}: {nB:+.3f}  (center {cB})")
print(f"  n_C @ r={r_small:.1f}: {nC:+.3f}  (center {cC})")
print(f"  n_outer @ r={r_outer:.1f}: {n_outer:+.3f}  (encloses all)")

print("\nBridge metrics (mean over narrow mid-bands):")
for k,v in bridge.items():
    print(f"  {k}:  Ā={v['A_mean']:.4f}   ⟨Bz²⟩={v['Bz2_mean']:.4e}")

print("\nInvariant snapshots (Φ_E, FVR, R) at t =", [0]+t_samples)
print("  Φ_E:", [f"{x:.3e}" for x in PhiE_hist])
print("  FVR:", [f"{x:.3f}" for x in FVR_hist])
print("  R  :", [f"{x:.3f}" for x in R_hist])

# ---------------------------
# Lightweight trend plots
# ---------------------------
fig, ax = plt.subplots(1, 3, figsize=(12, 3.2), constrained_layout=True)
ax[0].plot(PhiE_hist, marker='o'); ax[0].set_title("Φ_E snapshots"); ax[0].set_xlabel("sample #")
ax[1].plot(FVR_hist,  marker='o'); ax[1].set_title("FVR snapshots"); ax[1].set_xlabel("sample #")
ax[2].plot(R_hist,    marker='o'); ax[2].set_title("R snapshots");   ax[2].set_xlabel("sample #")
plt.show()

# Store for downstream (e.g., parameter sweeps, DDU variant)
exp3 = {
    "final": {"A": A, "theta": theta, "chi": chi},
    "centers": {"A": cA, "B": cB, "C": cC},
    "params": {"sep": sep, "r0": r0, "w": w, "mA": mA, "mB": mB, "mC": mC,
               "Da": Da, "Dth": Dth, "eta": eta, "cJ": cJ, "chi_sigma": chi_sigma,
               "T": T, "dt": dt},
    "metrics": {
        "nA": nA, "nB": nB, "nC": nC, "n_outer": n_outer,
        "bridge": bridge,
        "stats0": stats0, "statsF": statsF,
        "PhiE_hist": PhiE_hist, "FVR_hist": FVR_hist, "R_hist": R_hist
    }
}
```

```python
# Cell U1 — Utilities patch: masked floor + masked diagnostics (non-breaking)
# Adds support-aware invariants/diagnostics that compute the geometry floor only where A is present.

from dataclasses import dataclass
import numpy as np

@dataclass
class InvariantsMasked:
    H_coh: float
    kappa_var: float
    Phi_E: float
    rho_phi: float
    R_global: float
    L_metric: float
    L_geom: float
    FVR: float

def support_mask(A: np.ndarray, frac: float = 0.03) -> np.ndarray:
    """Binary mask over support of the field A, thresholded at 'frac' of A.max()."""
    thr = float(frac * (np.max(A) if np.max(A) > 0 else 1.0))
    return (A > thr)

def invariants_masked(A: np.ndarray, theta: np.ndarray, chi: np.ndarray, dx: float, mask_frac: float = 0.03) -> InvariantsMasked:
    eps = 1e-8
    M = support_mask(A, frac=mask_frac)

    # Coherence entropy proxy on support
    Apos = np.clip(A, 0.0, None)
    w = Apos / (np.mean(Apos[M]) + eps)
    H = float(np.mean(np.log1p(w[M]**2)))

    # Curvature variance of chi (global; small cost)
    kappa = laplacian(chi, dx)
    kappa_var = float(np.var(kappa))

    # Transport-aware current & energy proxy
    gx, gy = grad(theta, dx)
    Jx = (Apos**2) * gx
    Jy = (Apos**2) * gy
    Phi_E = float(np.mean((Jx[Jx==Jx])**2 + (Jy[Jy==Jy])**2))  # numeric safety

    # Topological defect density per area (use full for density)
    n_def = abs(count_phase_defects(theta))
    rho_phi = n_def / (A.size + 0.0)

    # Global order on support
    Rg = float(np.mean(Apos[M]) / (np.std(Apos[M]) + eps))

    # Masked geometry floor from |∇χ|
    cx, cy = grad(chi, dx)
    gmag = np.sqrt(cx*cx + cy*cy) + eps
    gM = gmag[M]
    if gM.size == 0:
        L_metric = 1e3
    else:
        L_local = 1.0 / gM
        L_metric = max(1e-3, float(np.percentile(L_local, 10.0)))
    L_geom = L_metric

    # Floor violation rate measured only on support
    gth = np.sqrt(gx*gx + gy*gy)
    viol = gth[M] > (1.0 / (L_geom + eps))
    FVR = float(np.mean(viol)) if viol.size else 0.0

    return InvariantsMasked(H, kappa_var, Phi_E, rho_phi, Rg, L_metric, L_geom, FVR)

def diagnostics_masked(A: np.ndarray, theta: np.ndarray, chi: np.ndarray, dx: float,
                       title: str = "", mask_frac: float = 0.03):
    inv = invariants_masked(A, theta, chi, dx, mask_frac=mask_frac)
    # Reuse your existing panel (expects 'Invariants' dataclass shape); build a shim
    inv_shim = Invariants(inv.H_coh, inv.kappa_var, inv.Phi_E, inv.rho_phi,
                          inv.R_global, inv.L_metric, inv.L_geom, inv.FVR)
    gx, gy = grad(theta, dx)
    Jx = (A**2) * gx
    Jy = (A**2) * gy
    Bz = curlz(Jx, Jy, dx)
    panel(A, theta, Jx, Jy, Bz, inv_shim, title=title)
    return {
        "H_coh": inv.H_coh, "kappa_var": inv.kappa_var, "Phi_E": inv.Phi_E,
        "rho_phi": inv.rho_phi, "R": inv.R_global, "L_metric": inv.L_metric,
        "L_geom": inv.L_geom, "FVR": inv.FVR
    }
```

```python
# Cell U2 — Center-tracking utilities
# Finds updated loop centers from amplitude peaks and rotated midline metrics.

import numpy as np
from scipy.ndimage import gaussian_filter

def track_centers(A: np.ndarray,
                  centers: list[tuple[float, float]],
                  search_radius: float = 12.0) -> list[tuple[float, float]]:
    """
    Track each loop center by locating the local amplitude maximum within
    a small search window (±search_radius) around the prior coordinates.
    Returns updated (cx, cy) list in same order.
    """
    n = A.shape[0]
    A_smooth = gaussian_filter(A, sigma=1.2)
    updated = []
    for (cx, cy) in centers:
        x0, y0 = int(round(cx)), int(round(cy))
        x_min, x_max = max(0, x0 - int(search_radius)), min(n, x0 + int(search_radius))
        y_min, y_max = max(0, y0 - int(search_radius)), min(n, y0 + int(search_radius))
        sub = A_smooth[y_min:y_max, x_min:x_max]
        if sub.size == 0:
            updated.append((cx, cy))
            continue
        j, i = np.unravel_index(np.argmax(sub), sub.shape)
        updated.append((x_min + i, y_min + j))
    return updated


def rotated_band_metrics(A: np.ndarray, Bz: np.ndarray,
                         p1: tuple[float, float],
                         p2: tuple[float, float],
                         width: int = 3) -> dict[str, float]:
    """
    Compute average A and Bz^2 in a rotated narrow band centered on the line p1→p2.
    Uses simple rotation to align the band with the axis, samples by nearest grid.
    """
    n = A.shape[0]
    x1, y1 = p1
    x2, y2 = p2
    L = int(np.hypot(x2 - x1, y2 - y1))
    if L <= 1:
        return {"A_mean": float(np.mean(A)), "Bz2_mean": float(np.mean(Bz**2))}
    t = np.linspace(0, 1, L)
    xs = x1 + (x2 - x1) * t
    ys = y1 + (y2 - y1) * t
    # sample band
    vals_A, vals_Bz2 = [], []
    for k in range(-width, width + 1):
        xn = xs - k * (y2 - y1) / L
        yn = ys + k * (x2 - x1) / L
        xi = np.clip(np.round(xn).astype(int), 0, n - 1)
        yi = np.clip(np.round(yn).astype(int), 0, n - 1)
        vals_A.append(A[yi, xi])
        vals_Bz2.append(Bz[yi, xi] ** 2)
    A_mean = float(np.mean(vals_A))
    Bz2_mean = float(np.mean(vals_Bz2))
    return {"A_mean": A_mean, "Bz2_mean": Bz2_mean}
```

```python
# Cell U3 — Winding estimators (support-masked ring + local disk + angle window)
import numpy as np

def ring_circulation_masked(theta: np.ndarray,
                            A: np.ndarray,
                            center: tuple[float, float],
                            r: float,
                            k: int = 4096,
                            amin_frac: float = 0.05,
                            angle_skip_deg: float | None = None,
                            skip_dir: tuple[float, float] | None = None) -> float:
    """
    Robust circulation around a ring:
      - samples θ and A on a ring of radius r,
      - masks low-support points where A < amin_frac * A.max(),
      - optional angle window: skip a ±angle_skip_deg wedge centered on 'skip_dir' (x,y).
      - computes unwrapped phase change on kept segments, then rescales by angular coverage.

    Returns: n ≈ (1/2π) ∮ ∇θ · dl  (topological charge)
    """
    n = theta.shape[0]
    cx, cy = center
    t = np.linspace(0, 2*np.pi, k, endpoint=False)
    xs = cx + r*np.cos(t)
    ys = cy + r*np.sin(t)
    xi = np.mod(np.round(xs).astype(int), n)
    yi = np.mod(np.round(ys).astype(int), n)

    th = theta[yi, xi]
    amp = A[yi, xi]

    # base mask: amplitude support
    A_thr = float(amin_frac * (np.max(A) if np.max(A) > 0 else 1.0))
    keep = amp >= A_thr

    # optional angle window skip (e.g., across the bridge direction)
    if angle_skip_deg is not None and skip_dir is not None:
        vx, vy = skip_dir
        ang_dir = np.arctan2(vy, vx)  # reference angle
        ang = np.mod(t - ang_dir + np.pi, 2*np.pi) - np.pi
        keep &= np.abs(ang) > np.deg2rad(angle_skip_deg)

    # nothing to do?
    if not np.any(keep):
        return 0.0

    # unwrap along full ring, then set masked diffs to 0 and rescale by coverage
    th_u = np.unwrap(th)
    dth = np.diff(np.r_[th_u, th_u[0] + 2*np.pi])  # length k
    # mask indices where either endpoint is dropped
    k_mask = keep & np.roll(keep, -1)
    total = float(np.sum(dth[k_mask]))
    # scale to 2π coverage (assumes charge uniform around ring)
    coverage = float(np.sum(k_mask)) / k
    if coverage <= 1e-6:
        return 0.0
    total_scaled = total / max(coverage, 1e-6)
    return total_scaled / (2*np.pi)


def local_disk_winding(theta: np.ndarray,
                       center: tuple[float, float],
                       radius: int) -> float:
    """
    Count net topological charge inside a square window ~ disk radius.
    Uses the existing plaquette counter on a cropped θ patch.
    """
    n = theta.shape[0]
    cx, cy = center
    r = int(max(2, radius))
    xi0 = int(round(cx)) - r
    yi0 = int(round(cy)) - r
    xs = slice(max(0, xi0), min(n, xi0 + 2*r + 1))
    ys = slice(max(0, yi0), min(n, yi0 + 2*r + 1))
    th_sub = theta[ys, xs]
    # reuse your plaquette-based counter
    return float(count_phase_defects(th_sub))
```

```python
# Cell 9 — Triad winding audit with two wedges + auto radius (uses Exp 3b state)
import numpy as np

def ring_circ_two_wedge(theta, A, center, r_base, v1, v2,
                        k=4096, amin_frac=0.05, angle_skip_deg=25.0):
    """Masked ring winding that skips two bridge wedges (center→center vectors v1, v2)."""
    n = theta.shape[0]
    cx, cy = center
    t = np.linspace(0, 2*np.pi, k, endpoint=False)
    xs = cx + r_base*np.cos(t)
    ys = cy + r_base*np.sin(t)
    xi = np.mod(np.round(xs).astype(int), n)
    yi = np.mod(np.round(ys).astype(int), n)
    th = theta[yi, xi]
    amp = A[yi, xi]

    A_thr = float(amin_frac * (np.max(A) if np.max(A) > 0 else 1.0))
    keep = amp >= A_thr

    def apply_skip(dir_vec, keep_mask):
        vx, vy = dir_vec
        ang_dir = np.arctan2(vy, vx)
        ang = np.mod(t - ang_dir + np.pi, 2*np.pi) - np.pi
        return keep_mask & (np.abs(ang) > np.deg2rad(angle_skip_deg))

    keep = apply_skip(v1, keep)
    keep = apply_skip(v2, keep)

    if not np.any(keep): 
        return 0.0
    th_u = np.unwrap(th)
    dth = np.diff(np.r_[th_u, th_u[0] + 2*np.pi])
    k_mask = keep & np.roll(keep, -1)
    total = float(np.sum(dth[k_mask]))
    coverage = float(np.sum(k_mask)) / k
    total_scaled = total / max(coverage, 1e-6)
    return total_scaled / (2*np.pi)

def auto_radius(A, center, r0, r_span=4, r_step=0.5):
    """Pick radius at annulus amplitude peak near r0."""
    cx, cy = center
    radii = np.arange(r0 - r_span, r0 + r_span + 1e-9, r_step)
    means = []
    for r in radii:
        t = np.linspace(0, 2*np.pi, 2048, endpoint=False)
        xi = np.clip(np.round(cx + r*np.cos(t)).astype(int), 0, A.shape[0]-1)
        yi = np.clip(np.round(cy + r*np.sin(t)).astype(int), 0, A.shape[0]-1)
        means.append(np.mean(A[yi, xi]))
    idx = int(np.argmax(means))
    return float(radii[idx]), float(np.ptp(np.array(means)))  # np.ptp(array) per project note

# --- Build two-wedge directions per core from tracked centers (Exp 3b)
cA, cB, cC = centers  # from Cell 8 tracking
vAB = (cB[0]-cA[0], cB[1]-cA[1])
vAC = (cC[0]-cA[0], cC[1]-cA[1])
vBA = (cA[0]-cB[0], cA[1]-cB[1])
vBC = (cC[0]-cB[0], cC[1]-cB[1])
vCA = (cA[0]-cC[0], cA[1]-cC[1])
vCB = (cB[0]-cC[0], cB[1]-cC[1])

def audit_core_two_wedge(label, C, d1, d2, r0):
    r_star, span = auto_radius(A, C, r0, r_span=4, r_step=0.5)
    n_plain   = [circulation_quanta(theta, center=C, r=r, k=4096) for r in (r_star-1.0, r_star, r_star+1.0)]
    n_win2    = [ring_circ_two_wedge(theta, A, C, r, d1, d2, k=4096, amin_frac=0.05, angle_skip_deg=25.0)
                 for r in (r_star-1.0, r_star, r_star+1.0)]
    n_disk    = local_disk_winding(theta, center=C, radius=int(max(6, np.ceil(r0*0.3))))
    print(f"\n[{label}] center={tuple(map(lambda z: int(round(z)), C))}  r*={r_star:.2f} (annulus span ptp={span:.3f})")
    print(f"  n_ring_plain @ r*±1 : {[f'{v:+.3f}' for v in n_plain]}")
    print(f"  n_ring_win2  @ r*±1 : {[f'{v:+.3f}' for v in n_win2]}")
    print(f"  n_disk (local window): {n_disk:+.3f}")

audit_core_two_wedge("core A (+)", cA, vAB, vAC, r0)
audit_core_two_wedge("core B (+)", cB, vBA, vBC, r0)
audit_core_two_wedge("core C (−)", cC, vCA, vCB, r0)

# Outer neutrality check (enclosing all)
n_outer = circulation_quanta(theta, center=(cx, cy), r=sep+8.0, k=4096)
print(f"\nOuter ring n_outer @ r≈{sep+8.0:.1f}: {n_outer:+.3f}")
```

```python
# Cell U5 — Tiny helpers (AM/FM schedule, normalized FFT + sideband finder)
import numpy as np

def hann(n: int) -> np.ndarray:
    n = int(n); 
    return np.ones(n) if n <= 1 else 0.5 - 0.5*np.cos(2.0*np.pi*np.arange(n)/(n-1))

def norm_rfft(x: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
    """Real FFT with Hann window; returns (freq, normalized power)."""
    x = np.asarray(x, float)
    x = x - np.mean(x)
    w = hann(x.size)
    X = np.fft.rfft(x * w)
    P = (X * np.conj(X)).real
    f = np.fft.rfftfreq(x.size, d=1.0)     # cycles / frame
    return f, P / (P.max() + 1e-12)

def find_sidebands(P: np.ndarray, f: np.ndarray, f0: float, df: float, kmax: int = 5, tol: float = 2e-3):
    """
    Locate up to kmax sidebands around carrier f0 at offsets ±n*df (n ≥ 1).
    tol is frequency bin tolerance.
    Returns list of (n, f_sideband, power_norm).
    """
    out = []
    for n in range(1, kmax+1):
        for sign in (+1, -1):
            target = f0 + sign * n * df
            idx = np.where(np.abs(f - target) <= tol)[0]
            if idx.size:
                j = idx[np.argmax(P[idx])]
                out.append((n*sign, float(f[j]), float(P[j])))
    return sorted(out, key=lambda t: (abs(t[0]), t[0] < 0))


# Cell 11 — Exp 3e: Dynamic UUD triad oscillations + ΦE sidebands

# --- Recreate triad if needed (reuses v5 parameters) ---
try:
    A, theta, chi, centers
except NameError:
    Gx = G
    X, Y = Gx.coords()
    n = Gx.n
    cx = cy = 0.5 * (n - 1)
    sep = 52.0; h = np.sqrt(3) * 0.5 * sep
    cA = (cx - 0.5*sep, cy + h/3.0)
    cB = (cx + 0.5*sep, cy + h/3.0)
    cC = (cx,             cy - 2*h/3.0)
    r0, w = 26.0, 6.0
    def ring_seed(X, Y, center, r0, w, m, noise=0.02):
        x0, y0 = center
        Rloc = np.hypot(X - x0, Y - y0)
        A = np.exp(-((Rloc - r0) ** 2) / (2.0 * (w ** 2)))
        th = m * np.arctan2(Y - y0, X - x0) + RNG.normal(scale=noise, size=Rloc.shape)
        return A.astype(np.float32), ((th + np.pi) % (2.0*np.pi) - np.pi).astype(np.float32)
    A_A, th_A = ring_seed(X, Y, cA, r0, w, +1)
    A_B, th_B = ring_seed(X, Y, cB, r0, w, +1)
    A_C, th_C = ring_seed(X, Y, cC, r0, w, -1)
    psi0 = A_A*np.exp(1j*th_A) + A_B*np.exp(1j*th_B) + A_C*np.exp(1j*th_C)
    A    = np.abs(psi0).astype(np.float32)
    theta= np.angle(psi0).astype(np.float32); theta = (theta + np.pi) % (2.0*np.pi) - np.pi
    chi  = gaussian_blur_fft(A, sigma=1.6).astype(np.float32)
    centers = [cA, cB, cC]

# --- Drive schedule (FM-like): η(t) = η0 [1 + m * sin(ω t)] ---
T      = 900          # frames
dt     = 0.12
Da     = 0.18
Dth    = 0.15
eta0   = 0.42
m_mod  = 0.10         # 10% modulation depth
f_drive= 0.006        # cycles/frame (choose small vs carrier)
chi_sigma = 1.6
mask_frac = 0.03
FVR_guard = 0.14

PhiE_series = np.zeros(T, float)
FVR_series  = np.zeros(T, float)
bridge_t    = {"AB":[], "BC":[], "CA":[]}

# helper for bridges
def bridgem(A, theta, cP, cQ):
    gx, gy = grad(theta, G.dx)
    Jx = (A**2)*gx; Jy = (A**2)*gy
    Bz = curlz(Jx, Jy, G.dx)
    return rotated_band_metrics(A, Bz, cP, cQ, width=3)

# --- Run with periodic η-modulation ---
for t in range(T):
    # centers refresh & bridge logs
    if (t % 40) == 0 or t == T-1:
        centers = track_centers(A, centers, search_radius=10)
        cA, cB, cC = centers
        bridge_t["AB"].append((t, bridgem(A, theta, cA, cB)))
        bridge_t["BC"].append((t, bridgem(A, theta, cB, cC)))
        bridge_t["CA"].append((t, bridgem(A, theta, cC, cA)))

    # masked invariants & drive
    inv = invariants_masked(A, theta, chi, G.dx, mask_frac=mask_frac)
    FVR_series[t] = inv.FVR
    eta = eta0 * (1.0 + m_mod*np.sin(2.0*np.pi*f_drive*t))
    if inv.FVR > FVR_guard:
        eta *= 0.85

    # evolve one step
    gx, gy = grad(theta, G.dx)
    Jx = (A**2)*gx; Jy = (A**2)*gy
    Bz = curlz(Jx, Jy, G.dx)

    theta = theta + dt*( Dth*laplacian(theta, G.dx) - eta*Bz )
    theta = (theta + np.pi) % (2.0*np.pi) - np.pi

    g2 = gx*gx + gy*gy
    A   = A + dt*( Da*laplacian(A, G.dx) - 0.10*g2*A )
    A   = np.clip(A, 0.0, None)
    chi = gaussian_blur_fft(np.minimum(A,1.0), sigma=chi_sigma).astype(np.float32)

    # log ΦE
    inv_now = invariants_masked(A, theta, chi, G.dx, mask_frac=mask_frac)
    PhiE_series[t] = float(inv_now.Phi_E)

# --- Quick look: spectra & sidebands ---
f, Pn = norm_rfft(PhiE_series)
kcar = int(np.argmax(Pn[1:])+1) if Pn.size>1 else 0
f0   = float(f[kcar])

sb   = find_sidebands(Pn, f, f0, df=f_drive, kmax=5, tol=2e-3)
print(f"Carrier (ΦE): f0={f0:.6f} cycles/frame; peak power=1.000")
for (n, ff, pw) in sb:
    print(f" sideband n={n:+d}: f={ff:.6f}, power={pw:.3f}")

# --- Post-run panel (final) ---
_ = diagnostics_masked(A, theta, chi, dx=G.dx, title="Exp 3e — UUD triad (final under η-modulation)", mask_frac=mask_frac)
```

```python
# Patch B — Lock-in amplitude at f_drive (and a couple harmonics)
import numpy as np

def lockin_amplitude(x: np.ndarray, f0: float, n_harm: int = 1) -> dict:
    """
    Return {n: (A, I, Q)} for harmonics n=1..n_harm at frequency n*f0.
    x is the time series; sample interval is 1 frame.
    """
    t = np.arange(len(x), dtype=float)
    out = {}
    for n in range(1, n_harm+1):
        w = 2.0*np.pi*(n*f0)*t
        c = np.cos(w); s = np.sin(w)
        I = float(2.0*np.mean(x * c))   # in-phase
        Q = float(2.0*np.mean(x * s))   # quadrature
        A = float(np.hypot(I, Q))       # amplitude
        out[n] = (A, I, Q)
    return out

# detrend ΦE slightly (remove slow drift/DC so lock-in is crisp)
PhiE_d = PhiE_series - np.mean(PhiE_series)

LI = lockin_amplitude(PhiE_d, f_drive, n_harm=3)
print("Lock-in amplitudes @ n·f_drive:")
for n,(A,I,Q) in LI.items():
    print(f"  n={n}:  A={A:.4e}  (I={I:.4e}, Q={Q:.4e})")
```
